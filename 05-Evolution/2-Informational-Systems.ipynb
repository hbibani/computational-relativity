In conventional theories of evolution, adaptation is explained through natural selection, inherited variation, and interactions between organisms and their environments.

In Computational Relativity (CR), we propose that **adaptation is primarily driven by information systems within the body**, which receive information from sensory organs and use it to guide changes and decisions.

Sensory organs such as antennae, ears, eyes, and others collect information about the surrounding environment. This information is then processed by internal systems that influence how an organism moves, responds, and interacts with its environment.

We also propose that **coordinated changes across multiple organisms are made possible through information systems passed from parents to offspring during reproduction**. These inherited systems may contain similar instructions for movement, timing, and responses. In this way, organisms that are physically separate could behave in a synchronized manner because they share similar information and internal processes.

Within CR, **information-processing systems are proposed as an important cause of the order observed in biological organisms**. Although existing evolutionary theories recognize sensory processing, inherited information, and responses to the environment, CR attempts to explain how these systems could also coordinate physical transitions and biological organization.

Consider an organism as a complex robotic system operating in a hostile environment. Without sensory information, internal processing, and the ability to respond to changing conditions, it would be extremely difficult for that organism to survive. The same problem applies to biological systems: **how could an organism consistently respond to its environment, maintain its internal structure, and survive without systems that collect and process information?**

CR proposes that these information systems play a central role in biological organization, adaptation, and the coordination of behaviour between organisms.

This document examines the idea that **organs and organisms develop in response to information gathered from their environment**, with sensory systems providing the information needed to guide decisions and adaptations.

One way to investigate this is to identify **the order in which three organisms evolved and developed their specialized organs**. By establishing when these changes occurred and how the organisms interacted, we can examine whether their evolutionary order supports the proposed mechanism.

The aim is to show that **information gathered from the environment influenced how these organisms developed**, and that the order of their adaptations provides evidence for this process.

## The Alcon Blue Butterfly, Myrmica Ants, and Ichneumon Wasp

The Alcon blue butterfly (*Phengaris alcon*) has a special relationship with *Myrmica* ants. Its larvae initially feed on gentian plants before developing a chemical disguise that makes them resemble ant larvae. The ants mistake the butterfly larvae for their own, carry them into their nests, and feed and protect them.

A second organism, the ichneumon wasp (*Ichneumon eumerus*), takes advantage of this relationship. **The wasp uses its antennae to detect chemical information and locate ant nests containing butterfly larvae.** Once inside, it releases chemicals that disrupt the ants' behaviour, causing them to attack one another. This allows the wasp to reach the butterfly larvae and lay its eggs inside them.


The wasp is therefore exploiting a relationship that already exists between the butterfly and the ants.

### The Role of Information in Adaptation

These interactions show how **organisms use information gathered through their sensory organs to make decisions and respond to their environment**.

The ants use chemical signals to recognize their young. The butterfly larvae produce similar signals to gain access to the nest. The wasp then uses its own sensory organs to identify and exploit this relationship.

Within Computational Relativity (CR), we propose that **information-processing systems play a role not only in how organisms behave, but also in how their organs and biological structures develop through evolution**.

To investigate this, we need to establish the evolutionary order in which these specialized adaptations appeared. If the order can be determined, we can examine whether later adaptations developed in response to information about organisms that already existed.



### The Proposed Mechanism

CR proposes that information gathered from the environment is processed by internal biological systems, which influence behaviour, adaptation, and the development of organs.

For example, the wasp relies on sensory information to locate the butterfly larvae inside the ant colony. Without this information, it would be much harder for the wasp to locate its target and reproduce successfully.

**The central question is whether the same type of information processing that guides an organism's behaviour also plays a role in the evolutionary development of its organs.**

Existing evolutionary biology already recognizes sensory processing, inherited information, and natural selection. CR proposes an additional role for information-processing systems in organizing physical changes during evolution.

If we can establish when these specialized adaptations evolved, we can investigate whether their order supports this proposed mechanism.

### Proposed Order of Specialized Adaptations

The interaction involves three main stages:

1. **Myrmica ants:** Develop chemical recognition systems that allow them to identify and care for their larvae.
2. **Alcon blue butterfly:** Develops chemical mimicry that allows its larvae to be accepted and cared for by the ants.
3. **Ichneumon wasp:** Develops specialized sensory and chemical mechanisms that allow it to locate and exploit the butterfly larvae inside ant nests.
