# **THE STRANGENESS OF THE EXPERIMENT DERIVING THE TAYLOR SERIES INDEPENDENTLY**

---

## **INTIAL STEPS**

When this investigation first began, it arose from an experiment concerning the observed anomalous movement of vehicles. The initial result was expressed through the relation

$$
\boxed{
p_{\mathrm{obs}}=\gamma p_{\mathrm{exp}}
}
$$

where the observed probability was related to the expected probability through the factor $\gamma$.

At this stage, the investigation was entirely statistical in origin and was formulated in relation to the binomial distribution. We had not yet identified a connection with the Boltzmann distribution, nor had we determined the mathematical structure that the theory would ultimately take.

---

## **COMPUTATIONAL TERMINOLOGY THROUGH THE ANALYSIS OF THE EXPERIMENT**

The analysis of the experiment required a set of specifically computational concepts, including processing, inputs, statefulness, and relative probability. These concepts subsequently provided the foundation for the terminology developed within Computational Relativity. The terminology is important because it expresses the underlying philosophy of the framework. In the development of theoretical systems, a sufficiently coherent conceptual structure can guide the construction of the mathematics itself. If the general concepts correctly describe the relationships within the system, they can provide strong clues as to the form that the corresponding mathematical terms should take. In this sense, the computational terminology was not introduced merely as a way of naming mathematical expressions after they had been derived. Rather, it served as a conceptual framework from which plausible mathematical formulas could be derived intuitively.

## **DERIVATION OF TAYLOR SERIES**


The most important point to understand is the order in which the mathematics developed. We assumed that the observed probability was dilated relative to the expected probability by a factor $\gamma$:

$$
p_{\mathrm{obs}}
\propto
\gamma
e^{-E_{\mathrm{exp}}/(k_B T)}.
$$

Using this energy relation as the guide for the method of postulation, using the derivative of the term above $\Delta E=-k_B T\ln\gamma$ which equals $\frac{k_b T}{\gamma}$, we estimated the corresponding temperature-dilation factor as

$$
\boxed{
\gamma_d
\approx
\frac{1}{
1-\frac{k_B T}{k_e\gamma}
}
}.
$$

The next objective was to transfer the thermodynamic structure into relativistic variables. This motivated the correspondence

$$
\boxed{
\frac{k_B T}{k_e\gamma}
\approx
\frac{mv^2}{2k_e}

$$


To amalgamate certain issues we inuitively stated that the dilation factor was striated, can be decomposed into separate velocity contributions,

$$
\frac{mv^2}{mc^2}

\frac{mv_1^2}{mc^2}

\frac{mv_2^2}{mc^2}

\frac{mv_3^2}{mc^2}

$$

\frac{mv^2}{2k_{e,n}}.

and subsequently arrived, through an independent line of reasoning, at the mathematical structure associated with the Taylor-series expansion of the relativistic dilation factor. This was not obtained by beginning with the known Taylor expansion and then constructing an interpretation around it. At the time the statistical formulation was being developed, we had not used the Taylor-series representation of the dilation term as the basis of the derivation. This was mainly because we had not been made aware of the Taylor series until a portion of the structure was already derived.

Within the proposed correspondence, each multiplicity-dependent field term may be associated with a Taylor-series sector according to

Where 

$$

This raises an important question concerning the original experiment: if a statistical formulation developed independently from the observations subsequently reproduces the mathematical structure of the relativistic Taylor series, then that correspondence itself requires explanation. The detailed derivation is contained in the `Proofs` folder.

Computational Relativity then goes a step further by proposing a physical interpretation for the individual terms appearing within that series. In doing so, the terminology introduced within Computational Relativity provides a more explicit definition of quantities that, within the formulations of Einstein and Lorentz, are treated primarily in mathematical terms without specifying an underlying computational mechanism.


In particular, the framework interprets the $n$th-order contribution as corresponding to a multiparticle system rather than treating the Taylor coefficients merely as abstract mathematical terms.

Schematically,

$$
\boxed{
\text{$n$th Taylor contribution}
\quad\longrightarrow\quad
\text{$n$-dependent multiparticle contribution}.
}
$$

Under this interpretation, the state of an individual particle is not determined solely by quantities associated with that particle in isolation. Instead, its state is determined through a computational process involving the states of other particles within the relevant system, potentially including particles located at considerable distances.

Thus, the central proposal of Computational Relativity is that

$$
\boxed{
\text{particle state}
\longrightarrow
\text{computational determination involving a multiparticle system}.
}
$$

The Taylor-series structure is therefore interpreted not simply as a mathematical approximation to relativistic dilation, but as encoding successive levels of multiparticle participation in the determination of a particle's state.

## **THE \(K-e\) BRIDGE DERIVED FROM THE TERMINOLOGY AND CONCEPTUAL FRAMEWORK OF COMPUTATIONAL RELATIVITY**

The terminology introduced within Computational Relativity led to the construction of the following bridge. This bridge arose primarily from the definitions and conceptual structure developed within the Computational Relativity framework.

The first step was to use intuition, together with a guiding postulate, to assign a more concrete physical interpretation to the relevant terms. This produced the energy partition

$$
\boxed{
K-e
}
$$

which expresses the idea that the total kinetic contribution may be separated into an ordinary kinetic component and an additional energy component associated with the computational process.

Expanding this relation leads to the following expression:

$$
\boxed{
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}
\left(
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}

mv'^2

$$

This introduces a form that does not ordinarily appear in standard relativistic formulations. Within Computational Relativity, it serves as the fundamental bridge through which quantities from other physical systems can be connected to relativistic dynamics.

The significance of this expression is that the relativistic ratio

$$
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
=
\frac{v^2}{c^2}
$$

is represented explicitly as an energy ratio. This provides a common energetic language through which otherwise distinct physical descriptions may be related to the relativistic framework.
