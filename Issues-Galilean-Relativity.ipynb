# Identification of Limitations in Galilean Relativity and Lorentzian Relativistic Formulations

From the experiment that we had conduted and the formulation that was produced from it, it's formulas specifies that the true definition of $v^2/c^2$ is in fact an energy ratio, $\tfrac{1}{2}mv^2 \big/ \tfrac{1}{2}mc^2$. The issue is that within Galilean relativity, any term produced relativistically omits the underlying causes of motion — and so this term could never be correctly defined within a Galilean framework. The Boltzmann distribution, by contrast, provides a better description of it, since it captures the causal, statistical origin of the motion that Galilean relativity leaves out. This is particularly important because, within our formulation—and more generally within any scientific framework—the construction of a bridge to relativity depends on this mechanism. Many quantities cannot be transferred directly into a relativistic description because of limitations inherent in the original formulations developed by Lorentz and Einstein. This mechanism therefore provides the means by which otherwise incompatible terms can be expressed in a form that permits a consistent connection with relativity.


The quantity

$$
\frac{k_B T}{k_e\gamma}
$$

suggests that the relativistic ratio

$$
\frac{v^2}{c^2}
$$

may be interpreted more fundamentally, within this framework, as a ratio of energies.

Specifically,

$$
\boxed{
\frac{v^2}{c^2}
=
\frac{mv^2}{mc^2}
=
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
}.
$$

Because the same mass $m$ and the factor $\frac{1}{2}$ occur in both numerator and denominator, they cancel, leaving the conventional dimensionless ratio $v^2/c^2$.

Computational Relativity therefore adopts the energy-ratio form

$$
\boxed{
\frac{mv^2}{mc^2}
}
$$

or, equivalently,

$$
\boxed{
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
}
$$

as the preferred representation going forward.

The purpose of this reformulation is to distinguish between a purely kinematic description and an energetic interpretation. A velocity ratio describes the state of motion directly, whereas the energy-ratio form makes explicit the energetic quantities associated with that motion.
