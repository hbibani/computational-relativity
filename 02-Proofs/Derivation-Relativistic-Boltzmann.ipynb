## **Introduction: Common System Motion and Individual Molecular Motion**

The diagram illustrates the distinction between two components of motion within a moving molecular system: the **common system motion** and the **individual molecular motion**.


<p align="center">
  <img src="../Assets/Diagram-54.svg" width="400">
</p>



Mathematically, these two components are distinguished as

$$ \underbrace{\vec v}_{\text{Common system motion}} \qquad \underbrace{\vec u'_i}_{\text{Individual molecular motion}} $$

The blue vectors represent the common horizontal motion shared by the molecular system, whereas the orange vectors represent the independent motion of each molecule relative to the moving system.

The important visual distinction is that **the blue motion belongs to the system collectively, whereas the orange motion belongs to each particle individually.**

For a particle moving purely vertically relative to the train, the transverse velocity measured by an external observer is

$$ \boxed{u_{y,i}=\frac{u'_{y,i}}{\gamma}} $$

where

$$ \gamma=\frac{1}{\sqrt{1-\frac{mv^2}{mc^2}}} $$

As the common horizontal velocity increases, the transverse velocity measured by the external observer decreases according to the inverse Lorentz factor:

$$ \boxed{ u_{y,i}=u'_{y,i}\sqrt{1-\frac{mv^2}{mc^2}} } $$

Within Computational Relativity, we propose examining the relationship between these two components of motion through the relativistic statistical formulation

$$ \boxed{ P(E)\propto\Omega e^{-\frac{\gamma E}{mc^2}} } $$

The objective is to investigate how common system motion influences the statistical description of individual molecular motion, particularly in the transverse direction.

Throughout this section, **contraction refers to the reduction in transverse velocity measured by an external observer, rather than a reduction in transverse molecular spacing**, since standard Lorentz transformations do not contract transverse lengths.

---

Starting with

$$ \Omega e^{-\frac{\gamma\epsilon}{mc^2}} $$

Express $\Omega$ as an exponential:

$$ =e^{\ln\Omega}e^{-\frac{\gamma\epsilon}{mc^2}} $$

Combine the exponents:

$$ =e^{-\frac{\gamma\epsilon}{mc^2}+\ln\Omega} $$

Factor out $-\frac{\gamma}{mc^2}$:

$$ \boxed{ =e^{-\frac{\gamma}{mc^2} \left(\epsilon-\frac{mc^2}{\gamma}\ln\Omega\right)} } $$

Therefore, the **change in energy** is

$$ \boxed{\Delta\epsilon=-\frac{mc^2}{\gamma}\ln\Omega} $$

Substituting the inverse Lorentz factor:

$$ \boxed{ \Delta\epsilon=-mc^2\sqrt{1-\frac{mv^2}{mc^2}}\ln\Omega } $$

This is mathematically analogous to the Boltzmann energy shift:

$$ \Delta E=-k_BT\ln\gamma $$

but with the contractive relativistic energy scale $mc^2/\gamma$ replacing $k_BT$.

---

## Recovering the contraction through the statistical formulation

From the proposed relativistic statistical formulation,

$$
P(\epsilon)\propto \Omega\, e^{-\frac{\gamma\epsilon}{mc^2}}
$$

we previously obtained

$$
\boxed{\Delta\epsilon=-\frac{mc^2}{\gamma}\ln\Omega}
$$

We now distinguish between two conditions.

### Condition A: In the absence of common system motion

When $v=0$,

$$
\gamma=1
$$

Therefore,

$$
\boxed{\Delta\epsilon_0=-mc^2\ln\Omega}
$$

### Condition B: In the presence of common system motion

When the system has a common horizontal velocity $v$,

$$
\boxed{\Delta\epsilon_v=-mc^2\sqrt{1-\frac{mv^2}{mc^2}}\,\ln\Omega}
$$

Substituting the stationary energy shift into the moving energy shift:

$$
\Delta\epsilon_v=\Delta\epsilon_0\sqrt{1-\frac{mv^2}{mc^2}}
$$

Rearranging:

$$
\boxed{\frac{\Delta\epsilon_v}{\Delta\epsilon_0}=\sqrt{1-\frac{mv^2}{mc^2}}}
$$

I also added a small space (`\,`) before $\ln\Omega$ and in $\Omega\, e^{\dots}$ so the terms are easier to read. The maths is unchanged.
