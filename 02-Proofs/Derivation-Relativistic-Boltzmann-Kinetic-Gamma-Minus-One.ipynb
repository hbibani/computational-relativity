# **Derivation of the Relativistic Statistical Factor for Entropic Multiplicity — Kinetic-Energy Formulation**

**Hidden multiplicity.** Starting from the equivalent relativistic kinetic-energy formulation,

$$
K=(\gamma-1)mc^2,\qquad \gamma=\frac{1}{\sqrt{1-\dfrac{v^2}{c^2}}}.
$$

Introducing the hidden multiplicity $x=e^{\gamma-1}$ gives

$$
\gamma-1=\ln x,\qquad K=mc^2\ln(e^{\gamma-1})=mc^2\ln x.
$$

We begin with the proposed kinetic-energy–multiplicity definition within Computational Relativity:

$$
\boxed{K_\Omega=mc^2\ln\Omega}
$$

Here, $\Omega$ represents entropic multiplicity, and $K_\Omega$ represents its associated kinetic energy, excluding rest energy.

Throughout this derivation, $m>0$ and $c$ are held fixed as parameters of the characteristic energy scale $mc^2$, and multiplicity is positive. The velocity $v$ may vary as kinetic energy changes, with $0\leq v<c$. The variable $K_\Omega$ is an independent energy coordinate of this proposed model. The single-particle kinetic-energy identity motivates the model but does not by itself establish it or its probability interpretation.

---


## **1. Express multiplicity as an exponential**

Starting with

$$
K_\Omega=mc^2\ln\Omega
$$

Divide both sides by $mc^2$:

$$
\frac{K_\Omega}{mc^2}=\ln\Omega
$$

Exponentiate both sides:

$$
e^{K_\Omega/(mc^2)}=e^{\ln\Omega}
$$

Therefore,

$$
\boxed{\Omega=e^{K_\Omega/(mc^2)}}
$$

We have now expressed entropic multiplicity as an exponential function of its associated kinetic energy.

---

## **2. Derive the logarithmic multiplicity derivative**

From our initial definition,

$$
K_\Omega=mc^2\ln\Omega
$$

Rearranging,

$$
\ln\Omega=\frac{K_\Omega}{mc^2}
$$

Differentiate both sides with respect to $K_\Omega$:

$$
\frac{\partial\ln\Omega}{\partial K_\Omega}=\frac{\partial}{\partial K_\Omega}\left(\frac{K_\Omega}{mc^2}\right)
$$

Since $m$ and $c$ are constant,

$$
\frac{\partial\ln\Omega}{\partial K_\Omega}=\frac{1}{mc^2}\frac{\partial K_\Omega}{\partial K_\Omega}
$$

Using

$$
\frac{\partial K_\Omega}{\partial K_\Omega}=1
$$

we obtain

$$
\boxed{\frac{\partial\ln\Omega}{\partial K_\Omega}=\frac{1}{mc^2}}
$$

Applying the chain rule,

$$
\frac{\partial\ln\Omega}{\partial K_\Omega}=\frac{1}{\Omega}\frac{\partial\Omega}{\partial K_\Omega}
$$

Therefore,

$$
\boxed{\frac{1}{\Omega}\frac{\partial\Omega}{\partial K_\Omega}=\frac{1}{mc^2}}
$$

This establishes that the reciprocal rest-energy scale represents the relative rate of change of entropic multiplicity with respect to its associated kinetic energy.

Rearranging further,

$$
\boxed{\frac{\partial\Omega}{\partial K_\Omega}=\frac{\Omega}{mc^2}}
$$

The rate of change of multiplicity is directly proportional to its existing magnitude.

---

## **3. Examine the higher-order logarithmic derivatives**

The first logarithmic derivative is

$$
\frac{\partial\ln\Omega}{\partial K_\Omega}=\frac{1}{mc^2}
$$

Taking the second derivative,

$$
\frac{\partial^2\ln\Omega}{\partial K_\Omega^2}=\frac{\partial}{\partial K_\Omega}\left(\frac{1}{mc^2}\right)
$$

Since $m$ and $c$ are constant,

$$
\boxed{\frac{\partial^2\ln\Omega}{\partial K_\Omega^2}=0}
$$

Taking the third derivative,

$$
\frac{\partial^3\ln\Omega}{\partial K_\Omega^3}=\frac{\partial}{\partial K_\Omega}(0)
$$

Therefore,

$$
\boxed{\frac{\partial^3\ln\Omega}{\partial K_\Omega^3}=0}
$$

Consequently, all logarithmic derivatives of order two and above vanish:

$$
\boxed{\frac{\partial^n\ln\Omega}{\partial K_\Omega^n}=0,\qquad n\geq2}
$$

### **Multiplicity and logarithmic multiplicity**

$$
\Omega=e^{K_\Omega/(mc^2)}
$$

$$
\ln\Omega=\frac{K_\Omega}{mc^2}
$$

Although multiplicity increases exponentially with energy, its logarithm is strictly linear.

This is important because it means that the Taylor expansion of logarithmic multiplicity terminates exactly after the first-order term.


---

## **4. Consider the transfer of energy**

Consider a larger system with initial multiplicity-associated kinetic energy $K_{\mathrm{total}}$, which transfers an amount of energy $E$ to a smaller microscopic system. We assume the transfer is drawn from this kinetic-energy sector, with $0\leq E\leq K_{\mathrm{total}}$, and the characteristic scale $mc^2$ remains fixed.

From conservation of energy,

$$
K_{\mathrm{total}}=K_R+E
$$

Rearranging,

$$
\boxed{K_R=K_{\mathrm{total}}-E}
$$

We define the initial multiplicity of the larger system as

$$
\boxed{\Omega_0=\Omega(K_{\mathrm{total}})}
$$

Following the transfer of energy, its remaining multiplicity becomes

$$
\boxed{\Omega_R=\Omega(K_{\mathrm{total}}-E)}
$$

We now determine how the multiplicity changes as a consequence of this energy transfer, assuming the energy removed directly reduces the multiplicity-associated kinetic energy.

---

## **5. Perform the Taylor expansion**

Recall the general Taylor expansion:

$$
\begin{aligned}
f(x-h)={}&f(x)-hf'(x)\\
&+\frac{h^2}{2!}f''(x)\\
&-\frac{h^3}{3!}f'''(x)+\cdots
\end{aligned}
$$

Applying this expansion to the logarithm of entropic multiplicity around $K_\Omega=K_{\mathrm{total}}$ (all derivatives below are evaluated there),

$$
\begin{aligned}
\ln\Omega(K_{\mathrm{total}}-E)={}&\ln\Omega(K_{\mathrm{total}})\\
&-E\frac{\partial\ln\Omega}{\partial K_\Omega}\\
&+\frac{E^2}{2!}\frac{\partial^2\ln\Omega}{\partial K_\Omega^2}\\
&-\frac{E^3}{3!}\frac{\partial^3\ln\Omega}{\partial K_\Omega^3}\\
&+\cdots
\end{aligned}
$$

From our previous derivations,

$$
\frac{\partial\ln\Omega}{\partial K_\Omega}=\frac{1}{mc^2}
$$

and

$$
\frac{\partial^n\ln\Omega}{\partial K_\Omega^n}=0,\qquad n\geq2
$$

Substituting these results into the Taylor expansion,

$$
\begin{aligned}
\ln\Omega(K_{\mathrm{total}}-E)={}&\ln\Omega(K_{\mathrm{total}})\\
&-E\left(\frac{1}{mc^2}\right)\\
&+\frac{E^2}{2!}(0)\\
&-\frac{E^3}{3!}(0)\\
&+\cdots
\end{aligned}
$$

All higher-order terms vanish, leaving

$$
\boxed{\ln\Omega(K_{\mathrm{total}}-E)=\ln\Omega(K_{\mathrm{total}})-\frac{E}{mc^2}}
$$

Using the definitions

$$
\Omega_R=\Omega(K_{\mathrm{total}}-E)
$$

and

$$
\Omega_0=\Omega(K_{\mathrm{total}})
$$

we obtain

$$
\boxed{\ln\Omega_R=\ln\Omega_0-\frac{E}{mc^2}}
$$

This expression is exact under the proposed fixed-scale kinetic-energy–multiplicity definition.

---

## **6. Obtain the change in logarithmic multiplicity**

Starting with

$$
\ln\Omega_R=\ln\Omega_0-\frac{E}{mc^2}
$$

Subtract $\ln\Omega_0$ from both sides:

$$
\ln\Omega_R-\ln\Omega_0=-\frac{E}{mc^2}
$$

Define the change in logarithmic multiplicity as

$$
\Delta\ln\Omega=\ln\Omega_R-\ln\Omega_0
$$

Therefore,

$$
\boxed{\Delta\ln\Omega=-\frac{E}{mc^2}}
$$

Using the logarithmic identity

$$
\ln a-\ln b=\ln\left(\frac{a}{b}\right)
$$

we obtain

$$
\boxed{\ln\left(\frac{\Omega_R}{\Omega_0}\right)=-\frac{E}{mc^2}}
$$

Alternatively, rearranging for the energy transferred,

$$
E=-mc^2\ln\left(\frac{\Omega_R}{\Omega_0}\right)
$$

Using the reciprocal logarithmic identity,

$$
\boxed{E=mc^2\ln\left(\frac{\Omega_0}{\Omega_R}\right)}
$$

The energy transferred is therefore related logarithmically to the ratio of initial and remaining multiplicities.

---

## **7. Express the remaining multiplicity as an exponential**

Starting with

$$
\ln\Omega_R=\ln\Omega_0-\frac{E}{mc^2}
$$

Exponentiate both sides:

$$
e^{\ln\Omega_R}=e^{\ln\Omega_0-E/(mc^2)}
$$

Using the exponential identity

$$
e^{a-b}=e^ae^{-b}
$$

we obtain

$$
e^{\ln\Omega_R}=e^{\ln\Omega_0}e^{-E/(mc^2)}
$$

Since

$$
e^{\ln\Omega_R}=\Omega_R
$$

and

$$
e^{\ln\Omega_0}=\Omega_0
$$

we obtain

$$
\boxed{\Omega_R=\Omega_0e^{-E/(mc^2)}}
$$

Dividing both sides by the initial multiplicity,

$$
\boxed{\frac{\Omega_R}{\Omega_0}=e^{-E/(mc^2)}}
$$

We have therefore derived the exponential factor

$$
\boxed{e^{-E/(mc^2)}}
$$

directly from the proposed energy–multiplicity definition.

### **Relative entropic multiplicity**

$$
\frac{\Omega_R}{\Omega_0}=e^{-E/(mc^2)}
$$

Example: for energy transferred $E/(mc^2)=1.00$ (provided the reservoir has sufficient available energy),

$$
\frac{\Omega_R}{\Omega_0}=e^{-1}\approx0.367879
$$

so the percentage of multiplicity remaining is approximately $36.7879\%$.

---

## **8. Introduce the statistical interpretation**

We introduce the statistical premise that the probability of a particular microscopic state is proportional to the remaining multiplicity of the larger system:

$$
\boxed{P(E)\propto\Omega(K_{\mathrm{total}}-E)}
$$

Using the exponential representation of multiplicity,

$$
\Omega=e^{\ln\Omega}
$$

we obtain

$$
P(E)\propto e^{\ln\Omega(K_{\mathrm{total}}-E)}
$$

Substituting the Taylor expansion derived previously,

$$
P(E)\propto e^{\ln\Omega(K_{\mathrm{total}})-E/(mc^2)}
$$

Separating the exponential terms,

$$
P(E)\propto e^{\ln\Omega(K_{\mathrm{total}})}e^{-E/(mc^2)}
$$

Since

$$
e^{\ln\Omega(K_{\mathrm{total}})}=\Omega(K_{\mathrm{total}})
$$

we obtain

$$
P(E)\propto\underbrace{\Omega(K_{\mathrm{total}})}_{\text{Constant}}e^{-E/(mc^2)}
$$

Using our definition of initial multiplicity,

$$
P(E)\propto\Omega_0e^{-E/(mc^2)}
$$

Since $\Omega_0$ is independent of the energy transferred, it can be absorbed into the proportionality constant.

Therefore,

$$
\boxed{P(E)\propto e^{-E/(mc^2)}}
$$

This establishes the proposed exponential statistical weighting for an individual microscopic state.


---

## **9. Derive the partition function**

Consider a system containing a collection of accessible microscopic states with energies (each within the available reservoir energy)

$$
E_1,E_2,E_3,\ldots,E_N
$$

The statistical weight associated with each microscopic state is

$$
\boxed{w_i=e^{-E_i/(mc^2)}}
$$

To obtain normalized probabilities, introduce a normalization constant $Z$:

$$
P_i=\frac{w_i}{Z}
$$

Substituting the statistical weight,

$$
P_i=\frac{e^{-E_i/(mc^2)}}{Z}
$$

Since the sum of all probabilities must equal unity,

$$
\sum_{i=1}^{N}P_i=1
$$

Substituting,

$$
\sum_{i=1}^{N}\frac{e^{-E_i/(mc^2)}}{Z}=1
$$

Since $Z$ is independent of $i$,

$$
\frac{1}{Z}\sum_{i=1}^{N}e^{-E_i/(mc^2)}=1
$$

Multiplying both sides by $Z$,

$$
\boxed{Z=\sum_{i=1}^{N}e^{-E_i/(mc^2)}}
$$

Expanding the summation,

$$
Z=e^{-E_1/(mc^2)}+e^{-E_2/(mc^2)}+e^{-E_3/(mc^2)}+\cdots+e^{-E_N/(mc^2)}
$$

Therefore, the complete normalized statistical distribution becomes

$$
\boxed{P_i=\frac{e^{-E_i/(mc^2)}}{\displaystyle\sum_{j=1}^{N}e^{-E_j/(mc^2)}}}
$$

For an infinite collection of accessible microscopic states, the same expression applies provided that the partition function converges.

## **11. Conclusion**

Beginning with the proposed definition of entropic multiplicity,

$$
K_\Omega=mc^2\ln\Omega
$$

we have established an exact mathematical relationship between the energy transferred from a system and the resulting change in its multiplicity:

$$
\boxed{\frac{\Omega_R}{\Omega_0}=e^{-E/(mc^2)}}
$$

The exponential dependence emerges from the constant logarithmic derivative of multiplicity with respect to energy. Because this derivative is constant, all higher-order terms in the Taylor expansion vanish, yielding an exact expression rather than an approximation.

By introducing the additional statistical premise that microscopic state probabilities are proportional to the remaining multiplicity, we obtain the proposed statistical distribution:

$$
\boxed{P_i=\frac{1}{Z}e^{-E_i/(mc^2)},\qquad Z=\sum_j e^{-E_j/(mc^2)}}
$$

Because the characteristic scale $mc^2$ is independent of velocity, changes in velocity do not introduce additional logarithmic derivatives when rest mass is fixed. At $v=0$, $\gamma-1=0$, so the hidden multiplicity is $x=1$ and the kinetic energy is zero.

The multiplicity ratio is an exact mathematical consequence of the initial definition at fixed rest mass. Its application as a physical probability distribution remains conditional on the statistical premise and the assumed relationship between energy transfer and entropic multiplicity.
