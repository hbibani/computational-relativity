## Derivation

Start with the Boltzmann-form probability (up to normalization):

$$
p_{\text{exp}} \propto e^{-E_{\text{exp}}/(k_B T)}.
$$


Assume the observed probability is dilated by a factor $\gamma$:

$$
p_{\text{obs}} = \gamma\, p_{\text{exp}}.
$$




Substitute the Boltzmann form into the dilation:

$$
p_{\text{obs}} \propto \gamma\, e^{-E_{\text{exp}}/(k_B T)}.
$$

Rewrite the multiplicative factor $\gamma$ as an exponential:

$$
\gamma = e^{\ln \gamma}.
$$

Combine exponentials:

$$
p_{\text{obs}} \propto e^{\ln\gamma}\, e^{-E_{\text{exp}}/(k_B T)}
= e^{\ln\gamma - E_{\text{exp}}/(k_B T)}.
$$

Now express $p_{\text{obs}}$ in canonical Boltzmann form with an *effective* energy $E_{\text{eff}}$:

$$
p_{\text{obs}} \propto e^{-E_{\text{eff}}/(k_B T)}.
$$

Match the exponents:

$$
\ln\gamma - \frac{E_{\text{exp}}}{k_B T} = -\frac{E_{\text{eff}}}{k_B T}.
$$

Multiply both sides by $-k_B T$:

$$
-k_B T \ln\gamma + E_{\text{exp}} = E_{\text{eff}}.
$$

Define the energy shift $\Delta E$ by $E_{\text{eff}} = E_{\text{exp}} + \Delta E$. Then:


$$
E_{\text{exp}} + \Delta E = E_{\text{exp}} - k_B T \ln\gamma
\Rightarrow\quad \boxed{\Delta E} = \boxed{-k_B T \ln\gamma}.
$$

### The Multiplicity Term in the Dilation Factor

This section identifies the multiplicity contained within the dilation factor, explains what it describes, and shows how its **form** determines the time required for a system to reach a state.

---

**1\. Origin of the multiplicity term**

The multiplicity term was first identified in the Taylor series derivation of the dilation factor. That derivation was carried out before the term's significance was recognised. The term is the relativistic analogue of the thermodynamic multiplicity $\Omega$, but it is distinct from it. The Boltzmann distribution therefore serves as a reference system: it provides the basis for interpreting how the two multiplicities act within the computational framework of Computational Relativity.

---

**2\. The multiplicity within the dilation factor**

The dilation factor

$$
\gamma = \frac{1}{\sqrt{1-\frac{mv^2}{mc^2}}}
$$

has the Taylor series expansion

$$
\gamma = \sum_{n=0}^{\infty} \frac{\binom{2n}{n}}{4^n}\left(\frac{mv^2}{mc^2}\right)^n.
$$

The coefficient of each term is exactly the thermodynamic multiplicity $\binom{q+N-1}{q}$, evaluated with $N=\tfrac{1}{2}$ and $q=n$:

$$
\binom{q+N-1}{q}\bigg|_{N=\frac{1}{2},\,q=n} = \binom{n-\frac{1}{2}}{n} = \frac{\binom{2n}{n}}{4^n}.
$$

The dilation factor is therefore a sum of multiplicities, one for each order $n$.

---

**3\. What the multiplicity describes**

Because the dilation factor describes a single moving particle, its multiplicity does not count configurations of many particles, as $\Omega$ does. Instead, it describes how the **internal constituents** of that one particle are partitioned.

Writing the multiplicity at order $n$ as

$$
\Omega_n = \binom{n-\frac{1}{2}}{n} = \frac{\binom{2n}{n}}{4^n},
$$

the first four orders are

| $n$ | $\binom{n-\frac{1}{2}}{n}$ | $\frac{\binom{2n}{n}}{4^n}$ | $\Omega_n$ |
|---:|---:|---:|---:|
| $1$ | $\binom{\frac{1}{2}}{1}$ | $\frac{2}{4}$ | $\frac{1}{2}$ |
| $2$ | $\binom{\frac{3}{2}}{2}$ | $\frac{6}{16}$ | $\frac{3}{8}$ |
| $3$ | $\binom{\frac{5}{2}}{3}$ | $\frac{20}{64}$ | $\frac{5}{16}$ |
| $4$ | $\binom{\frac{7}{2}}{4}$ | $\frac{70}{256}$ | $\frac{35}{128}$ |

Multiplying the Taylor series of the dilation factor by $mc^2$ decomposes the relativistic energy into contributions, one for each order:

$$
\boxed{
E
=
mc^2
+\frac{1}{2}mv^2
+\frac{3}{8}\frac{(mv^2)^2}{mc^2}
+\frac{5}{16}\frac{(mv^2)^3}{(mc^2)^2}
+\frac{35}{128}\frac{(mv^2)^4}{(mc^2)^3}
+\cdots
}
$$

We define $E_n$ as the portion of the kinetic energy held in the form associated with order $n$:

$$
E_n = \Omega_n\, mc^2\left(\frac{mv^2}{mc^2}\right)^n.
$$

The first four contributions are

$$
E_1=\frac{1}{2}mv^2,
\qquad
E_2=\frac{3}{8}\frac{(mv^2)^2}{mc^2},
\qquad
E_3=\frac{5}{16}\frac{(mv^2)^3}{(mc^2)^2},
\qquad
E_4=\frac{35}{128}\frac{(mv^2)^4}{(mc^2)^3}.
$$

The internal constituents of the particle can therefore be described as a **division of its kinetic energy** into these terms:

$$
E - mc^2 = E_1 + E_2 + E_3 + E_4 + \cdots
$$

Each multiplicity $\Omega_n$ sets the **maximum value** that can be attributed to a computational unit at that order. Since $\frac{mv^2}{mc^2}<1$, every energy term satisfies

$$
E_n < \Omega_n\, mc^2,
$$

and approaches this maximum as $v\rightarrow c$:

| $n$ | $E_n$ | Maximum, $\Omega_n\,mc^2$ |
|---:|---|---:|
| $1$ | $\frac{1}{2}mv^2$ | $\frac{1}{2}mc^2$ |
| $2$ | $\frac{3}{8}\frac{(mv^2)^2}{mc^2}$ | $\frac{3}{8}mc^2$ |
| $3$ | $\frac{5}{16}\frac{(mv^2)^3}{(mc^2)^2}$ | $\frac{5}{16}mc^2$ |
| $4$ | $\frac{35}{128}\frac{(mv^2)^4}{(mc^2)^3}$ | $\frac{35}{128}mc^2$ |

For example, at $n=3$ the multiplicity $\frac{5}{16}$ describes the maximum energy, $\frac{5}{16}mc^2$, that can be held in that form within the particle's internal state.

---

**4\. The form of the multiplicity**

To reach any state, a system must pass through a sequence of transitions between configurations. As shown in Proposition 6, the multiplicity counts the transition sequences that lead to a given state, and its internal structure determines which of those sequences the system follows. We refer to this internal structure as the **form** of the multiplicity, and interpret it as the **computational component** of the system.

The thermodynamic and relativistic multiplicities share the same expression, $\binom{q+N-1}{q}$, but differ in form:

| | Thermodynamic $\Omega$ | Relativistic (dilation factor) |
|---|---|---|
| $N$ | whole number (number of oscillators) | $\tfrac{1}{2}$ |
| Describes | random partitioning among many particles | partitioning of one particle's internal constituents |

Because the form determines the transitions, replacing one form with another would cause the particles to follow different transitions and drive the system towards a different type of state.

It follows that two dilation terms of equal magnitude are not necessarily the same. Two dilation terms $D_1$ and $D_2$ may satisfy

$$
|D_1| = |D_2|
$$

while still being distinct,

$$
D_1 \neq D_2,
$$

because the form of each term, and therefore the transitions that produced it, may differ.

The expression

$$
\left(\frac{mv^2}{mc^2}\right)^n
$$

can therefore be used to represent different classes of computational systems according to the organization and interpretation of their states. For example, one may define

$$
\boxed{
\left(\frac{mv^2}{mc^2}\right)^n
\text{ as a system of entropy}
}
$$

while another may be defined as

$$
\boxed{
\left(\frac{mv^2}{mc^2}\right)^n
\text{ as a system of biology}.
}
$$

The distinction therefore arises from the **computational form** imposed upon the same underlying relativistic quantity. In this sense, computational relativity provides a framework in which \(v^2/c^2\) can describe not merely a limit on energy, but the computational structure through which different physical systems are organized.

---

**5\. Multiplicity and the time to reach a state**

Without a multiplicity of the corresponding form, the transitions required to realise a particular state could not occur, and the time required to reach that state would approach infinity. Multiplicity therefore reduces the time required to produce a biological state.

In Computational Relativity, this reduction in time is represented as a reduction in dilation. In the limiting case of no dilation,

$$
\sqrt{1-\frac{mv^2}{mc^2}} = 1.
$$

$$
\boxed{
\text{Multiplicity of the corresponding form}\rightarrow
\text{finite time to reach the state}
}
$$

$$
\boxed{
\text{No multiplicity}\rightarrow
t_{\mathrm{eq}}\rightarrow\infty
}
$$
