## Proposition 9 – Distinguishing Universal and Sporadic Computational Fields, and Their Relationship to Evolution

All physical fields can be understood as computational fields. This includes classical fields such as the electric field as well as informational fields, as evidenced by the observed number-plate phenomena.

Within this framework, computational fields fall into two broad categories:

1. **Universal fields.** These fields are persistent, stable, and reproducible. Their effects are consistent across time and objects, making them invariant features of the physical or informational environment.

2. **Diminished (sporadic) fields.** These fields exhibit limited strength, duration, or scope. For example, the diminished field seen with the DA-prefix is not universal in time: it can be switched on and off. An enormous number of diminished fields exist, each operating with a distinct information-processing structure, and differences in that processing cause variation in the field's informational output.

---

### Scholium 1

Diminished fields are essential for explaining how life can arise spontaneously, because their influence is neither universal nor constant across space and time. Life does not spontaneously generate in all places at all times.

### Scholium 2

Distinct diminished fields process information independently. Variations in the informational processing of a given diminished field modify the probability structure of the system it acts upon. Biological variation emerges as a consequence of these differences in the character of the field.

## Proposition 9a – Holographness and the Theory of Everything

Holographness can be a property of both universal and diminished fields.

Holographness refers to the partial nature of a field, whose influence does not extend uniformly across all objects. As an analogy, one may be able to speak to a ghost and be heard, yet the ghost can walk through the person without physical resistance. The interaction is real but incomplete: some channels of interaction are available (information exchange), while others are absent (mechanical or material coupling). The ghost occupies the same space, but the coupling between entities is selective rather than total. In such cases, the field acts selectively rather than universally.

### Demonstration 1: Diminished Fields – The DA-Prefix Number Plate Case

In the case of diminished fields, an informational field may affect only a specific subset of objects, such as vehicles with DA-prefix number plates. This selectivity makes the field's manifestation partial, with only certain objects exhibiting measurable effects.

Beyond this, the field's influence is not localized to a single object but is distributed across the system in a non-uniform, holographic manner.

### Demonstration 2: The Invariance of Light as Holographness

As a train moves, its material components acquire the train's velocity, whereas light does not. The train's motion therefore influences only certain internal elements, which implies a holographic property: a system's characteristics apply selectively rather than universally.

### Scholium 3

Holographness is a foundational property that enables the construction of a theory of everything within a computational framework. Under this view, fields may overlap and may be either diminished or universal in scope.

## Proposition 9c – Granularity and Holographness

A field's influence on a physical system is highly granular, so that only specific molecular properties may be altered. This is especially true of diminished fields, where even the position of a single molecule within a large ensemble can be affected while the rest of the system remains effectively unchanged.
