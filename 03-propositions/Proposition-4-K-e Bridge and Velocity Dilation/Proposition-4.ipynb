## Proposition 4: The \(K-e\) Energy-Allocation Principle


With appropriate guidance from the explicit terminology developed within Computational Relativity, we were able to interpret the term \(v^2/c^2\) energetically as the ratio

$$
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}.
$$

This provided a direct energetic interpretation of the relativistic quantity and, through the terminology and conceptual framework of Computational Relativity, allowed us to make an accurate conjecture about the required mathematical relationship, leading to the development of the \(K\)-\(e\) relation.

$$
\boxed{
K' = K - e
}
$$

which states that the resulting kinetic state is determined by the available kinetic energy after a portion of that energy has been allocated to another physical or computational requirement.

Within the relativistic formulation, this principle was represented as

$$
\boxed{
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}
\left(
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
\right)
mv'^2
}
$$

The important feature of this expression is that the quantity controlling the subtraction is itself an **energy ratio**:

$$
\boxed{
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
=
\frac{mv^2}{mc^2}
=
\frac{v^2}{c^2}
}
$$

The factors of \(\frac{1}{2}\) and \(m\) cancel, leaving the familiar normalized relativistic term

$$
\frac{v^2}{c^2}.
$$

Within Computational Relativity, however, this quantity is interpreted not merely as a ratio between the squared velocity \(v^2\) and the limiting squared velocity \(c^2\), but equivalently as the ratio between the kinetic-energy scale associated with the current motion and the corresponding limiting energy scale:

$$
\boxed{
\frac{v^2}{c^2}
\equiv
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}.
}
$$

It can therefore be interpreted as the **fraction of the available energetic capacity represented by the current state of motion**. This energetic interpretation provides the basis for subsequently relating the relativistic term to the \(k_e\) energy scale and to the thermodynamic expression involving \(k_B T\).

### Corollary 2: Derivation $mc^2$ using K-e


$$
\frac12 mv'^2
=
\frac{\frac12 mv^2}{1-mv^2/mc^2}.
$$

Since $v'=\gamma v$,

$$
\frac12 mv'^2=\frac12 m\gamma^2v^2.
$$

Now rewrite the right-hand side:

$$
\frac12 m\gamma^2v^2
=
\frac12 mc^2\left(\gamma^2\frac{mv^2}{mc^2}\right).
$$

From the definition of $\gamma$,

$$
\gamma^2\frac{v^2}{c^2}=\gamma^2-1.
$$


Therefore,

$$
\frac12 mv'^2=\frac12 mc^2(\gamma^2-1).
$$

Differentiate both sides:

$$
d\left(\frac12 mv'^2\right)
=
mc^2\gamma\,d\gamma.
$$

Now use the relativistic work–energy theorem:

$$
dK=F\,dx=v\,dp.
$$

Since $p=mv'$, we have

$$
d\left(\frac12 mv'^2\right)
=
\frac{p}{m}\,dp.
$$

But

$$
\frac{p}{m}=\gamma v.
$$

Consequently,

$$
d\left(\frac12 mv'^2\right)
=
\gamma v\,dp
=
\gamma\,dK.
$$

Combine this with the differentiated expression:

$$
\gamma\,dK=mc^2\gamma\,d\gamma.
$$

Cancel $\gamma$:

$$
dK=mc^2\,d\gamma.
$$

Integrate from rest, where $\gamma=1$ and $K=0$, to the final velocity:

$$
K=\int_1^\gamma mc^2\,d\gamma.
$$

Therefore,

$$
K=mc^2(\gamma-1).
$$

This is the standard relativistic kinetic energy.

Adding the rest energy $E_0=mc^2$ gives

$$
E=K+E_0
$$

and hence

$$
\boxed{E=\gamma mc^2}.
$$

### Corollary 3: Getting True Kinetic Energy in terms of mv^2

From earlier, we have:

$$ \frac12 mv'^2=\frac12 mc^2(\gamma^2-1) $$

The term $\gamma^2-1$ is a difference of squares, so it factors as:

$$ \frac12 mv'^2 = \frac12 mc^2(\gamma-1)(\gamma+1) $$

The relativistic kinetic energy is:

$$ K=mc^2(\gamma-1) $$

Substituting $K$ into the factored expression gives:

$$ \frac12 mv'^2=\frac{\gamma+1}{2}K $$

Rearranging for $K$:

$$ \boxed{ K=\frac{mv'^2}{\gamma+1}} $$

This equals:

$$ \boxed{ K=\frac{\gamma^2mv^2}{\gamma+1} }$$

Which can be re-arranged to:


$$ \boxed{ K=\frac{mv^2}{\gamma^{-1}+\gamma^{-2}} } $$



This gives a direct conversion between the dilated kinetic-energy expression $\frac12 mv'^2$ and Einstein's relativistic kinetic energy $K$.

The key physical point is this: momentum $p=mv'$ carries a single factor of $\gamma$, so squaring it introduces $\gamma^2$. Relativistic kinetic energy, however, comes from integrating the work done as momentum changes, $dK=v\,dp$, and that integration produces $\gamma-1$ rather than $\gamma^2-1$. The factor $\frac{2}{\gamma+1}$ accounts for the difference between the two, and it approaches 1 as $v \to 0$, where both expressions reduce to the classical $\frac12 mv^2$.

### Corollary 4: Getting True Kinetic Energy in Terms of $mv^2$

Define $F(x)$ as the total field contribution, represented by the sum of individual Taylor-series sectors $f_n(x)$, where $x$ is the normalized velocity ratio:

$$
x=\frac{mv^2}{mc^2}=\frac{v^2}{c^2}
$$

The field contribution is defined as:

$$
F(x)=\sum_{n=1}^{\infty}f_n(x)=1-\sqrt{1-x}
$$

Each sector contributes a proportion determined by its corresponding Taylor coefficient:

$$
f_n(x)=\frac{\binom{2n}{n}}{4^n(2n-1)}x^n
$$

Consequently, $F(x)$ represents the cumulative proportion subtracted from unity, leaving the inverse Lorentz factor:

$$
1-F(x)=\gamma^{-1}
$$

The original kinetic-energy denominator is:

$$
\gamma^{-1}+\gamma^{-2}
$$

We now have:

$$
\gamma^{-1}=1-F(x)
$$

and

$$
\gamma^{-2}=1-x
$$

Therefore:

$$
\gamma^{-1}+\gamma^{-2}=(1-F(x))+(1-x)
$$

Expanding and substituting the sector series:

$$
\gamma^{-1}+\gamma^{-2}=2-x-\sum_{n=1}^{\infty}f_n(x)
$$

Thus, relativistic kinetic energy becomes:

$$
K=\frac{mv^2}{\displaystyle 2-\frac{mv^2}{mc^2}-\sum_{n=1}^{\infty}\frac{\binom{2n}{n}}{4^n(2n-1)}\left(\frac{mv^2}{mc^2}\right)^n}
$$

Expanding the Taylor series:

$$
K=\frac{mv^2}{\displaystyle 2-\frac32x-\frac18x^2-\frac1{16}x^3-\frac5{128}x^4-\cdots}
$$

However, taking the reciprocal of the complete denominator produces a positive Taylor series:

$$
\frac{1}{\gamma^{-1}+\gamma^{-2}}=\frac{\gamma-1}{x}
$$

Expanding the Lorentz factor and dividing by $x$:

$$
\frac{1}{\gamma^{-1}+\gamma^{-2}}=\frac12+\frac38x+\frac5{16}x^2+\frac{35}{128}x^3+\frac{63}{256}x^4+\cdots
$$

Consequently, kinetic energy can alternatively be expressed as:

$$
K=mv^2\left[\frac12+\frac38\left(\frac{mv^2}{mc^2}\right)+\frac5{16}\left(\frac{mv^2}{mc^2}\right)^2+\frac{35}{128}\left(\frac{mv^2}{mc^2}\right)^3+\cdots\right]
$$

This establishes two complementary mathematical representations: the denominator contains progressively subtracted field-sector contributions, while its reciprocal produces positive relativistic kinetic-energy corrections. Within Computational Relativity, these may be interpreted as complementary aspects of the proposed field decomposition, although their physical identification with particle sectors remains hypothetical.

### Scholium 1: Getting True Kinetic Energy in Terms of $mv^2$


The Lorentz factor is:

$$
\gamma=\frac{1}{\sqrt{1-\frac{v^2}{c^2}}}
$$

Its Taylor series is:

$$
\gamma
=
1+\frac12x+\frac38x^2+\frac5{16}x^3+\frac{35}{128}x^4+\frac{63}{256}x^5+\cdots
$$

where

$$
x=\frac{mv^2}{mc^2}=\frac{v^2}{c^2}.
$$

Subtracting 1 removes the constant term:

$$
\boxed{
\gamma-1
=
\frac12x+\frac38x^2+\frac5{16}x^3+\frac{35}{128}x^4+\frac{63}{256}x^5+\cdots
}
$$

##### General Taylor-series expression

$$
\boxed{
\gamma-1
=
\sum_{n=1}^{\infty}
\frac{\binom{2n}{n}}{4^n}
\left(\frac{mv^2}{mc^2}\right)^n
}
$$

This converges for $|x|<1$, corresponding to physical speeds $v<c$.

##### Relationship to relativistic kinetic energy

Since

$$
K=(\gamma-1)mc^2,
$$

we obtain:

$$
K=mc^2\left[
\frac12\left(\frac{mv^2}{mc^2}\right)
+\frac38\left(\frac{mv^2}{mc^2}\right)^2
+\frac5{16}\left(\frac{mv^2}{mc^2}\right)^3+\cdots
\right].
$$

Simplifying:

$$
\boxed{
K=\frac12mv^2
+\frac38\frac{mv^4}{mc^2}
+\frac5{16}\frac{mv^6}{mc^4}
+\frac{35}{128}\frac{mv^8}{mc^6}
+\cdots
}
$$

The important distinction for your Computational Relativity framework:

- $\gamma-1$ has positive Taylor coefficients $1/2,3/8,5/16,\ldots$.
- $1-\gamma^{-1}$, your proposed $F(x)$, has coefficients $1/2,1/8,1/16,\ldots$.

They are different series, although both begin with $\frac12x$.

### Scholium 2: A simpler and quicker method for deriving the term associated with K-e

An alternative and considerably simpler method of deriving the relativistic kinetic-energy expression in terms of $mv^2$ is to factor out the normalized velocity ratio $x$ directly from the Taylor expansion of $\gamma-1$. This avoids the more extensive algebraic derivation involving the inverse-Lorentz-factor denominator.

Starting with:

$$
K=mc^2\left[
\frac12x+\frac38x^2+\frac5{16}x^3+\frac{35}{128}x^4+\cdots
\right]
$$

Since every term contains at least one factor of $x$, we can factor it out:

$$
K=mc^2x\left[
\frac12+\frac38x+\frac5{16}x^2+\frac{35}{128}x^3+\cdots
\right]
$$

Substitute:

$$
x=\frac{mv^2}{mc^2}
$$

Therefore:

$$
K=mc^2\frac{mv^2}{mc^2}\left[
\frac12+\frac38x+\frac5{16}x^2+\frac{35}{128}x^3+\cdots
\right]
$$

Cancelling $mc^2$ gives:

$$
K=mv^2\left[
\frac12+\frac38x+\frac5{16}x^2+\frac{35}{128}x^3+\cdots
\right]
$$

Finally, substituting the normalized velocity ratio:

$$
\boxed{
K=mv^2\left[
\frac12
+\frac38\left(\frac{mv^2}{mc^2}\right)
+\frac5{16}\left(\frac{mv^2}{mc^2}\right)^2
+\frac{35}{128}\left(\frac{mv^2}{mc^2}\right)^3
+\cdots
\right]
}
$$

This demonstrates that the alternative kinetic-energy formulation can be obtained directly from the Taylor series through a simple factorization, without requiring the more extensive intermediate derivation. The resulting expression retains $mv^2$ as the principal factor, while the remaining series accounts for the relativistic corrections.
