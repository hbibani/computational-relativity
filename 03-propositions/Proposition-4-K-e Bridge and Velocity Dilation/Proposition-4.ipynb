## Proposition 4: The \(K-e\) Energy-Allocation Principle


With appropriate guidance from the explicit terminology developed within Computational Relativity, we were able to interpret the term $v^2/c^2$ energetically as the ratio

$$
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}.
$$

This provided a direct energetic interpretation of the relativistic quantity and, through the terminology and conceptual framework of Computational Relativity, allowed us to make an accurate conjecture about the required mathematical relationship, leading to the development of the \(K\)-\(e\) relation.

$$
\boxed{
K' = K - e
}
$$

which states that the resulting kinetic state is determined by the available kinetic energy after a portion of that energy has been allocated to another physical or computational requirement.

Within the relativistic formulation, this principle was represented as

$$
\boxed{
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}
\left(
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
\right)
mv'^2
}
$$

The important feature of this expression is that the quantity controlling the subtraction is itself an **energy ratio**:

$$
\boxed{
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
=
\frac{mv^2}{mc^2}
=
\frac{v^2}{c^2}
}
$$

The factors of \(\frac{1}{2}\) and \(m\) cancel, leaving the familiar normalized relativistic term

$$
\frac{v^2}{c^2}.
$$

Within Computational Relativity, however, this quantity is interpreted not merely as a ratio between the squared velocity \(v^2\) and the limiting squared velocity \(c^2\), but equivalently as the ratio between the kinetic-energy scale associated with the current motion and the corresponding limiting energy scale:

$$
\boxed{
\frac{v^2}{c^2}
\equiv
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}.
}
$$

It can therefore be interpreted as the **fraction of the available energetic capacity represented by the current state of motion**. This energetic interpretation provides the basis for subsequently relating the relativistic term to the \(k_e\) energy scale and to the thermodynamic expression involving \(k_B T\).

### Corollary 2: A Simpler Derivation of the K–e Relation

Starting with $mv'^2=\gamma^2 mv$:

$$
\frac12 mv'^2
=\frac12 m\gamma^2v^2
=\frac12 mc^2\gamma^2 (\frac{mv^2}{mc^2})
=\frac12 mc^2(\gamma^2-1)
=\frac12 mc^2(\gamma-1)(\gamma+1).
$$

Using $K=mc^2(\gamma-1)$, we obtain:

$$
K=\frac{\gamma^2mv^2}{\gamma+1}
=\frac{mv^2}{\gamma^{-1}+\gamma^{-2}}.
$$

**Alternatively, a simpler derivation follows directly from the Taylor expansion.**

Let $x=v^2/c^2=mv^2/mc^2$. Then:

$$
K=mc^2(\gamma-1)
=mc^2\left[\frac12x+\frac38x^2+\frac5{16}x^3+\frac{35}{128}x^4+\cdots\right].
$$

Factoring out $x$ and cancelling $mc^2$:

$$
K=mv^2\left[
\frac12+\frac38x+\frac5{16}x^2+\frac{35}{128}x^3+\cdots
\right].
$$

Therefore:

$$
\boxed{
K=mv^2\left[
\frac12+\frac38\left(\frac{mv^2}{mc^2}\right)
+\frac5{16}\left(\frac{mv^2}{mc^2}\right)^2
+\cdots
\right]
=\frac{mv^2}{\gamma^{-1}+\gamma^{-2}}
}
$$

This demonstrates that the same relativistic kinetic-energy expression can be obtained directly by factoring the Taylor series, avoiding the more extensive intermediate algebra.

### Corollary 3: Deriving Relativistic Kinetic Energy from the Field–Motion Partition

The purpose of this section is to highlight the difficulty of directly interpreting the mathematical expressions produced by the K–e Bridge and assigning physical meaning to them.
In Corollary 2, the best we could achieve was an alternative formulation of the relativistic kinetic-energy equation. However, interpreting the resulting terms required us to rely on relationships already established by Einstein, rather than deriving their physical meaning independently.

In this second attempt, we achieved a more promising result. By applying the terminology and principles introduced in Computational Relativity, we were able to assign meaning to the individual mathematical contributions within the framework itself.

Furthermore, as demonstrated, the derivation can once again be expressed in just three lines of mathematics.

The significance of this result is that it demonstrates a potential pathway toward deriving the energy relationships associated with (E=mc^2) directly through Computational Relativity, without explicitly relying on Einstein's original derivation.

We begin with the relationship previously established in Proposition 3: Probability Dilation.

$$

\frac{1}{k_e^{(n-1)}}.

This implies,

$$
D_{\mathrm{motion}}+D_{\mathrm{field}}=\frac{mv^2}{mc^2}.
$$

Multiplying by \(mc^2\),

$$
mc^2D_{\mathrm{motion}}+mc^2D_{\mathrm{field}}=mv^2.

From the Taylor expansion of the Lorentz factor, we identify

$$

$$

Rearranging,

$$

Since \(K=(\gamma-1)mc^2\), we obtain

$$
K=mc^2D_{\mathrm{field}}=(\gamma-1)mc^2.
$$

Thus, the field contribution corresponds mathematically to relativistic kinetic energy when its Taylor-series coefficients are defined to reproduce \(\gamma-1\). The remaining motion contribution is the difference between \(mv^2\) and \(K\).
