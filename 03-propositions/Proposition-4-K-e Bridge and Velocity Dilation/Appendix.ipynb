## Appendix: Supporting Derivations

### A.1 Relating $d\left(\tfrac12 mv'^2\right)$ to $dK$

Starting from the momentum written in terms of proper velocity,

$$ p = mv' $$

we show that

$$ d\left(\tfrac12 mv'^2\right) = \gamma\,dK $$

**Differentiate the energy term.** Treating $m$ as constant,

$$ d\left(\tfrac12 mv'^2\right) = \tfrac12 m(2v')\,dv' = mv'\,dv' $$

**Differentiate the momentum.** Since $m$ is constant,

$$ dp = m\,dv' $$

**Substitute.** Grouping the terms as $v'(m\,dv')$ and using $m\,dv' = dp$,

$$ d\left(\tfrac12 mv'^2\right) = v'\,dp = \frac{p}{m}\,dp $$

**Introduce the Lorentz factor.** With $v' = \gamma v$,

$$ d\left(\tfrac12 mv'^2\right) = \gamma v\,dp $$

The relativistic work–energy theorem gives $dK = v\,dp$, so multiplying by $\gamma$:

$$ \gamma\,dK = \gamma v\,dp $$

Therefore,

$$ \boxed{ d\left(\tfrac12 mv'^2\right) = \gamma\,dK } $$

The factor of $\gamma$ arises because $v'$ is the proper velocity $dx/d\tau$, while $v$ is the coordinate velocity $dx/dt$, and the two are related by $v' = \gamma v$.

### A.2 The identity $\gamma^2 \dfrac{v^2}{c^2} = \gamma^2 - 1$

Starting from the square of the Lorentz factor,

$$ \gamma^2 = \frac{1}{1-\dfrac{v^2}{c^2}} $$

multiply both sides by the denominator and expand:

$$ \gamma^2\left(1-\frac{v^2}{c^2}\right) = 1 \quad\Longrightarrow\quad \gamma^2 - \gamma^2\frac{v^2}{c^2} = 1 $$

Rearranging gives

$$ \boxed{ \gamma^2\frac{v^2}{c^2} = \gamma^2 - 1 } $$

This is purely algebraic; no additional physics is involved.

### A.3 Rewriting $K$ with negative powers of $\gamma$

Starting from

$$ K = \frac{\gamma^2 mv^2}{\gamma+1} $$

divide the numerator and denominator by $\gamma^2$:

$$ K = \frac{mv^2}{\dfrac{\gamma+1}{\gamma^2}} $$

Split the denominator:

$$ \frac{\gamma+1}{\gamma^2} = \frac{\gamma}{\gamma^2} + \frac{1}{\gamma^2} = \frac{1}{\gamma} + \frac{1}{\gamma^2} = \gamma^{-1} + \gamma^{-2} $$

Therefore,

$$ \boxed{ K = \frac{mv^2}{\gamma^{-1}+\gamma^{-2}} } $$

Note that $\gamma^{-1}$ means $1/\gamma$ (not $\gamma - 1$), and $\gamma^{-2}$ means $1/\gamma^2$ (not $\gamma - 2$).

Both forms use the coordinate velocity $v$, with $\gamma$ evaluated at that same $v$:

$$ \frac{\gamma^2 mv^2}{\gamma+1} = \frac{mv^2}{\gamma^{-1}+\gamma^{-2}} $$
