# Agent Orange
### Dedicated to the Great Jeremy Bentham
---
## I. Purpose
### Proposition 1 — Purpose
A system of government cannot be rationally directed unless the purpose toward which it is intended to operate has first been established.
### Proposition 2 — Theological Purpose
The theological purpose is the proposed objective of human existence determined through theological investigation.
### Proposition 3 — Investigation of Purpose
Because the theological purpose is not immediately evident, its determination requires investigation, comparison, criticism, and the continual reassessment of the information from which it is inferred.
### Proposition 4 — Government and Purpose
Where a purpose has been established, knowledge of that purpose should inform the principles by which government determines its objectives.
### Proposition 5 — Aimless Government
A government that possesses no conception of the purpose toward which its institutions are directed is, in this respect, an aimless government.

---
# II. Utility
### Proposition 6 — Utility
The principle of utility evaluates an action according to the pleasures, pains, benefits, and injuries that follow from it.
### Proposition 7 — Evaluation of an Action
The evaluation of an action requires consideration not merely of the immediate pleasure produced by the action, but also of its immediate and remote consequences for the individual and for others.
### Proposition 8 — Scientific Information
Information discovered through theology, psychology, medicine, economics, science, or any other legitimate field of investigation may alter the calculation by which an action or policy is judged.
### Proposition 9 — Utility and Purpose
The principle of utility and the principle of purpose are not necessarily contradictory, because knowledge of purpose may itself become information incorporated into the calculation of utility.
### Proposition 10 — Deficiency of Knowledge
An error in government may arise not because the principle by which an action is judged is necessarily defective, but because the information supplied to that principle is incomplete or false.
### Proposition 11 — Human Implementation
Even a rational principle may produce defective government when its practical implementation is distorted by the psychological attributes of the people responsible for applying it.

---
# III. Common Interest
### Proposition 12 — Common Interest
Human beings possess common interests insofar as certain conditions are generally required for their preservation, security, development, and continued existence.
### Proposition 13 — Conditions of Preservation
Food, water, shelter, clothing, medicine, sanitation, education, infrastructure, energy, transportation, security, industry, markets, technology, and government are examples of conditions that may contribute to human preservation and development.
### Proposition 14 — Priority
Where some interests are necessary for preservation while others are not, the preservation of the necessary interests possesses a higher governmental priority.

---

# IV. Burden
### Proposition 15 — Burden
The conditions of the world impose burdens upon individuals that constrain their ability to live and to fulfil the purpose.
### Proposition 16 — Solution
A solution is an action, object, institution, or arrangement that removes, reduces, or prevents a burden.
### Proposition 17 — Value of a Solution
The value of a solution consists principally in its capacity to preserve, protect, improve, or enable the condition of those upon whom the burden operates.
### Proposition 18 — Provision
A provision is a solution organized or supplied for the satisfaction of a requirement.
### Proposition 19 — Governmental Provision
A governmental provision is a solution organized, maintained, or made accessible through government for the purpose of removing or reducing a burden affecting the population.
### Proposition 20 — Objective of Government
The objective of government is to identify the burdens that obstruct the common interest and to arrange provisions capable of removing or diminishing those burdens.
### Proposition 21 — Organization
Because the production and maintenance of provisions require coordinated human activity, government necessarily involves the organization of labour, knowledge, resources, and institutions.

---

# V. The Motive
### Proposition 22 — Motive
The motive is an internal force that applies pressure to the body and disposes the body toward action.
### Proposition 23 — Desire
A desire is the intellectual expression of a motive toward an imagined object or objective.
### Proposition 24 — Action
An action occurs when the attributes and capacities of the identity are employed in response to a motive.
### Proposition 25 — Desornation
**Desornation** is the process by which the attributes of the identity are employed as means toward the acquisition or realization of a desire after that desire has influenced the motive.
### Proposition 26 — Identity as Instrument
Intelligence, memory, imagination, knowledge, physical capacity, social position, and other attributes of the identity may become instruments of Desornation.
### Proposition 27 — Neutrality of Capacity
The possession of intelligence or ability does not determine the moral quality of an action, because the same capacity may be employed in the service of radically different desires.

---

# VI. Affects and Action
### Proposition 28 — Affect
An affect modifies the motive by producing a disposition toward or away from an object, person, condition, or imagined outcome.
### Proposition 29 — Affect and Reason
Reason may determine the means by which an objective is pursued without necessarily determining the objective that the individual desires to pursue.
### Proposition 30 — Irrational Direction
A rational capacity may therefore be employed efficiently in pursuit of an irrational objective whenever the motive has been directed by a destructive affect.
### Proposition 31 — Persistence
Where an affect persists, the pressure that it exerts upon the motive may repeatedly generate actions directed toward the same general objective.

---

# VII. Fragsonevy
### Proposition 32 — Fragsonevy
**Fragsonevy** is the intellectual contradiction between the objectives of two or more desires.
### Proposition 33 — Contradictory Desires
Two desires are contradictory when the realization of one desire obstructs, damages, or prevents the realization of another desire possessed by the same individual.
### Proposition 34 — Inconstation
**Inconstation** is the internal contradiction produced when opposing affects generate desires whose objectives cannot be simultaneously satisfied.
### Proposition 35 — Fragmentation of Objective
Fragsonevy produces fragmentation when the individual alternately acts according to desires directed toward incompatible objectives.
### Proposition 36 — Self-Obstruction
An individual may consequently act against an objective that the same individual strongly desires because another affect temporarily exerts the greater pressure upon the motive.
### Proposition 37 — Love and Envy
Love may produce a desire to approach, preserve, praise, or benefit another person, while envy toward the same person may simultaneously produce a desire to diminish, reject, criticize, or harm that person.
### Proposition 38 — Contradictory Behaviour
Where love and envy operate simultaneously, the actions produced by the individual may appear inconsistent even though each action corresponds to one of the competing desires acting upon the motive.

---

# VIII. Destructive Affects
### Proposition 39 — Hostility Toward the Good
A person may experience pain, depression, hostility, or resentment from the perception of something regarded as good or fortunate.
### Proposition 40 — Destructive Desire
Such an affect may produce a desire to damage, remove, diminish, or prevent the thing whose existence generated the painful idea.
### Proposition 41 — Reluctance to Benefit Another
A person may experience pain from the idea that another person will become fortunate through an action available to them, thereby producing a reluctance to perform an otherwise beneficial action.
### Proposition 42 — Pleasure in Misfortune
A person may experience pleasure or relief from the idea that another person has suffered a misfortune.
### Proposition 43 — Pleasure in Harm
Where the idea of another person's suffering itself becomes pleasurable, a desire may arise to produce or preserve the conditions from which that suffering follows.
### Proposition 44 — Attraction to Destructive Attributes
A person may admire or become attracted to attributes that are destructive to themselves or others and may consequently seek either to acquire those attributes or to associate with people who possess them.

---

# IX. Greed
### Proposition 45 — Greed
Greed is a desire to acquire or preserve wealth where the desired acquisition is pursued without sufficient regard for the interests of others.
### Proposition 46 — Fear of Loss
The anticipated loss of wealth may produce pain or anger and thereby create a motive directed toward preventing that loss.
### Proposition 47 — Retaliation for Loss
Where another person is believed to have caused a loss, the pain associated with that loss may be converted into a desire for retaliation against the person believed to be responsible.

---

# X. Indignation and Spite
### Proposition 48 — Indignation
Indignation is a hostile affect produced by the perception of an act or attribute regarded as wrongful, offensive, or objectionable.
### Proposition 49 — Perceived Injury
The idea that one has been wronged may generate hatred toward the person believed to have caused the wrong.
### Proposition 50 — Spite
Spite is the desire to cause injury to another because that person is believed to have previously caused an injury.
### Proposition 51 — Retaliatory Pleasure
A person affected by spite may experience pleasure from the knowledge that misfortune has occurred to the person believed to have wronged them.

---

# XI. Envy
### Proposition 52 — Envy
Envy is pain produced by the idea of some good, advantage, attribute, possession, or fortune belonging to another person.
### Proposition 53 — Denial Under Envy
Envy may produce a desire to deny, diminish, or reinterpret the positive attribute that generated the envy.
### Proposition 54 — Search for Defect
Envy may produce a desire to discover a defect in the person toward whom the envy is directed.
### Proposition 55 — Communication of Defect
The desire to diminish the envied person may extend into an attempt to persuade others that the envied person possesses some defect.
### Proposition 56 — Injury Under Envy
Where envy becomes sufficiently strong, the desire to diminish another person's fortune may develop into a desire to cause that person some form of harm.
### Proposition 57 — Relief Through Loss
The loss of the object, advantage, or attribute that generated envy may produce relief in the envious person.

---

# XII. Competition
### Proposition 58 — Competition
Competition occurs when an individual interprets the fortune of another as relevant to their own comparative position.
### Proposition 59 — Competitive Pain
A competitor may experience pain when another person whom they regard as a rival acquires some advantage or fortune.
### Proposition 60 — Destructive Competition
Competition becomes destructive where the individual attempts to improve their relative position principally by diminishing the fortune of the perceived competitor.
### Proposition 61 — Competitive Relief
A person affected by destructive competition may experience relief when a perceived competitor suffers a misfortune.
### Proposition 62 — Relative and Absolute Improvement
The destruction of another person's advantage does not necessarily improve one's own condition, even though it may improve one's perceived relative position.

---

# XIII. Praise, Recognition and Pride
### Proposition 63 — Pain of Recognition
A person may experience pain from openly recognizing a positive attribute possessed by another person.
### Proposition 64 — Pain of Praise
Where recognition of another person's good attribute produces pain, the person may become reluctant to praise the other even when the praise is believed to be deserved.
### Proposition 65 — Concealment of Admiration
The pain associated with praise may produce an attempt to conceal admiration through silence, minimization, criticism, or some other deceptive expression.
### Proposition 66 — Desire for Recognition
The desire to be recognized, respected, or admired may become destructive when recognition is pursued at the expense of the interests or happiness of others.
### Proposition 67 — Pride and Collective Work
An excessive desire for personal recognition may prevent an individual from cooperating effectively in a common objective where the individual cannot personally claim the success of the project.
### Proposition 68 — Hostility Toward Pride
The observation of pride in another person may itself generate hostility in those who interpret that pride as offensive or threatening.
### Proposition 69 — Denial of Quality
A person who experiences pain from another person's positive quality may attempt first to deny the existence of the quality and subsequently to persuade the other person that the quality does not exist.

---

# XIV. Failure
### Proposition 70 — Pain of Failure
The perception that one's own actions are failing may produce psychological pain.
### Proposition 71 — Displacement of Failure
Rather than correcting the cause of personal failure, an individual may attempt to produce failure in others in order to reduce the comparative significance of their own failure.
### Proposition 72 — Relief Through Shared Failure
The knowledge that others have also failed may produce relief without altering the original cause of one's own failure.

---

# XV. Safety and Consideration of Others
### Proposition 73 — Consideration of Others
The capacity to act in a common interest requires an individual to represent within their reasoning the safety, welfare, and interests of persons other than themselves.
### Proposition 74 — Failure of Consideration
Where the mind fails to adequately represent the interests of others, actions may be produced that are locally advantageous to the individual while damaging the wider system of which the individual forms a part.

---

# XVI. Psychological Influence Upon Institutions
### Proposition 75 — Institutions and Persons
An institution possesses no psychology independently of the persons whose actions produce and maintain it.
### Proposition 76 — Psychological Transmission
The psychological attributes of individuals may therefore become expressed in the policies, procedures, decisions, delays, conflicts, and failures of the institutions in which those individuals operate.
### Proposition 77 — Institutional Motive
A person occupying an institutional position may use the authority and resources attached to that position as instruments of Desornation.
### Proposition 78 — Divergence of Interest
An official may act against the interests of the organization to which they belong when an internal desire exerts greater pressure upon the motive than the desire to preserve the institutional objective.
### Proposition 79 — Internal Hostility
Hatred, envy, indignation, competition, greed, pride, fear, or despondency may produce conduct contrary to the common interest even in the absence of external instruction.
### Proposition 80 — Appearance of External Influence
Persistent conduct against an organization's interests may externally resemble the conduct that would be expected from a person acting under foreign or external influence.
### Proposition 81 — Appearance and Cause
Similarity between two patterns of behaviour does not establish similarity between their causes.
### Proposition 82 — Internal and External Causation
An action contrary to the common interest may arise either from an internally generated policy of the individual or from an externally generated objective communicated by another person or organization.
### Proposition 83 — Evidence
The distinction between internally and externally generated conduct must be determined by evidence concerning communication, coordination, incentive, intention, and behaviour rather than by appearance alone.

---

# XVII. Policy and Security
### Proposition 84 — Security Review
Where a governmental decision may materially affect the security of a population or state, the decision should be subjected to an appropriate process of investigation and review.
### Proposition 85 — Psychological Independence
The security of an institution requires that decisions intended to protect the common interest remain sufficiently independent from the destructive private affects of the individuals responsible for those decisions.
### Proposition 86 — Rational Assessment
The evaluation of a policy must distinguish the objective consequences of the policy from the private emotional states of those evaluating it.
### Proposition 87 — Institutional Safeguards
Where individual psychological defects can influence decisions of great consequence, institutions require procedures capable of detecting error, reviewing decisions, distributing authority, and correcting harmful conduct.

---

# XVIII. Preservation of the Union
### Proposition 88 — Preservation
The primary objective of a political union is the preservation of the conditions required for the continued security and functioning of that union.
### Proposition 89 — Common Security
Preservation of the union includes protection of the common interests shared among its members.
### Proposition 90 — Individual Members
Preservation of the union includes protection of the legitimate security interests of each individual member within the union.
### Proposition 91 — External Interests
A union may possess obligations extending beyond its own members where its actions materially affect the security and welfare of persons outside the union.
### Proposition 92 — Economic Interest
The preservation of the union includes the maintenance of economic conditions necessary for the continued functioning and security of its members.
### Proposition 93 — Environment
The preservation of the natural environment forms part of the preservation of the union wherever environmental conditions determine the health, resources, or continued existence of its members.
### Proposition 94 — Citizens
The security of citizens forms an essential component of the security of the political system established for their preservation.

---

# XIX. Internal and External Threats
### Proposition 95 — Internally Generated Threat
An internally generated institutional threat arises when an individual acts against the common interest according to motives produced within the individual rather than through coordination with an external organization.
### Proposition 96 — Externally Generated Threat
An externally generated institutional threat arises when a person knowingly coordinates with an external actor and acts according to an objective supplied or directed by that actor against the interests of the institution or union.
### Proposition 97 — Distinction of Threats
Internally and externally generated threats may produce similar observable consequences while differing substantially in origin.
### Proposition 98 — Classification
The proper classification of a threat therefore depends upon establishing its causal origin rather than merely observing its effects.
### Proposition 99 — Response
Institutional responses to suspected threats should be lawful, evidence-based, proportionate, reviewable, and directed toward preservation rather than retaliation.

---

# XX. Compartmentalization
### Proposition 100 — Division of Information
An organization may divide information among different persons or groups so that each component possesses only the information necessary for its particular function.
### Proposition 101 — Compartmentalization
Compartmentalization limits the quantity of information available to any single subordinate component of an organization.
### Proposition 102 — Organizational Knowledge
In a highly compartmentalized system, knowledge of the complete objective may be concentrated in a comparatively small coordinating authority.
### Proposition 103 — Operational Independence
The separation of organizational functions may increase operational independence between components.
### Proposition 104 — Accountability Risk
The same compartmentalization that protects sensitive information may reduce the ability of subordinate participants to evaluate the wider purpose, legality, or morality of the actions they are instructed to perform.
### Proposition 105 — Centralized Risk
Where knowledge and authority are concentrated within a central command, defects in the motives or objectives of that command may propagate throughout the organization without being visible to subordinate components.
### Proposition 106 — Institutional Danger
An efficient organizational structure is not necessarily a good organizational structure, because efficiency determines how successfully an objective is pursued rather than whether the objective itself is beneficial.
### Proposition 107 — Means and Ends
The moral character of an organizational method cannot be determined independently of the objective toward which that method is directed.

---

# XXI. Psychological Corruption of Collective Objectives
### Proposition 108 — Common Objective
A collective system functions toward the common interest only insofar as the actions of its members remain sufficiently aligned with the objective for which the system exists.
### Proposition 109 — Corruption of Objective
Where private destructive affects become dominant within positions of authority, institutional resources may be redirected from the production of solutions toward the production of additional burdens.
### Proposition 110 — Inversion
An institution becomes inverted when mechanisms created for preservation are systematically employed in ways that increase the burdens they were intended to remove.
### Proposition 111 — Psychological Origin of Institutional Failure
Some institutional failures may therefore be understood not only through defective laws or structures, but through the psychological attributes of the persons who operate those structures.
### Proposition 112 — Structural Amplification
An institution may amplify an individual psychological defect when the authority, resources, or reach of the institution enables one person's motive to affect a substantially larger population.
### Proposition 113 — Necessity of Restraint
The greater the power attached to an institutional position, the greater the importance of mechanisms that restrain the conversion of private destructive motives into public action.

---

# XXII. Investigation
### Proposition 114 — Psychological Investigation
The investigation of human behaviour requires the identification of the affects, desires, motives, and intellectual processes from which observable actions arise.
### Proposition 115 — Behaviour and Cause
The observation of behaviour alone does not necessarily disclose the psychological cause from which the behaviour originated.
### Proposition 116 — Repetition
Repeated observation may increase confidence in a proposed psychological explanation where the same proposed cause repeatedly predicts the same class of behaviour.
### Proposition 117 — Revision
A psychological theory must remain open to reassessment where later observations contradict, modify, or refine an earlier interpretation.
### Proposition 118 — Error Correction
The discovery that an earlier psychological judgment was affected by incomplete information or by the condition of the investigator requires that the judgment be reconsidered rather than preserved merely for consistency.

---

# XXIII. Objectives of the Psychological System
### Proposition 119 — Identification
The first objective of the system is to identify the psychological attributes that influence human behaviour.
### Proposition 120 — Representation
The second objective is to construct a theoretical system capable of representing psychological phenomena in a consistent form.
### Proposition 121 — Prediction
The third objective is to determine whether the theoretical system can be used to predict patterns of behaviour from identified psychological conditions.
### Proposition 122 — Modification
The fourth objective is to use knowledge of psychological causation to understand how beneficial modifications in behaviour may be produced.
### Proposition 123 — Organizations and Communities
The fifth objective is to determine how individual psychological attributes influence the behaviour of organizations, governments, institutions, and communities.
### Proposition 124 — Systemic Weakness
The sixth objective is to identify weaknesses that arise within a system as a consequence of the psychological attributes of the individuals responsible for producing and operating that system.

---

# XXIV. General Principle
### Proposition 125 — Human Systems
Every human institution is simultaneously a structural system and a psychological system because its formal structure is operated through the actions of human beings.
### Proposition 126 — Structure and Psychology
A sound institutional structure may be weakened by destructive psychological attributes, while a defective institutional structure may amplify those attributes.
### Proposition 127 — Government as the Production of Solutions
The quality of government may therefore be measured, in part, according to its ability to identify burdens, produce solutions, preserve the common interest, and prevent private destructive motives from being converted into institutional action.
### Proposition 128 — Final Principle
A rational political system is one in which knowledge of purpose determines the general objective, knowledge of utility assists in evaluating consequences, knowledge of psychology explains the forces that influence human action, and institutional organization converts this knowledge into provisions directed toward the preservation and improvement of life.
