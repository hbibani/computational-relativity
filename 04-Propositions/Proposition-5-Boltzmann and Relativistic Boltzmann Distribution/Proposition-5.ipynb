## Proposition 5: The Boltzmann distribution describes equilibrium probabilities

The Boltzmann distribution describes a small system, which may be as small as a single particle, in thermal contact with a large heat reservoir with which it exchanges energy. It gives the probability of finding the small system in a state of energy $E$ as $P(E)\propto e^{-E/k_BT}$, where $T$ is the temperature of the reservoir.

</p

For a microscopic system in thermal equilibrium with a heat bath,

$$
p_i=\frac{e^{-E_i/(k_B T)}}{Z},
$$

where

$$
Z=\sum_i e^{-E_i/(k_B T)}.
$$

The small system described by the Boltzmann distribution can occupy any of several states, each with its own energy $E_i$ and an associated probability. As the diagram shows, the system may be found in any one of these states.

</p


The interaction with the heat bath is represented implicitly through the equilibrium temperature (T). The distribution does not specify how energy is exchanged or how (E_i) is physically composed. 

Without an additional multiplicity factor,

$$
p_i\propto e^{-E_i/(k_B T)},
$$

### Corollary 1: Separating Boltzmann Distribution into two sections: probability and partitioning

The Boltzmann distribution gives the relative likelihood of a system's equilibrium energy states. It says nothing about where that energy comes from, how it is structured, or how it is partitioned within each state. Students often find the distribution difficult because two distinct kinds of dynamics can be at play:

1. **Occupation of a state:** the probability that the system is found in a state of energy $E_i$, without reference to how that energy is partitioned internally. This is governed by the Boltzmann factor,

$$
p_i \propto e^{-E_i/(k_BT)}.
$$

2. **Restructuring of a state:** a change that alters the internal dynamics of the system's components and, in doing so, also changes the likelihood of the state. This is governed by a multiplicative factor applied to the Boltzmann factor, either $\gamma$ or $\Omega$:

$$
p_i \propto \gamma\, e^{-E_i/(k_BT)}
\qquad\text{or}\qquad
p_i \propto \Omega\, e^{-E_i/(k_BT)},
$$

where $\gamma$ corresponds to **biological partitioning** and $\Omega$ to **random partitioning**.

### Corollary 3: Two Energy Scales for the Exponential Statistical Factor

The exponential factor derived from hidden multiplicity depends on the energy scale associated with the logarithm. In this corollary, we consider two equivalent formulations of relativistic kinetic energy, excluding the rest-energy contribution so that both formulations describe the same physical quantity.

We begin with two dilation terms, each associated with a different energy scale:

$$
\boxed{K=(\gamma-1)mc^2=\gamma_z mv^2}
$$

where

$$
\gamma=\frac{1}{\sqrt{1-\dfrac{mv^2}{mc^2}}}
$$

and

$$
\gamma_z=\frac{1}{\gamma^{-1}+\gamma^{-2}} =\frac{\gamma^2}{\gamma+1}
$$

The two dilation terms are therefore related through:

$$
\boxed{\gamma-1=\gamma_z\frac{mv^2}{mc^2}}
$$

**Route A: Rest-energy scale ($mc^2$)**

Starting from relativistic kinetic energy,

$$
K=(\gamma-1)mc^2=mc^2\ln(e^{\gamma-1})
$$

we propose the corresponding entropic energy definition:

$$
\boxed{E_\Omega=mc^2\ln\Omega_A}
$$

**Route B: Kinetic-energy scale ($mv^2$)**

Using the equivalent kinetic-energy formulation,

$$
K=\gamma_z mv^2=mv^2\ln(e^{\gamma_z})
$$

we obtain the alternative entropic energy definition:

$$
\boxed{E_\Omega=mv^2\ln\Omega_B}
$$

#### Generalized statistical factor

Both formulations follow the same mathematical structure. Provided their respective energy scales remain constant and positive during energy transfer, the previous multiplicity derivation applies without modification.

| Property               | Route A               | Route B               |
| ---------------------- | --------------------- | --------------------- |
| Energy scale           | $mc^2$            | $mv^2$            |
| Dilation term          | $\gamma-1$        | $\gamma_z$        |
| Hidden multiplicity    | $e^{\gamma-1}$    | $e^{\gamma_z}$    |
| Logarithmic derivative | $\dfrac{1}{mc^2}$ | $\dfrac{1}{mv^2}$ |
| Exponential factor     | $e^{-E/(mc^2)}$   | $e^{-E/(mv^2)}$   |

The distinction between the two formulations is not in the kinetic energy they describe, but in how that energy is expressed through its characteristic energy scale and hidden multiplicity.

#### Relationship to the Boltzmann factor

Comparing the two formulations with the conventional Boltzmann factor,

$$
\boxed{ e^{-E/(k_BT)} \qquad e^{-E/(mc^2)} \qquad e^{-E/(mv^2)} }
$$

we observe that Route A establishes a fixed energy scale determined by rest mass, whereas Route B establishes a variable energy scale determined by velocity.

Under classical equipartition, for one translational degree of freedom,

$$
m\langle v_x^2\rangle=k_BT
$$

Route B therefore reproduces the conventional Boltzmann factor when velocity represents the corresponding root-mean-square thermal velocity.

Furthermore, as velocity approaches the speed of light, both proposed exponential factors converge for fixed transferred energy:

$$
\boxed{ \lim_{v\to c}e^{-E/(mv^2)} =e^{-E/(mc^2)} }
$$

#### Condition of applicability

Both derivations assume that their respective energy scales remain constant during energy transfer. For Route B, this requires velocity to be treated as a fixed characteristic property of the reservoir. Otherwise, higher-order logarithmic derivatives generally reappear.

#### Relationship between the multiplicities and the statistical factors

The two hidden multiplicities are:

$$
\boxed{ \Omega_A=e^{\gamma-1}, \qquad \Omega_B=e^{\gamma_z} }
$$

Since both formulations describe identical kinetic energy,

$$
mc^2\ln\Omega_A=mv^2\ln\Omega_B
$$

Rearranging,

$$
\ln\Omega_A=\frac{mv^2}{mc^2}\ln\Omega_B
$$

Therefore,

$$
\boxed{\Omega_A=\Omega_B^{mv^2/mc^2}}
$$

Although the multiplicities differ in magnitude, their corresponding energy scales compensate for this difference, producing identical kinetic energy.
