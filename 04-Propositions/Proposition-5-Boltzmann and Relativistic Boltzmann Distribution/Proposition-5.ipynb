## Proposition 5: The Boltzmann distribution describes equilibrium probabilities

The Boltzmann distribution describes a small system, which may be as small as a single particle, in thermal contact with a large heat reservoir with which it exchanges energy. It gives the probability of finding the small system in a state of energy $E$ as $P(E)\propto e^{-E/k_BT}$, where $T$ is the temperature of the reservoir.

<p align="center">
  <img src="../../Assets/Diagram-19.gif" width="600">
</p

For a microscopic system in thermal equilibrium with a heat bath,

$$
p_i=\frac{e^{-E_i/(k_B T)}}{Z},
$$

where

$$
Z=\sum_i e^{-E_i/(k_B T)}.
$$

The small system described by the Boltzmann distribution can occupy any of several states, each with its own energy $E_i$ and an associated probability. As the diagram shows, the system may be found in any one of these states.

<p align="center">
  <img src="../../Assets/Diagram-20.gif" width="600">
</p


The interaction with the heat bath is represented implicitly through the equilibrium temperature (T). The distribution does not specify how energy is exchanged or how (E_i) is physically composed. 

Without an additional multiplicity factor,

$$
p_i\propto e^{-E_i/(k_B T)},
$$

### **Corollary 1: Relativistic Boltzmann Distribution in the Presence of Common System Motion**

This corollary brings the Lorentz factor into the statistical exponent by distinguishing **common system motion** from **individual molecular motion**. The two original energy scales, $mc^2$ and $mv^2$, are kept, but their effective statistical energy scales contract to

$$
\boxed{\frac{mc^2}{\gamma}\qquad\text{and}\qquad\frac{mv^2}{\gamma}}
$$

#### **1. Two equivalent dilation terms**

Excluding rest energy, relativistic kinetic energy can be written on two energy scales:

$$
\boxed{K=(\gamma-1)mc^2=\gamma_z\,mv^2}
$$

where

$$
\gamma=\frac{1}{\sqrt{1-\dfrac{mv^2}{mc^2}}},
\qquad
\gamma_z=\frac{1}{\gamma^{-1}+\gamma^{-2}}=\frac{\gamma^2}{\gamma+1},
\qquad
\gamma-1=\gamma_z\frac{mv^2}{mc^2}
$$

Each route defines a hidden multiplicity and an entropic energy:

| | Route A: rest-energy scale | Route B: kinetic-energy scale |
|---|---|---|
| Kinetic energy | $K=mc^2\ln\!\left(e^{\gamma-1}\right)$ | $K=mv^2\ln\!\left(e^{\gamma_z}\right)$ |
| Hidden multiplicity | $\Omega_A=e^{\gamma-1}$ | $\Omega_B=e^{\gamma_z}$ |
| Entropic energy | $E_\Omega=mc^2\ln\Omega_A$ | $E_\Omega=mv^2\ln\Omega_B$ |

Because both describe the same $K$,

$$
mc^2\ln\Omega_A=mv^2\ln\Omega_B
\quad\Longrightarrow\quad
\boxed{\Omega_A=\Omega_B^{\,mv^2/mc^2}}
$$

The multiplicities differ in magnitude, but their energy scales compensate exactly.

#### **2. How common system motion enters**

Two components of motion are distinguished:

$$
\underbrace{\vec v}_{\text{Common system motion}}
\qquad
\underbrace{\vec u'_i}_{\text{Individual molecular motion}}
$$

The Lorentz transformation relates the internal energy transfer $\epsilon$ (system rest frame) to the external energy transfer $E$:

$$
E=\gamma\left(\epsilon+v\,\Delta p'_x\right)
$$

For a transfer with no longitudinal momentum component, $\Delta p'_x=0$, so

$$
\boxed{E=\gamma\epsilon}
$$

Substituting into the original factors $e^{-E/mc^2}$ and $e^{-E/mv^2}$ gives

$$
\boxed{e^{-\frac{\gamma\epsilon}{mc^2}}\qquad e^{-\frac{\gamma\epsilon}{mv^2}}}
$$

#### **3. The multiplicity-associated energy shift (general form)**

Let $\mathcal E$ be a fixed characteristic energy scale ($mc^2$ for Route A, $mv^2$ for Route B). Then

$$
P(\epsilon)\propto\Omega\,e^{-\frac{\gamma\epsilon}{\mathcal E}}
=e^{-\frac{\gamma\epsilon}{\mathcal E}+\ln\Omega}
=\exp\!\left[-\frac{\gamma}{\mathcal E}\left(\epsilon-\frac{\mathcal E}{\gamma}\ln\Omega\right)\right]
$$

so the multiplicity acts as an energy shift

$$
\boxed{\Delta\epsilon=-\frac{\mathcal E}{\gamma}\ln\Omega}
$$

Applying this to each route:

| Property | Route A | Route B |
|---|---|---|
| Statistical weighting | $P_A\propto\Omega_A\,e^{-\gamma\epsilon/mc^2}$ | $P_B\propto\Omega_B\,e^{-\gamma\epsilon/mv^2}$ |
| Log-derivative, $\partial\ln\Omega_R/\partial\epsilon$ | $-\dfrac{\gamma}{mc^2}$ | $-\dfrac{\gamma}{mv^2}$ |
| Effective energy scale | $\dfrac{mc^2}{\gamma}$ | $\dfrac{mv^2}{\gamma}$ |
| Energy shift | $\Delta\epsilon_A=-\dfrac{mc^2}{\gamma}\ln\Omega_A$ | $\Delta\epsilon_B=-\dfrac{mv^2}{\gamma}\ln\Omega_B$ |

Common system motion steepens the exponential decay and puts the inverse Lorentz factor into both effective energy scales.

#### **4. Equivalence is preserved**

Dividing $mc^2\ln\Omega_A=mv^2\ln\Omega_B$ by $\gamma$ gives $\Delta\epsilon_A=\Delta\epsilon_B$. Substituting the hidden multiplicities,

$$
\Delta\epsilon_A=-\frac{mc^2}{\gamma}(\gamma-1),
\qquad
\Delta\epsilon_B=-\frac{mv^2}{\gamma}\gamma_z
$$

and therefore

$$
\boxed{\Delta\epsilon_A=\Delta\epsilon_B=-\frac{K}{\gamma}}
$$

The two kinetic-energy formulations stay equivalent after the contraction is introduced.

#### **5. Recovery of the contraction factor**

Hold $\mathcal E$ and $\Omega$ fixed. At rest ($\gamma=1$), $\Delta\epsilon_0=-\mathcal E\ln\Omega$. In motion, $\Delta\epsilon_v=-(\mathcal E/\gamma)\ln\Omega$. The ratio is

$$
\boxed{\frac{\Delta\epsilon_v}{\Delta\epsilon_0}=\frac{1}{\gamma}=\sqrt{1-\frac{mv^2}{mc^2}}}
$$

For a particle moving purely transversely in the moving system ($u'_x=0$), special relativity gives

$$
u_y=\frac{u'_y}{\gamma}=u'_y\sqrt{1-\frac{mv^2}{mc^2}}
$$

Hence

$$
\boxed{\frac{\Delta\epsilon_v}{\Delta\epsilon_0}=\frac{u_y}{u'_y}=\frac{1}{\gamma}}
$$

The statistical energy-scale contraction has exactly the same dependence on common system velocity as the transverse velocity transformation.

#### **6. Relationship to the conventional Boltzmann factor**

With the $\gamma$-multiplicity of the framework as a multiplicative factor (distinct from the Lorentz factor),

$$
P(E)\propto\gamma\,e^{-\frac{E}{k_BT}}
=\exp\!\left[-\frac{1}{k_BT}\left(E-k_BT\ln\gamma\right)\right]
\quad\Longrightarrow\quad
\Delta E=-k_BT\ln\gamma
$$

This gives three characteristic statistical energy scales:

$$
\boxed{
\underbrace{k_BT}_{\text{Thermodynamic}}
\qquad
\underbrace{\frac{mc^2}{\gamma}}_{\text{Route A}}
\qquad
\underbrace{\frac{mv^2}{\gamma}}_{\text{Route B}}
}
$$

**Thermal-energy scale.** If the Route B scale is set by the *internal thermal velocity* rather than the common system velocity, equipartition for one translational degree of freedom gives $m\langle u_x'^2\rangle=k_BT$, so

$$
P(\epsilon)\propto\Omega\,e^{-\frac{\gamma\epsilon}{k_BT}}=\Omega\,e^{-\frac{\epsilon}{k_BT/\gamma}}
\qquad\Longrightarrow\qquad
\boxed{\mathcal E_T=\frac{k_BT}{\gamma}}
$$

This is a mathematical correspondence within the proposed formulation, not an established transformation of physical temperature.

#### **7. Behaviour as $v\to c$**

$$
\frac{\mathcal E_A}{mc^2}=\sqrt{1-\frac{v^2}{c^2}},
\qquad
\frac{\mathcal E_B}{mc^2}=\frac{v^2}{c^2}\sqrt{1-\frac{v^2}{c^2}},
\qquad
\boxed{\frac{\mathcal E_B}{\mathcal E_A}=\frac{v^2}{c^2}\xrightarrow{\,v\to c^-\,}1}
$$

Both energy scales go to zero while their ratio goes to unity. For any fixed $\epsilon>0$,

$$
\lim_{v\to c^-}e^{-\frac{\gamma\epsilon}{mc^2}}=\lim_{v\to c^-}e^{-\frac{\gamma\epsilon}{mv^2}}=0
$$

so the two statistical factors converge in their limiting behaviour.

#### **8. Conditions of applicability**

- The common system velocity $v$, and hence $\gamma$, is constant during each microscopic energy transfer.
- The longitudinal momentum transfer is zero, so $E=\gamma\epsilon$.
- The characteristic reservoir energy scale $\mathcal E$ is fixed.
- Route B needs a nonzero characteristic velocity. If its scale is set by $v$, it cannot be evaluated at $v=0$, because $mv^2=0$ there.
- Common system velocity and internal thermal velocity are different physical quantities, and only the second enters equipartition.
- Isolating the $1/\gamma$ contraction requires $\mathcal E$ and $\Omega$ to be held constant.
- A uniform multiplicity prefactor cancels on normalisation. Physically distinguishable energy shifts therefore need relative or state-dependent multiplicities.
- "Contraction" here means the statistical energy scale and its correspondence with the transverse velocity measured by an external observer. It does not mean contraction of transverse molecular spacing.

#### **Summary**

$$
\boxed{
\begin{aligned}
&K=(\gamma-1)mc^2=\gamma_z\,mv^2\\[6pt]
&E=\gamma\epsilon\\[6pt]
&P_A(\epsilon)\propto\Omega_A\,e^{-\frac{\gamma\epsilon}{mc^2}},
\qquad
P_B(\epsilon)\propto\Omega_B\,e^{-\frac{\gamma\epsilon}{mv^2}}\\[6pt]
&\Delta\epsilon_A=\Delta\epsilon_B=-\frac{K}{\gamma}\\[6pt]
&\frac{\Delta\epsilon_v}{\Delta\epsilon_0}=\frac{u_y}{u'_y}=\frac{1}{\gamma}
\end{aligned}
}
$$

**Common system motion enters the statistical formulation through the relativistic transformation of energy, and its effect on the multiplicity-associated energy scale appears as a contraction.**

This gives a mathematical correspondence between the proposed entropic statistical formulation and the transverse velocity transformation of special relativity, and a basis for studying how common system motion relates to individual molecular motion within Computational Relativity. The correspondence is exact under the stated assumptions. Reading it as a physical change in intrinsic molecular statistics is a proposed extension that still needs independent validation.

### Corollary 2: Separating Boltzmann Distribution into two sections: probability and partitioning

The Boltzmann distribution gives the relative likelihood of a system's equilibrium energy states. It says nothing about where that energy comes from, how it is structured, or how it is partitioned within each state. Students often find the distribution difficult because two distinct kinds of dynamics can be at play:

1. **Occupation of a state:** the probability that the system is found in a state of energy $E_i$, without reference to how that energy is partitioned internally. This is governed by the Boltzmann factor,

$$
p_i \propto e^{-E_i/(k_BT)}.
$$

2. **Restructuring of a state:** a change that alters the internal dynamics of the system's components and, in doing so, also changes the likelihood of the state. This is governed by a multiplicative factor applied to the Boltzmann factor, either $\gamma$ or $\Omega$:

$$
p_i \propto \gamma\, e^{-E_i/(k_BT)}
\qquad\text{or}\qquad
p_i \propto \Omega\, e^{-E_i/(k_BT)},
$$

### Corollary 3: Computational Fields reduce the computational time to produce specific states

The relationship between the Boltzmann distribution and relativity in a multiparticle system can be expressed through probability dilation. When probability dilation is represented as a computational field, the multiplicity provides the corresponding representation of that field within the system. The presence of this multiplicity changes the rate at which the system generates its energetic configuration as it approaches equilibrium. Consequently, the equilibrium state can be reached over a reduced elapsed time relative to the corresponding system in the absence of the multiplicity field.
