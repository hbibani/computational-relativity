## Thought Experiment: Perpendicular Motion Inside a Relativistic Train

Consider a train moving horizontally at velocity

$$
V=0.9c.
$$

Inside the train, an object moves vertically upward with velocity

$$
v.
$$

The velocity (v) is measured in the train’s own reference frame. We want to determine the effective vertical velocity measured by an observer standing on the ground.

We compare two cases:

1. The train is at rest relative to the ground observer.
2. The train moves horizontally at (0.9c).

## 2. Derivation Using Distance and Dilated Time

## Method 1: Relativistic velocity transformation

$$
u'_x=0,
\qquad
u'_y=v
$$

$$
u_y=\frac{u'_y}{\gamma\left(1+\frac{u'_xV}{c^2}\right)}
=\frac{v}{\gamma}
$$

$$
u_y=v\sqrt{1-\frac{V^2}{c^2}}
$$

$$
V=0.9c
\quad\Longrightarrow\quad
\boxed{u_y\approx0.4359v}
$$

## Method 2: Distance over dilated time

$$
v=\frac{s}{t},
\qquad
t'=\frac{t}{\sqrt{1-\frac{V^2}{c^2}}}
$$

$$
v_{\mathrm{eff}}=\frac{s}{t'}
=\frac{s}{t}\sqrt{1-\frac{V^2}{c^2}}
$$

$$
v_{\mathrm{eff}}
=v\sqrt{1-\frac{V^2}{c^2}}
$$

$$
V=0.9c
\quad\Longrightarrow\quad
\boxed{v_{\mathrm{eff}}\approx0.4359v}
$$


---

If the **observed perpendicular velocity** is set equal to the train velocity (V), then

$$
u_{\mathrm{eff}}=u\sqrt{1-\frac{V^2}{c^2}}=V
$$

Therefore,

$$
\boxed{
u=\frac{V}{\sqrt{1-\frac{V^2}{c^2}}}
