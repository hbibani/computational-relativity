## Derivation: From $\tfrac12 mv'^2$ to $E = \gamma mc^2$

Throughout, $v$ is the coordinate velocity $dx/dt$, $v' = \gamma v$ is the proper velocity $dx/d\tau$, and $m$ is constant.

---

### 1. Express $\tfrac12 mv'^2$ in terms of $\gamma$

Since $v' = \gamma v$ and $\gamma^2 = \dfrac{1}{1-v^2/c^2}$,

$$ \frac12 mv'^2 = \frac12 m\gamma^2 v^2 = \frac{\frac12 mv^2}{1-v^2/c^2} $$

Pull out a factor of $c^2$:

$$ \frac12 m\gamma^2 v^2 = \frac12 mc^2\left(\gamma^2\frac{v^2}{c^2}\right) $$

To simplify the bracket, multiply the definition of $\gamma^2$ through by its denominator and expand:

$$ \gamma^2\left(1-\frac{v^2}{c^2}\right) = 1 \quad\Longrightarrow\quad \gamma^2 - \gamma^2\frac{v^2}{c^2} = 1 \quad\Longrightarrow\quad \gamma^2\frac{v^2}{c^2} = \gamma^2 - 1 $$

Therefore,

$$ \boxed{ \frac12 mv'^2 = \frac12 mc^2(\gamma^2-1) } $$

---

### 2. Differentiate with respect to $\gamma$

$$ d\left(\tfrac12 mv'^2\right) = \frac12 mc^2(2\gamma)\,d\gamma = mc^2\gamma\,d\gamma $$

---

### 3. Relate the same differential to $dK$

Differentiate $\tfrac12 mv'^2$ directly:

$$ d\left(\tfrac12 mv'^2\right) = \tfrac12 m(2v')\,dv' = mv'\,dv' $$

Since $p = mv'$, we have $dp = m\,dv'$, so grouping as $v'(m\,dv')$:

$$ d\left(\tfrac12 mv'^2\right) = v'\,dp = \frac{p}{m}\,dp = \gamma v\,dp $$

The relativistic work–energy theorem gives

$$ dK = F\,dx = v\,dp $$

so multiplying by $\gamma$:

$$ \boxed{ d\left(\tfrac12 mv'^2\right) = \gamma\,dK } $$

The factor of $\gamma$ appears because $v' = \gamma v$: the energy differential built from proper velocity runs $\gamma$ times faster than the one built from coordinate velocity.

---

### 4. Solve for $K$

Equating the results of Steps 2 and 3:

$$ \gamma\,dK = mc^2\gamma\,d\gamma $$

Cancel $\gamma$:

$$ dK = mc^2\,d\gamma $$

Integrate from rest, where $\gamma = 1$ and $K = 0$, to the final velocity:

$$ K = \int_1^{\gamma} mc^2\,d\tilde\gamma = mc^2(\gamma-1) $$

This is the standard relativistic kinetic energy.

---

### 5. Total energy

Adding the rest energy $E_0 = mc^2$:

$$ E = K + E_0 = mc^2(\gamma - 1) + mc^2 $$

$$ \boxed{ E = \gamma mc^2 } $$

---

### 6. Kinetic energy in terms of $mv'^2$ and $mv^2$

Factor the difference of squares from Step 1:

$$ \frac12 mv'^2 = \frac12 mc^2(\gamma-1)(\gamma+1) = \frac{\gamma+1}{2}K $$

so

$$ \boxed{ K = \frac{mv'^2}{\gamma+1} = \frac{\gamma^2 mv^2}{\gamma+1} } $$

Dividing the numerator and denominator by $\gamma^2$ and splitting the denominator,

$$ \frac{\gamma+1}{\gamma^2} = \frac{1}{\gamma} + \frac{1}{\gamma^2} = \gamma^{-1} + \gamma^{-2} $$

which gives

$$ \boxed{ K = \frac{mv^2}{\gamma^{-1}+\gamma^{-2}} } $$

Here $\gamma^{-1}$ means $1/\gamma$ (not $\gamma - 1$), and $\gamma^{-2}$ means $1/\gamma^2$ (not $\gamma - 2$). As $v \to 0$, $\gamma \to 1$ and this reduces to the classical $\tfrac12 mv^2$.
