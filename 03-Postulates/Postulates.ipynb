## Postulate 1: Absolute Reference Frame, Time Dilation and Absolute Time

**Premise.** Computational processes need an absolute reference frame to coordinate and synchronise interactions. A *computational space* is proposed in which organised computational structures exist independently of material structures, and which is hypothesised to underlie the emergence of life.

**Time dilation** happens when a computational process takes extra time to finish relative to the absolute frame. Each change of a particle's position is a computational process. Different processes produce a spectrum of dilation terms that differ in both magnitude and internal form.

**Properties of the absolute frame:**
1. It is at rest and/or infinitely far from gravitational fields.
2. No time dilation occurs within it.
3. All observers agree on the time elapsed relative to it.
4. The most time elapses within it.

**Mathematics.** With bibs-rate constant $k_b = B/S$ (bibs per second):

$$
\Delta b = k_b\,\gamma\,\Delta t, \qquad \gamma = \frac{1}{\sqrt{1 - \dfrac{mv^2}{mc^2}}}
$$

Absolute time is proposed as a hidden variable underlying the Lorentz transformations.

## Postulate 2: Probability Dilation

**Premise.** Motivated by the DA-prefix observations documented in `experiment.ipynb`, this postulate proposes that a computational field $\gamma$ interacts with a probabilistic system, modifying its statistical behaviour by altering the underlying physical transitions that govern its movement:

$$
p_{\text{obs}} = \gamma\, p_{\text{exp}} \quad\Longleftrightarrow\quad \gamma = \frac{p_{\text{obs}}}{p_{\text{exp}}}
$$

**Interpretation.** This is an abstract description of how a field can reshape a system's probability space. The postulate does not yet say where those probabilities come from in energy or mechanics. Postulates 3–5 build that link.

## Postulate 3: The K–e Bridge

**Premise.** The relativistic ratio is read as a ratio of energy scales:

$$
\frac{\tfrac12 mv^2}{\tfrac12 mc^2} = \frac{mv^2}{mc^2} = \frac{mv^2}{mc^2}
$$

**K–e relation.** The kinetic state that results is the kinetic energy left over after part of it has been allocated elsewhere:

$$
K' = K - e
$$

**Relativistic form.** The quantity that controls the subtraction is itself an energy ratio:

$$
\tfrac12 mv^2 = \tfrac12 mv'^2 - \tfrac12\left(\frac{mv^2}{mc^2}\right) mv'^2
$$

**Interpretation.** $v^2/c^2$ compares the kinetic-energy scale of motion with a reference energy scale. This sets up the later links to the computational energy scale $k_e$ and the thermal scale $k_BT$.

## Postulate 4: Energy Partitioning Between Motion and Computational Fields

**Premise.** The normalised energy ratio from Postulate 3 splits into a motion share and a field share:


$$
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{mv^2}{mc^2}
\left(
1-
\sum_{n=1}^{\infty}
\frac{1}{k_e^{(n-1)}}
\right)
+
\frac{mv^2}{mc^2}
\sum_{n=1}^{\infty}
\frac{1}{k_e^{(n-1)}}.
$$

When $S$ converges, the two shares add up to the total:

$$
D_{\text{motion}} + D_{\text{field}} = \frac{mv^2}{mc^2}
$$

From this formulation, we can subsequently derive the mass–energy equivalence relation, $E=mc^2$

$$
K=mc^2D_{\mathrm{field}}=(\gamma-1)mc^2.
$$

**Interpretation.** The same energy magnitude can be realised in different physical forms. Magnitude and partitioning are separate properties, and this is the basis for telling transition systems apart.

## Postulate 5: Hidden Multiplicity and Logarithmic Relativistic Energy

**Premise.** Multiplicity functions characterise different *computational transition systems*. Each one affects how accessible configurations are weighted and how energy is partitioned internally.

**Occupation vs. restructuring of states:**

$$
\underbrace{p_i \propto e^{-E_i/k_BT}}_{\text{occupation}}
\qquad\qquad
\underbrace{p_i \propto \Omega_i\, e^{-E_i/k_BT} \;\;\text{or}\;\; p_i \propto \gamma_i\, e^{-E_i/k_BT}}_{\text{restructuring}}
$$

| Weight | Partitioning type | Meaning |
|---|---|---|
| $\Omega$ | Random | Entropic multiplicity: accessible microstates |
| $\gamma$ | Biological | Proposed dilation-dependent computational organisation |

*This is a conceptual distinction, not an established microscopic model.*

**Hidden multiplicity.** Starting from $K = (\gamma-1)\,mc^2$:



$$
K = mc^2 \ln\!\left(e^{\gamma-1}\right), \qquad x_A \equiv e^{\gamma-1} \;\Rightarrow\; \gamma-1 = \ln x_A, \qquad K = mc^2 \ln x_A
$$

Equivalently, starting from $K = \gamma_z\,mv^2$:

$$
K = mv^2 \ln\!\left(e^{\gamma_z}\right), \qquad x_B \equiv e^{\gamma_z} \;\Rightarrow\; \gamma_z = \ln x_B, \qquad K = mv^2 \ln x_B
$$

Where

$$
\gamma_z=\frac{1}{\gamma^{-1}+\gamma^{-2}}.
$$

Note that the modified $\gamma$ term is specifically defined to account for the dilation of kinetic energy, thereby establishing the mathematical equivalence between the two expressions.

$$K = mv^2 \ln\!\left(e^{\gamma_z}\right) =  mc^2 \ln\!\left(e^{\gamma-1}\right) $$

**Transition systems:**

$$
E_\gamma = mc^2 \ln x_A \quad \text{(biological)}, \qquad E_\Omega = mc^2 \ln \Omega \quad \text{(entropic)}
$$

$$
E_{\gamma_z} = mv^2 \ln x_B \quad \text{(biological)}, \qquad E_\Omega = mv^2 \ln \Omega \quad \text{(entropic)}
$$
**Statement.** Relativistic dilation is proposed to have a hidden multiplicity $x = e^\gamma$. By analogy with thermodynamics, different multiplicity structures describe different transition systems. **Two systems can have the same energy and still be organised differently.**

## Open Mathematical Notes

- **P1:** Write $k_b$ (bibs rate) so it isn't confused with Boltzmann's $k_B$.
- **P2:** $p_{\text{obs}}$ must stay in $[0,1]$ and remain normalised. Because $\gamma \ge 1$, that needs $p_{\text{exp}} \le 1/\gamma$ or a renormalisation step.
- **P4:** For $k_e > 1$, the series sums to $S = \dfrac{k_e}{k_e - 1} > 1$, so $D_{\text{motion}} < 0$. You'll want either to redefine $S$ (for example, start at $n = 2$, which gives $S = \tfrac{1}{k_e - 1}$) or to explain what a negative motion share means. If $k_e$ is an energy, it also has to be made dimensionless (for example, $k_e / k_BT$).
- **P5:** $\ln(e^\gamma) = \gamma$ is an exact identity. Treating $e^\gamma$ as a *physical* multiplicity is a separate hypothesis and needs its own justification.
