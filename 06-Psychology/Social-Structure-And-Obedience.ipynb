### Proposition 1

The human body is affected by external objects in different ways according to the category to which those objects belong.

### Demonstration 1

The human body responds to another human being differently from the way it responds to an inanimate object or to an animate organism that is not human. The nature of the impression produced upon the human body therefore depends, in part, upon the kind of object producing it.

---

### Proposition 2

Biological organisms that live in close association with one another tend to develop social structures directed toward particular collective or biological purposes.


### Corollary 1

The manner in which one human being is impressed by the presence and actions of another may therefore serve a regulatory function, influencing and constraining human behavior within a social structure.

---

### Proposition 3

The human body experiences pleasure or pain as a consequence of the actions of other human beings. These experiences produce corresponding impressions upon the body and mind.

---

### Proposition 4

When the human mind has not been carefully educated in matters of psychology, it tends to rely upon these immediate impressions—particularly the pleasures and pains associated with particular actions—to determine the value of the actions performed by other human beings.

By extension, the mind may also use these impressions to form judgments concerning the value or character of the individuals who perform those actions.

---

### Proposition 5

Because the human mind possesses a highly developed capacity for reasoning, memory, anticipation, and social judgment, human behavior is regulated through a complex system of pleasures, pains, and the impressions associated with particular actions and their consequences.

Every significant human action may therefore produce an impression upon the body or mind. These impressions influence how actions are evaluated, remembered, and responded to, and consequently contribute to the regulation of future human behavior.


---

### Proposition 6:



**Filisific Calculus** is a rational mechanism employed by advanced human societies for evaluating human actions and their consequences independently of immediate emotional impressions.

Its purpose is to regulate the internal impressions associated with particular actions by subjecting those impressions to rational, proportional, and systematic judgment.

In particular, Filisific Calculus seeks to restrain violent, excessively punitive, or otherwise disproportionate impressions that may arise from the pain associated with another person's actions. Without such regulation, these impressions may influence social or political decisions in ways that produce harm disproportionate to the conduct that originally produced them.

In this sense, Filisific Calculus functions as a corrective mechanism between the immediate impressions produced within the human body and the deliberate formation of judgments, laws, and social policy.
