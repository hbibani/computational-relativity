## Proposition 1: Definition of Hatred

Hatred is an emotional state characterized by feelings of animosity, hostility, or disdain directed toward another individual.

---

### Corollary 1: On the Influence of Hatred in the Characterization and Interpretation of Human Beings

Hatred has the propensity to distort, exaggerate, or otherwise influence the descriptions and judgments attributed to the character of other human beings. In this respect, systems of hatred and the evaluative processes through which individuals interpret one another may be understood as systems that construct descriptions of their experiences of objects, individuals, and phenomena within the world.

Consequently, feelings of animosity may influence not only an individual's emotional response toward another person but also the manner in which that person's character, intentions, and behavior are perceived, interpreted, and described. Such descriptions may therefore reflect the subjective emotional disposition of the observer rather than an impartial representation of the individual being observed.

---

## Proposition 2: Classification of Hatred

Hatred can be subdivided into three distinct categories according to the relationship between its underlying cause and the behavior of the individual toward whom it is directed:

1. **Direct hatred:** Hatred arising from circumstances directly attributable to an individual's behavior, whereby the animosity is based on their observable actions or conduct. Such hatred may have a justifiable basis.

2. **Indirect hatred:** Hatred arising from circumstances that are not directly attributable to an individual's behavior, whereby the animosity originates from characteristics, comparisons, or psychological processes unrelated to their conduct. Such hatred is considered unjustifiable within this framework.

3. **Concomitant hatred:** A behavioral phenomenon in which an individual's actions resemble those ordinarily associated with feelings of hatred or animosity toward another person, despite the absence of any underlying hostility or animosity. In this case, the outward expression of behavior is similar to that produced by hatred, but the emotional state responsible for such behavior is fundamentally different or entirely absent.

---

## Proposition 3: Proportionality of Direct Hatred

Hatred arising from an individual's behavior may be further classified according to whether the intensity of the emotional response is proportionate to the conduct being observed.

1. **Proportionate hatred:** An emotional response in which the degree of animosity corresponds to the nature, severity, or consequences of the individual's behavior.

2. **Disproportionate hatred:** An emotional response in which the degree of animosity exceeds what can reasonably be justified by the individual's behavior or is otherwise inconsistent with the nature of their conduct.

Therefore, even when hatred originates from an identifiable behavioral cause, its justification depends upon the proportionality between the observed conduct and the resulting emotional response.

---

### Corollary 1: Forms of Direct Hatred

The identification of a behavioral cause does not, in itself, establish that the resulting hatred is justified. Both indignation and hatred arising from displays of pride may be proportionate or disproportionate depending upon the nature of the conduct and the intensity of the resulting emotional response.Direct hatred may also be distinguished according to the behavioral qualities or characteristics that give rise to the emotional response.

1. **Indignation:**  A form of animosity arising from the direct experience or observation of negative qualities, vices, or objectionable conduct in another individual. Such hatred originates from the individual's perception of morally reprehensible or otherwise objectionable behavior.

2. **Hatred arising from displays of pride:** A form of animosity directed toward individuals who exhibit percieved pride, arrogance and superiority, through their conduct. This form of hatred arises from the perception or experience of these qualities as they are manifested in the individual's behavior, rather than from the mere possession of self-esteem or personal confidence.

---

### Corollary 2: Distinguishing Events and Qualities In People

The world is far more complex than what the human mind can easily understand through experience. Human behavior is influenced by complicated psychological processes that are often difficult to recognize or understand. As a result, people may struggle to identify the underlying causes of behavior and distinguish between different psychological characteristics.

---

#### Demonstration 1:

An example of this difficulty is the inability to distinguish between confidence and pride. Although these psychological characteristics may produce similar outward expressions of behavior, their underlying motivations and psychological mechanisms can differ considerably.

---

### Corollary 3: Extreme Dysfunctionality of Direct Hatred

Conditions may arise in which feelings of animosity become associated with nearly all forms of human behavior, regardless of their actual nature or intent. In such circumstances, individuals who are innocent of any wrongdoing may be perceived as possessing the same negative qualities or engaging in the same objectionable conduct toward which the original hatred was directed.

## Proposition 4: The Unjustifiable Nature of Indirect Hatred

Within this classification, indirect hatred is considered inherently unjustifiable because its underlying causes are independent of the individual's conduct. Such animosity may arise from:

1. **Physical appearance or beauty:** Hatred directed toward another individual because of their physical attractiveness or appearance, neither of which, in itself, constitutes a behavioral justification for animosity.

2. **Intelligence or intellectual capability:** Hatred arising from another individual's intellectual abilities or perceived intelligence, independently of how those abilities are exercised through their behavior.

3. **Envy or self-referential comparison and perceived deprivation:** Hatred originating from an individual's comparison of themselves with another person, whereby the perception of lacking a desirable quality possessed by the other person generates resentment, hostility, or animosity.

4. **Jealousy arising from perceived deprivation:** Hatred directed toward another individual because they possess something that the person experiencing the animosity lacks or perceives themselves to have been deprived of. Such hatred originates from resentment toward another person's possessions, opportunities, advantages, or desirable qualities, rather than from any objectionable behavior on their part.

5. **Transitional hatred arising from the juxtaposition of opposing emotional states:** The emergence of hatred through the successive transition between two contrasting emotional states, particularly pride and hatred, whereby an initial feeling of pride is followed by animosity or hostility toward another individual. This mechanism describes how opposing emotional responses may succeed one another, producing a transition from positive self-regard to negative feelings directed toward another person.

6. **Animosity arising from blame and praise:** The emergence of extreme feelings of hostility, resentment, or hatred in response to blame or praise. This mechanism describes how two seemingly opposing forms of social evaluation may produce similar emotional responses, whereby both negative judgments, expressed through blame, and positive judgments, expressed through praise, become sources of intense animosity.

7. **Animosity arising from infamy and uncritical acceptance of information:** The development of feelings of hatred, hostility, or disdain toward an individual based upon their perceived reputation, accusations, or information communicated by others, without exercising sufficient judgment, impartiality, or critical examination to establish the truth of such claims. Such mechanisms may be heavily influenced by other individuals, particularly when intense feelings of jealousy, envy, or resentment constitute the primary psychological basis upon which information is received, interpreted, and accepted prejudicially. Under such circumstances, pre-existing feelings of jealousy may predispose an individual to accept unfavorable information about another person without adequate scrutiny, while disregarding contradictory evidence or alternative explanations.  

In each case, the emotional response arises not from wrongdoing or objectionable behavior on the part of the individual being observed, but from characteristics or psychological evaluations that do not independently establish a justification for hatred.

---

## Proposition 5: On the Effects of Personality Disturbances on Evaluative and Re-Evaluative Processes

Re-evaluative processes involve the examination of additional information, contradictory evidence, and alternative interpretations that may challenge or invalidate the original basis upon which feelings of animosity were established. However, disturbances in personality functioning may interfere with these evaluative and re-evaluative processes, producing a condition, whereby an individual remains unwilling or unable to revise their negative judgment despite the presentation of contradictory evidence that no longer supports the original emotional response.

---

## Proposition 6: On Vices and Virtues

The causes of these passions may be understood as arising from the representations of virtues and vices attributed to other individuals. However, an individual governed by disproportionate or unregulated passions may possess little or no genuine understanding of the virtues and vices upon which their judgments are based. In cases involving disturbances in personality functioning, the regulation of emotional responses and conduct may lack a consistent evaluative framework grounded in the utility, consequences, or inherent moral worth of the actions being observed. Thus, systems of information governed solely by such passions and disturbances may produce individuals whose capacity to judiciously determine the legitimacy, justification, and rightful appropriation of property is severely impaired.
