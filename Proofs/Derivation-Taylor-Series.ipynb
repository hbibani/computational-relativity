## From Probability Dilation to the Field–Motion Partition

Our original investigation began with the probability relation obtained from the experiment and its interaction with the Boltzmann distribution.

We assumed that the observed probability was dilated relative to the expected probability by a factor $\gamma$:

$$
p_{\mathrm{obs}}
=
\gamma p_{\mathrm{exp}}.
$$

Substituting the Boltzmann form for the expected probability gives

$$
p_{\mathrm{obs}}
\propto
\gamma
e^{-E_{\mathrm{exp}}/(k_B T)}.
$$

From this relation, the corresponding energy shift can be written as

$$
\Delta E
=
-k_B T\ln\gamma.
$$

Using this energy relation as the guide for the method of postulation, we estimated the corresponding temperature-dilation factor as

$$
\boxed{
\gamma_d
\approx
\frac{1}{
1-\frac{k_B T}{k_e\gamma}
}
}.
$$

The next objective was to transfer the thermodynamic structure into relativistic variables. This motivated the correspondence

$$
\boxed{
\frac{k_B T}{k_e\gamma}
\approx
\frac{mv^2}{2k_e}
}.
$$

A deliberate mathematical strategy was required at this stage. It was understood that the multiplicity term could not be transferred directly into the relativistic analogue while preserving the structural correspondence between the two systems ($k_e \gamma \rightarrow k_e$). Instead, the contribution associated with multiplicity was absorbed into the constant, leaving its explicit relativistic definition to be derived at a later stage.

More generally, when two mathematical systems are structurally related because of underlying conceptual similarities, not every term in one system will necessarily possess a direct analogue in the other. When such a term cannot be transferred explicitly without disrupting the shared mathematical structure, its contribution may be provisionally absorbed into an appropriate constant or composite quantity. The concealed contribution can then be recovered and defined independently once the corresponding structure is identified within the target system.


We therefore introduced this quantity while recognizing that the multiplicity contribution absorbed within $k_e$ still required an explicit relativistic analogue. In this document, we describe the sequence of reasoning and mathematical developments that led to the identification of that analogue.

$$
\boxed{
f_e
=
\frac{mv^2}{2k_e}
}.
$$

At this stage, it is important to distinguish between the **origin of this term** and its later, secondary interpretation.

The quantity

$$
\frac{mv^2}{2k_e}
$$

was initially introduced through the transfer from the thermodynamic probability-dilation framework. It was subsequently incorporated into a broader relativistic construction through the **unification-operator interpretation of computational relativity**, in which the field associated with multiplicity, or the enumeration of particles, was unified with the field associated with motion.

This produced the field contribution through the unification operator $\times$, which mathematically combined the field of motion with the field associated with multiplicity into a single unified structure.

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\times
\frac{mv^2}{2k_e}
}
$$



where

$$
\boxed{
V_t^2
=
\sum_i v_i^2
}.
$$


During the earlier stages of the investigation, before the significance of the Taylor-series structure of the Lorentz factor had been recognized, we incorrectly incorporated the product directly inside a Lorentz-like square-root expression. This was a mistake because interpreting the expression solely as the unification of two independent contributions would lead to a formulation that alters the relativistic capacity and therefore conflicts with the requirement that the speed of light remain the invariant limiting constant. 

We had mistakenly placed the terms in this form:
$$
\gamma
=
\frac{1}{
\sqrt{
1-\frac{mv^4}{2k_e c^2}
}
}.
$$

The solution to this problem required a **secondary definition** to the term 

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\times
\frac{mv^2}{2k_e}
}
$$

This means that the expression cannot be understood solely as the unification of two distinct systems through the $\times$ operator. Rather, it admits two complementary interpretations:

1. It represents the **unification of the two fields in terms of their respective contributions via the unification operator**: the field of motion, represented by $\frac{V_t^2}{c^2}$, and the field associated with multiplicity, represented by $\frac{mv^2}{2k_e}$.

2. It represents the **partitioning of the total normalized capacity** $\frac{V_t^2}{c^2}$, where the factor $\frac{mv^2}{2k_e}$ determines the fraction of that already existing contribution assigned to the field component.

Thus, the same mathematical term can be understood both as a unification of two fields and as an internal partition of the total relativistic contribution.

By understanding this alternative definition, we could formulate a solution to the problem, which is described in the next section.

## Striating the Total Motion Contribution

The issue identified in the previous section was resolved by changing the role assigned to

$$
\frac{mv^2}{2k_e}.
$$

Rather than allowing this quantity to introduce an additional contribution to the relativistic capacity, it was reinterpreted as a fraction that acts **within** the already existing normalized motion contribution.

The first step was therefore to recognize that the total normalized motion term could itself be decomposed, or **striated**, into a collection of individual contributions:

$$
1
-
\frac{v_1^2}{c^2}
-
\frac{v_2^2}{c^2}
-
\frac{v_3^2}{c^2}
-
\frac{v_4^2}{c^2}
-
\frac{v_5^2}{c^2}.
$$

More generally,

$$
\boxed{
V_t^2
=
\sum_i v_i^2
}
$$

and therefore

$$
\boxed{
\frac{V_t^2}{c^2}
=
\sum_i
\frac{v_i^2}{c^2}
}.
$$

Within the computational interpretation, each term

$$
\frac{v_i^2}{c^2}
$$

represents an individual contribution to the total normalized demand, while

$$
\frac{V_t^2}{c^2}
$$

represents their combined contribution.

The remaining capacity is therefore determined entirely by this total:

$$
\boxed{
C_{\mathrm{remaining}}
=
1-\frac{V_t^2}{c^2}
}.
$$

This observation provided the key to resolving the earlier problem. The field term did not need to be introduced as an additional contribution outside

$$
\frac{V_t^2}{c^2}.
$$

Instead, the already existing total contribution could be divided internally between two complementary components.

We therefore use

$$
\boxed{
f_e
=
\frac{mv^2}{2k_e}
}
$$

as the fraction of the total normalized contribution assigned to the field-related component.

The field contribution becomes

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
f_e
}
$$

or explicitly,

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_e}
}.
$$

Because this quantity is now understood as a partition of an existing total, the remainder of that same total must belong to the motion component. The complementary fraction is therefore

$$
1-f_e,
$$

giving

$$
\boxed{
D_{\mathrm{motion}}
=
\frac{V_t^2}{c^2}
\left(1-f_e\right)
}
$$

or

$$
\boxed{
D_{\mathrm{motion}}
=
\frac{V_t^2}{c^2}
\left(
1-\frac{mv^2}{2k_e}
\right)
}.
$$

The crucial difference from the earlier formulation is that neither term now exists independently of the original normalized motion contribution. Both are subdivisions of

$$
\frac{V_t^2}{c^2}.
$$

The total contribution is therefore

$$
D_{\mathrm{total}}
=
D_{\mathrm{motion}}
+
D_{\mathrm{field}}.
$$

Substituting the two components gives

$$
D_{\mathrm{total}}
=
\frac{V_t^2}{c^2}
\left(
1-\frac{mv^2}{2k_e}
\right)
+
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_e}.
$$

Factoring out the common normalized motion term,

$$
D_{\mathrm{total}}
=
\frac{V_t^2}{c^2}
\left[
1
-
\frac{mv^2}{2k_e}
+
\frac{mv^2}{2k_e}
\right].
$$

The two internal fractions cancel:

$$
-
\frac{mv^2}{2k_e}
+
\frac{mv^2}{2k_e}
=
0,
$$

leaving

$$
\boxed{
D_{\mathrm{total}}
=
\frac{V_t^2}{c^2}
}.
$$

This is the mathematical resolution of the problem identified in the previous section.

The unification operator produces the field contribution

$$
\frac{V_t^2}{c^2}
\times
\frac{mv^2}{2k_e},
$$

but the secondary interpretation of

$$
\frac{mv^2}{2k_e}
$$

as a partitioning fraction ensures that this contribution does not increase the total normalized demand.

Instead,

$$
\frac{V_t^2}{c^2}
$$

is conserved while being internally divided between the motion and field components.

Consequently, the remaining capacity remains

$$
\boxed{
C_{\mathrm{remaining}}
=
1-D_{\mathrm{total}}
=
1-\frac{V_t^2}{c^2}
}.
$$

The complete partition can therefore be written as

$$
\boxed{
\frac{V_t^2}{c^2}
=
\underbrace{
\frac{V_t^2}{c^2}
\left(
1-\frac{mv^2}{2k_e}
\right)
}_{D_{\mathrm{motion}}}
+
\underbrace{
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_e}
}_{D_{\mathrm{field}}}
}.
$$

The resolution therefore consists of two related steps.

First, the total normalized motion contribution

$$
\frac{V_t^2}{c^2}
$$

is treated as the complete contribution constrained by the limiting capacity.

Second,

$$
\frac{mv^2}{2k_e}
$$

is applied internally as a partitioning fraction that determines how much of that fixed contribution belongs to the field component, with the complementary fraction assigned to motion.

The important conceptual correction is therefore not the removal of the unified term, but a change in its interpretation:

$$
\boxed{
\text{additional demand}
\quad\longrightarrow\quad
\text{internal partition of an existing demand}
}.
$$

This preserves the total normalized contribution while retaining the unification between the field of motion and the field associated with multiplicity.

## The Field Fraction Can Itself Be Striated by Multiplicity

A further point is that the quantity

$$
\frac{mv^2}{2k_e}
$$

does not necessarily have to be interpreted as one indivisible field contribution.

Within the computational-relativity interpretation, the field may contain multiple contributions distinguished according to the **number of particles, multiplicity, or particle sector** associated with them. Each sector \(n\) is associated with its own corresponding energy scale \(k_{e,n}\).

In other words, just as

$$
\frac{V_t^2}{c^2}
$$

can be decomposed into separate velocity contributions,

$$
\frac{V_t^2}{c^2}
=
\frac{v_1^2}{c^2}
+
\frac{v_2^2}{c^2}
+
\frac{v_3^2}{c^2}
+\cdots,
$$

the field fraction may itself possess an internal decomposition.

We may therefore define

$$
\boxed{
f_{e,n}
=
\frac{mv^2}{2k_{e,n}}
}
$$

where each

$$
f_{e,n}
$$

represents the field contribution associated with a particular particle-number or multiplicity sector.

The complete field fraction is then

$$
\boxed{
f_e
=
\sum_n f_{e,n}
=
\sum_n
\frac{mv^2}{2k_{e,n}}
}.
$$

Schematically,

$$
\boxed{
f_e
=
\frac{mv^2}{2k_{e,1}}
+
\frac{mv^2}{2k_{e,2}}
+
\frac{mv^2}{2k_{e,3}}
+
\frac{mv^2}{2k_{e,4}}
+\cdots
}.
$$

The index \(n\) may therefore distinguish sectors containing different numbers of particles or, more generally, different multiplicity configurations, with each sector possessing its own corresponding \(k_{e,n}\).

The field contribution then becomes

$$
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\sum_n
\frac{mv^2}{2k_{e,n}}.
$$

Equivalently,

$$
\boxed{
D_{\mathrm{field}}
=
\sum_n
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_{e,n}}
}.
$$

Thus, the quantity previously represented by the single term

$$
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_e}
$$

may instead be understood as a **series of field contributions**:

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_{e,1}}
+
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_{e,2}}
+
\frac{V_t^2}{c^2}
\frac{mv^2}{2k_{e,3}}
+\cdots
}.
$$

This introduces two distinct forms of striation into the model.

First, the total motion term is striated according to its different computational duties:

$$
\boxed{
\frac{V_t^2}{c^2}
=
\sum_i\frac{v_i^2}{c^2}
}.
$$

Second, the field fraction is striated according to its multiplicity or particle-number sectors:

$$
\boxed{
f_e
=
\sum_n
\frac{mv^2}{2k_{e,n}}
}.
$$

The full field expression consequently takes the form

$$
\boxed{
D_{\mathrm{field}}
=
\left(
\sum_i\frac{v_i^2}{c^2}
\right)
\left(
\sum_n\frac{mv^2}{2k_{e,n}}
\right)
}.
$$

Expanding this produces

$$
D_{\mathrm{field}}
=
\sum_i\sum_n
\frac{v_i^2}{c^2}
\frac{mv^2}{2k_{e,n}}.
$$

Therefore, what initially appeared to be a single field term can, in principle, represent an entire collection of contributions distinguished both by computational duty and by multiplicity, with each multiplicity sector possessing its own corresponding \(k_{e,n}\).

Importantly, the total field fraction is now defined by the sum of these sector-dependent contributions:

$$
\boxed{
f_e
=
\sum_n
\frac{mv^2}{2k_{e,n}}
}.
$$

The complementary motion contribution then becomes

$$
\boxed{
D_{\mathrm{motion}}
=
\frac{V_t^2}{c^2}
\left(
1-
\sum_n
\frac{mv^2}{2k_{e,n}}
\right)
}
$$

and the field contribution becomes

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\sum_n
\frac{mv^2}{2k_{e,n}}
}.
$$

Adding them gives

$$
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\left(
1-
\sum_n
\frac{mv^2}{2k_{e,n}}
\right)
+
\frac{V_t^2}{c^2}
\sum_n
\frac{mv^2}{2k_{e,n}}.
$$

The entire multiplicity series cancels between the complementary partitions:

$$
\boxed{
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
}.
$$

This is important because the cancellation is no longer restricted to a single field fraction. Even if the field contribution contains many distinct particle-number sectors, each with its own \(k_{e,n}\), their detailed internal structure can be preserved while the total contribution still collapses to the same normalized relativistic term.

## The Unification Operator

A useful concept in developing the first part of the expression was what we refer to as the **unification operator**. In this framework, multiplication, written as $\times$, is interpreted as an operation that combines distinct attributes or contributions into a single physical quantity or structure.

A simple example can be taken from Newtonian mechanics. Momentum is defined by

$$
p = m \times v,
$$

where mass $m$ and velocity $v$ describe different physical attributes. Multiplication combines these attributes into the single quantity of momentum. In this sense, the multiplication sign may be viewed as performing a **unifying operation** between the two quantities.

This idea provided a useful way of approaching the first structural component required in computational relativity.

Rather than initially attempting to derive the complete higher-order expression, we considered repeated normalized velocity contributions of the form

$$
\left(\frac{v^2}{c^2}\right)
\times
\left(\frac{v^2}{c^2}\right)
\times
\left(\frac{v^2}{c^2}\right)
\times \cdots
$$

continued through $n$ factors.

This immediately gives

$$
\boxed{
\left(\frac{v^2}{c^2}\right)^n
}
$$

and provides a simple mechanism by which multiple field contributions may be **unified into a single term**.

Thus, the structural part of the expression was comparatively straightforward to identify:

$$
\boxed{
\underbrace{
\left(\frac{v^2}{c^2}\right)
\times
\left(\frac{v^2}{c^2}\right)
\times \cdots \times
\left(\frac{v^2}{c^2}\right)
}_{n\text{ factors}}
=
\left(\frac{v^2}{c^2}\right)^n
}
$$

Within the computational interpretation, each factor may be regarded as a contribution associated with a field or computational component, while the multiplication operator unifies those contributions into one higher-order structure.

The more difficult problem was therefore **not obtaining the power**

$$
\left(\frac{v^2}{c^2}\right)^n,
$$

but determining the correct **coefficient associated with each value of $n$**.

The expression ultimately requires the form

$$
a_n
\left(\frac{v^2}{c^2}\right)^n,
$$

where $a_n$ determines the weighting of the $n$th unified contribution.

The next task is therefore to determine where these coefficients come from and why they take their particular values. This leads directly to the coefficient structure associated with the relevant series expansion.

## Later Recognition of the Taylor-Series Structure

An important feature of this development is that the partitioning argument was constructed **before we recognized its connection with the Taylor expansion of the inverse Lorentz factor**.

The inverse Lorentz factor is

$$
\frac{1}{\gamma}
=
\sqrt{1-\beta^2},
$$

where

$$
\beta=\frac{v}{c}.
$$

Its Taylor expansion is

$$
\frac{1}{\gamma}
=
1
-\frac12\beta^2
-\frac18\beta^4
-\frac1{16}\beta^6
-\frac5{128}\beta^8
-\frac7{256}\beta^{10}
-\cdots.
$$

Squaring the complete series gives

$$
\left(\frac{1}{\gamma}\right)^2
=
\left(\sqrt{1-\beta^2}\right)^2,
$$

and therefore

$$
\boxed{
\frac{1}{\gamma^2}
=
1-\beta^2
}.
$$

Equivalently,

$$
\boxed{
\frac{1}{\gamma^2}
=
1-\frac{v^2}{c^2}
}.
$$

The striking feature is that all higher-order contributions in the Taylor expansion cancel under the convolution produced by squaring the complete series.

For example, setting

$$
x=\beta^2,
$$

we have

$$
\frac1\gamma
=
1
-\frac12x
-\frac18x^2
-\frac1{16}x^3
-\cdots.
$$

When this series is squared, the coefficient of \(x^2\) becomes

$$
-\frac18
+
\frac14
-
\frac18
=
0,
$$

and the coefficient of \(x^3\) similarly becomes

$$
-\frac1{16}
+
\frac1{16}
+
\frac1{16}
-
\frac1{16}
=
0.
$$

The same cancellation continues through every higher power, leaving only

$$
\boxed{
\left(\frac1\gamma\right)^2
=
1-x
}.
$$

Substituting

$$
x=\frac{v^2}{c^2},
$$

gives

$$
\boxed{
\left(\frac1\gamma\right)^2
=
1-\frac{v^2}{c^2}
}.
$$

There is also an important complementary result for the positive Lorentz factor itself:

$$
\gamma
=
1
+\frac12\beta^2
+\frac38\beta^4
+\frac5{16}\beta^6
+\frac{35}{128}\beta^8
+\cdots.
$$

Squaring this series gives

$$
\boxed{
\gamma^2
=
\frac{1}{1-\beta^2}
}
$$

and therefore

$$
\boxed{
\gamma^2
=
1+\beta^2+\beta^4+\beta^6+\beta^8+\cdots
}.
$$

Thus the two directions behave differently:

$$
\boxed{
\frac{1}{\gamma^2}
=
1-\beta^2
}
$$

while

$$
\boxed{
\gamma^2
=
1+\beta^2+\beta^4+\beta^6+\cdots
}.
$$

The inverse series collapses under squaring, whereas the positive series becomes an infinite geometric series whose coefficients are all equal to \(1\).

## Connection with the Earlier Partitioning Argument

This later Taylor-series result reveals a structural connection with the earlier computational partition.

From the computational argument,

$$
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}.
$$

If the field contribution is itself expanded according to multiplicity, with each multiplicity sector \(n\) associated with its own energy scale \(k_{e,n}\), then

$$
\boxed{
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\sum_n
\frac{mv^2}{2k_{e,n}}
}.
$$

The complementary motion contribution is therefore

$$
\boxed{
D_{\mathrm{motion}}
=
\frac{V_t^2}{c^2}
\left(
1-
\sum_n
\frac{mv^2}{2k_{e,n}}
\right).
}
$$

Consequently,

$$
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}
\left(
1-
\sum_n
\frac{mv^2}{2k_{e,n}}
\right)
+
\frac{V_t^2}{c^2}
\sum_n
\frac{mv^2}{2k_{e,n}}.
$$

The internal multiplicity series cancels between the two complementary partitions, leaving

$$
\boxed{
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{V_t^2}{c^2}.
}
$$

Therefore,

$$
1-
\left(
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
\right)
=
1-\frac{V_t^2}{c^2}.
$$

The relativistic identity has the same compact form:

$$
\boxed{
\frac{1}{\gamma^2}
=
1-\frac{V_t^2}{c^2}
}
$$

when \(V_t\) is the velocity quantity entering the proposed dilation structure.

Thus,

$$
\boxed{
\frac{1}{\gamma^2}
=
1-
D_{\mathrm{motion}}
-
D_{\mathrm{field}}.
}
$$

Substituting the multiplicity-expanded form gives

$$
\frac{1}{\gamma^2}
=
1
-
\frac{V_t^2}{c^2}
\left(
1-
\sum_n
\frac{mv^2}{2k_{e,n}}
\right)
-
\frac{V_t^2}{c^2}
\sum_n
\frac{mv^2}{2k_{e,n}}.
$$

Combining the terms,

$$
\frac{1}{\gamma^2}
=
1
-
\frac{V_t^2}{c^2}
\left[
1
-
\sum_n
\frac{mv^2}{2k_{e,n}}
+
\sum_n
\frac{mv^2}{2k_{e,n}}
\right].
$$

Therefore,

$$
\boxed{
\frac{1}{\gamma^2}
=
1-\frac{V_t^2}{c^2}.
}
$$

This makes the structural correspondence more interesting than in the original single-term formulation. The computational model does not necessarily contain only one field contribution. Instead, the field fraction may contain a series of contributions,

$$
\boxed{
\frac{mv^2}{2k_{e,1}},
\quad
\frac{mv^2}{2k_{e,2}},
\quad
\frac{mv^2}{2k_{e,3}},
\quad
\ldots
}
$$

with each \(k_{e,n}\) corresponding to a particular particle-number or multiplicity sector.

Yet after these internal contributions are distributed between the field and motion portions, the complete expression still reduces to the simple normalized term

$$
1-\frac{V_t^2}{c^2}.
$$

The later Taylor-series investigation revealed an analogous mathematical phenomenon. The inverse Lorentz factor contains an infinite sequence of higher-order terms,

$$
1
-\frac12\beta^2
-\frac18\beta^4
-\frac1{16}\beta^6
-\cdots,
$$

but after the complete series is squared, those higher-order contributions combine and cancel so that only

$$
1-\beta^2
$$

remains.

The earlier computational construction therefore came close to reproducing an important **structural principle** before the Taylor-series connection had been identified: a system may contain a much richer internal decomposition than is visible in its final compact relativistic expression.

In the computational construction, the hidden structure consists of different computational duties together with multiple particle-number or multiplicity sectors, each represented by its own energy scale

$$
k_{e,n}.
$$

The individual field contributions are therefore

$$
\boxed{
\frac{mv^2}{2k_{e,n}}.
}
$$

In the Taylor expansion, the hidden structure consists of the successive powers and coefficients of the binomial series.

Within the proposed correspondence, each multiplicity-dependent field term may be associated with a Taylor-series sector according to

$$
\boxed{
\frac{mv^2}{2k_{e,n}}
=
a_n
\left(
\frac{v^2}{c^2}
\right)^n
}
$$

where

$$
\boxed{
a_n
=
\frac{\binom{2n}{n}}{4^n}.
}
$$

Thus, each \(n\)-term has both its own multiplicity-dependent energy scale \(k_{e,n}\) and its corresponding Taylor coefficient \(a_n\).

These are not yet mathematically identical derivations. The multiplicity partition is an interpretive construction within Computational Relativity, whereas the Taylor-series cancellation follows exactly from the binomial expansion of the Lorentz factor. Nevertheless, the structural similarity provides a possible correspondence between the individual \(k_{e,n}\) field contributions and the distinct coefficients and powers appearing in the Lorentz series.

Under such an interpretation, the apparently simple expression

$$
\boxed{
1-\frac{v^2}{c^2}
}
$$

would represent the collapsed form of a considerably richer internal structure containing multiple sector-dependent contributions,

$$
\boxed{
\left\{
\frac{mv^2}{2k_{e,n}}
\right\}_{n=1}^{\infty}.
}
$$
