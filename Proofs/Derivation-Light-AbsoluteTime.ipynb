### Corollary 2: How Lorentz Obtained Correct Results Without an Absolute Reference Frame

Lorentz derived the correct time-dilation equations without invoking absolute time or a computational field. This is not a coincidence. Light runs at a fixed rate in absolute time (one bib-rate), so it acts as a direct manifestation of absolute time, since the two tick at the same rate. Any derivation that treats light as invariant therefore tracks absolute time implicitly, and reaches correct conclusions about dilation without ever naming it.

**Absolute time.** Let $db$ be the number of bibs elapsed in the absolute reference frame. Light is not dilated and stays in sync with absolute time, so its proper time is

$$
d\tau_{\text{light}} \approx db.
$$

**Light as a synchronization tool.** Because light's rate is the same in every reference frame and is unaffected by motion, it serves as a common standard. An observer can find a system's dilation factor $\rho$ by comparing its local matter clock against light:

$$
\frac{d\tau_{\text{matter}}}{d\tau_{\text{light}}} = \frac{\rho\,db}{db} = \rho.
$$

**The train thought experiment.** The ground observer sees the light follow a longer path while still travelling at the same speed. This yields the Lorentz relation

$$
p = \frac{v^2}{u^2} \in [0,1], 
\qquad
\rho = \sqrt{1 - p} = \sqrt{1 - \frac{v^2}{u^2}}.
$$

Here $p$ is the **processing factor**: the fraction of absolute computational resources diverted into motion rather than local ticking. The remaining fraction, $\rho$, goes to local ticking. The classical derivation therefore relies on absolute time indirectly, through the invariance of light, to reach the same result.
