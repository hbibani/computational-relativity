## Proposition 1: Absolute Reference Frame and Computational Relativity

Within this framework, an absolute reference frame is proposed as a fundamental requirement for the execution of computational processes, providing a common reference through which interactions between processes can be coordinated and synchronized. Secondly, a computational space is required in which computation can occur, allowing organized computational structures to exist independently of the physical structures observed in the material world. Within the framework of Computational Relativity, the material world alone is proposed to be insufficient to establish the organized conditions necessary for life to emerge through random processes. Consequently, computational space is proposed to provide the underlying structure required for the emergence of life, consistent with the principles established in Computational Relativity.

In Computational Relativity (CR), the central interpretation is that time dilation arises because a computational process requires additional time to complete relative to an absolute reference frame. In Computational Relativity (CR), we introduce a **spectrum of dilation terms**, each distinguished by its **internal form rather than its magnitude alone**. Conceptually, these dilation terms differ according to the types of computational processes they perform. Within this framework, every time a particle transitions from one position to another, an extensive computational process is proposed to occur, altering the particle's internal structure. Thus, the distinction between dilation terms lies not merely in their numerical values, but in the nature of the computations they represent and the resulting transformations they produce within the particle.

We reinterpret Lorentz’s results using a computational-field view in which only an absolute reference frame is truly inertial. Within this framework, the absolute reference frame is defined by the following premises:

1. the absolute reference frame is one which is non-moving, and/or infinitely far away from a gravitational field
2. no dilation occurs in the absolute reference frame
3. everyone agrees on the same amount of time that elapsed relative to the absolute reference frame
4. the greatest amount of time elapses in the absolute reference frame



Let $k_B=\frac{B}{S}$ be the bibs-rate constant (units: bibs/s), representing a system-specific computational overhead that scales how proper time is counted in bibs. The bibs elapsed over a proper-time interval $\Delta t$ is

$$
\boxed{\Delta b=k_b\gamma \Delta t}
$$

Where 

$$ \gamma = \frac{1}{\sqrt{1 - \frac{mv^2}{mc^2}}} $$



Within the framework of Computational Relativity, absolute time is proposed as a hidden variable underlying the Lorentz transformations.

### Scholium 1: Absolute Time the classical definition

Newton's definition establishes three important properties with absolute time:

1. Uniformity: Absolute time progresses at a constant rate.
2. Independence: Its progression is independent of physical motion or external phenomena.
3. Universality: The same absolute time applies throughout the universe, irrespective of an observer's motion.

Classically, absolute time was defined as a universal measure of time that does not undergo dilation, with all observers agreeing on the amount of time elapsed, irrespective of their respective frames of reference.

Within the framework of **Computational Relativity**, this classical conception corresponds to the interpretation of the formula presented above. Specifically, it describes a reference frame in which no time dilation occurs and relative to which all observers can agree upon a common measure of elapsed time.

### Corollary 1: Interpreting Process Requirements


Within Computational Relativity, dilation occurs because a process is taking too long to complete, the term

$$
1-\frac{mv^2}{mc^2}
=
\frac{mc^2}{mc^2}-\frac{mv^2}{mc^2}

represents the proportion of computational capacity remaining available to the system as its velocity increases. The term may also be interpreted as a measure of computational complexity, representing the amount of physical change that can be produced through space and time for a given amount of energy. As this capacity decreases, more energy is required to produce or maintain the same degree of motion or change.

For example, the increased energy required to maintain the same velocity can be seen in the following comparison:

$$

$$

Therefore, additional energy is required to compensate for this reduction and preserve the same level of complexity in terms of motion. Furthermore, when the available energy is partitioned, the amount of physical change that can be produced through space and time is correspondingly reduced.

### Corollary 2: The Stationary Observer and Absolute Time

In the train thought experiment, the stationary observer is used as the reference for measuring the motion of the train.

Within Computational Relativity, this stationary frame represents **absolute time**, because it is the frame in which no motion dilation occurs.

If $$db$$ is the amount of absolute time that passes, then for the stationary observer,

$$
d\tau_{\mathrm{stationary}} = db.
$$

For the moving observer,

$$
d\tau_{\mathrm{moving}} = \rho\,db.
$$

Therefore,

$$
\frac{d\tau_{\mathrm{moving}}}{d\tau_{\mathrm{stationary}}}
=
\rho.
$$

From the train thought experiment,

$$
\rho
=
\sqrt{1-\frac{mv^2}{mc^2}}.
$$

Thus,

$$
d\tau_{\mathrm{moving}}
=
\sqrt{1-\frac{mv^2}{mc^2}}\,db.
$$

Equivalently,

$$
\gamma
=
\frac{1}{\sqrt{1-\frac{mv^2}{mc^2}}}.
$$

In this interpretation, the stationary observer acts as the reference for **absolute time**, while the moving observer is dilated relative to it.

### Scholium 3: Interpreting Time Dilation

**Time dilation arises when the internal computational process governing a system requires a longer interval to complete. As the duration of the internal process increases, a greater energy input is required to produce the same external change in motion.**

Equivalently, if an object's internal clock is identified with the completion rate of its underlying computational processes, then a slowing of that clock corresponds to an increased computational cost of changing the object's state of motion. Thus,

$$
\boxed{
\text{greater computational duration}
\;\Longrightarrow\;
\text{greater time dilation}
\;\Longrightarrow\;
\text{greater energy required for further motion}
}
$$

Within Computational Relativity, time dilation is therefore treated not merely as an observed change in elapsed time, but as a consequence of an underlying process requiring progressively greater computational capacity to complete.

### Corollary 5: Computational Space and the Absolute Reference Frame

The primary requirement of the absolute reference frame is that it provides a space in which computation can occur. In `Experiment.ipynb`, it was proposed that the observed phenomena could not occur without some underlying computational process.

There are two possible mechanisms through which such computation could occur:

1. Through the absolute reference frame, acting as an underlying computational space.
2. Through an unknown mechanism associated with the biological structure of the human body.

We exclude the second possibility on the basis of the conditions required for biological organisms to originate and develop within initially unstructured or random environments. If the computational mechanism were entirely dependent upon an already-existing biological structure, then it could not explain the processes responsible for the emergence of that biological structure in the first place. Before an organism existed, there would be no biological system to which the proposed mechanism could be attributed.

Furthermore, the work presented in `Evolution.ipynb` argues that the events considered there cannot be adequately produced under the specified environmental conditions without some form of underlying computational field or information-processing structure.

`Experiment.ipynb` additionally produced a mathematical expression intended to describe how such events could occur. Importantly, the resulting formulation exhibits a close mathematical relationship with the Boltzmann distribution, suggesting that the proposed computational mechanism may be expressible through established statistical-mechanical principles.

Under this interpretation, the absolute reference frame is therefore not merely a reference for absolute time or motion. It also provides the underlying computational space within which the required probabilistic and informational processes can occur.
