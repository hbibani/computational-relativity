## Proposition 5: The Boltzmann distribution describes equilibrium probabilities

The Boltzmann distribution describes a small system, which may be as small as a single particle, in thermal contact with a large heat reservoir with which it exchanges energy. It gives the probability of finding the small system in a state of energy $E$ as $P(E)\propto e^{-E/k_BT}$, where $T$ is the temperature of the reservoir.

</p

For a microscopic system in thermal equilibrium with a heat bath,

$$
p_i=\frac{e^{-E_i/(k_B T)}}{Z},
$$

where

$$
Z=\sum_i e^{-E_i/(k_B T)}.
$$

The small system described by the Boltzmann distribution can occupy any of several states, each with its own energy $E_i$ and an associated probability. As the diagram shows, the system may be found in any one of these states.

</p


The interaction with the heat bath is represented implicitly through the equilibrium temperature (T). The distribution does not specify how energy is exchanged or how (E_i) is physically composed. 

Without an additional multiplicity factor,

$$
p_i\propto e^{-E_i/(k_B T)},
$$

### Corollary 1: Separating Boltzmann Distribution into two sections: probability and partitioning

The Boltzmann distribution gives the relative likelihood of a system's equilibrium energy states. It says nothing about where that energy comes from, how it is structured, or how it is partitioned within each state. Students often find the distribution difficult because two distinct kinds of dynamics can be at play:

1. **Occupation of a state:** the probability that the system is found in a state of energy $E_i$, without reference to how that energy is partitioned internally. This is governed by the Boltzmann factor,

$$
p_i \propto e^{-E_i/(k_BT)}.
$$

2. **Restructuring of a state:** a change that alters the internal dynamics of the system's components and, in doing so, also changes the likelihood of the state. This is governed by a multiplicative factor applied to the Boltzmann factor, either $\gamma$ or $\Omega$:

$$
p_i \propto \gamma\, e^{-E_i/(k_BT)}
\qquad\text{or}\qquad
p_i \propto \Omega\, e^{-E_i/(k_BT)},
$$

where $\gamma$ corresponds to **biological partitioning** and $\Omega$ to **random partitioning**.

### Corollary 2: Computational Fields reduce the computational time to produce specific states

The relationship between the Boltzmann distribution and relativity in a multiparticle system can be expressed through probability dilation. When probability dilation is represented as a computational field, the multiplicity provides the corresponding representation of that field within the system. The presence of this multiplicity changes the rate at which the system generates its energetic configuration as it approaches equilibrium. Consequently, the equilibrium state can be reached over a reduced elapsed time relative to the corresponding system in the absence of the multiplicity field.

Computational relativity describes thermodynamics in a computational manner, with its terminology identifying conceptual relationships between structures in thermodynamics and corresponding structures in computational relativity. $$
\boxed{
Z
=
\sum_i e^{-E_i/(k_B T)}
}
$$

The **Helmholtz free energy** is then defined as

$$
\boxed{
F
=
-k_B T\ln Z
}
$$

Within the interpretation of computational relativity, free energy is defined in relation to a range of possible states. Some of these states may never come into existence, which implies that free energy is logically tied to states that are almost imaginary. States that are not presently realized can still contribute to the description of the current state.

This can only be understood through a computational framework: an unrealized state could contribute to the current state if it were used as a computational input. The partition function (Z) and its relation to free energy should therefore have suggested to those who derived it that we are in a computer or this could only possibly happen within a computational system. This matters because it has further implications within computational relativity itself. Not every structure in thermodynamics can be translated mathematically into the corresponding terms of computational relativity, but conceptual bridges can still be built between the two frameworks, and these can help guide the theory's continued development.

For example, the relation between the partition function and free energy can be read as showing that states which have not come into existence, including states of the past and future, can still contribute to the description of the current state of a particle.
