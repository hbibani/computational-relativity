# Appendix 2 — The K–e Bridge and Relativistic Kinetic Energy

This appendix presents the argument in a single forward sequence:

1. **Corollary 2:** derive \(K=mc^2(\gamma-1)\) using the K–e bridge.
2. **Corollary 3:** rewrite \(K\) as expressions proportional to \(mv'^2\) or \(mv^2\).
3. **Corollary 4:** express the denominator and its reciprocal as Taylor series.
4. **Numerical verification:** check that all three kinetic-energy formulas agree.

The supporting algebraic steps are developed separately in **Appendix 1**.

**Definitions.** For a particle with constant rest mass \(m\),

$$
\gamma=\frac{1}{\sqrt{1-v^2/c^2}},\qquad
v'=\gamma v,\qquad
p=mv'=\gamma mv,\qquad
|v|<c.
$$

Here \(v\) is coordinate velocity, \(v'\) is proper velocity, and \(K\) is the relativistic kinetic energy. The quantity \(\tfrac12mv'^2\) is used as an **auxiliary expression**, not as the physical kinetic energy \(K\).


## Corollary 2 — Deriving relativistic kinetic energy through the K–e bridge


### Step 1. Express the auxiliary term using \(\gamma\)

Because \(v'=\gamma v\),

$$
\frac12mv'^2=\frac12m\gamma^2v^2
=\frac{\frac12mv^2}{1-v^2/c^2}.
$$

Insert \(c^2/c^2\) and use the identity from Appendix 1, \(\gamma^2 v^2/c^2=\gamma^2-1\):

$$
\begin{aligned}
\frac12mv'^2
&=\frac12mc^2\left(\gamma^2\frac{v^2}{c^2}\right)\\[4pt]
&=\frac12mc^2(\gamma^2-1).
\end{aligned}
$$


### Step 2. Differentiate

With \(m\) and \(c\) constant,

$$
d\left(\frac12mv'^2\right)
=\frac12mc^2\,d(\gamma^2)
=mc^2\gamma\,d\gamma.
$$


### Step 3. Introduce the work–energy theorem

The relativistic work–energy theorem is

$$
dK=F\,dx=v\,dp.
$$

Since \(p=mv'\), the identity established in Appendix 1 gives

$$
\begin{aligned}
d\left(\frac12mv'^2\right)
&=v'\,dp\\
&=\gamma v\,dp\\
&=\gamma\,dK.
\end{aligned}
$$


### Step 4. Equate the two differentials

$$
\gamma\,dK=mc^2\gamma\,d\gamma.
$$

As \(\gamma>0\), cancel \(\gamma\):

$$
dK=mc^2\,d\gamma.
$$


### Step 5. Integrate from rest

At rest, \(\gamma=1\) and \(K=0\). Therefore,

$$
\begin{aligned}
K&=\int_{1}^{\gamma}mc^2\,d\gamma\\[4pt]
&=mc^2(\gamma-1).
\end{aligned}
$$

Thus,

$$
K=mc^2(\gamma-1).
$$


### Step 6. Recover the total-energy expression

Using the standard special-relativistic rest-energy relation \(E_0=mc^2\),

$$
E=E_0+K=mc^2+mc^2(\gamma-1),
$$

so

$$
E=\gamma mc^2.
$$

**Scope of the derivation:** Steps 1–5 derive the kinetic-energy expression from relativistic momentum and work. The final total-energy expression also uses the rest-energy relation \(E_0=mc^2\); it is **not** independently established by the preceding algebra.


## Corollary 3 — Equivalent expressions in \(mv'^2\) and \(mv^2\)

From Corollary 2,

$$
\frac12mv'^2=\frac12mc^2(\gamma^2-1).
$$

Factor the difference of squares:

$$
\frac12mv'^2=\frac12mc^2(\gamma-1)(\gamma+1).
$$

Since \(K=mc^2(\gamma-1)\),

$$
\frac12mv'^2=\frac{\gamma+1}{2}K.
$$

Rearranging yields

$$
K=\frac{mv'^2}{\gamma+1}.
$$

Substitute \(v'=\gamma v\), then divide numerator and denominator by \(\gamma^2\):

$$
\begin{aligned}
K
&=\frac{\gamma^2mv^2}{\gamma+1}\\[5pt]
&=\frac{mv^2}{\gamma^{-1}+\gamma^{-2}}.
\end{aligned}
$$

All three forms are equivalent:

$$

K=mc^2(\gamma-1)
=\frac{mv'^2}{\gamma+1}
=\frac{\gamma^2mv^2}{\gamma+1}
=\frac{mv^2}{\gamma^{-1}+\gamma^{-2}}.

$$

The expression \(\tfrac12mv'^2\) differs from \(K\) by a factor of \((\gamma+1)/2\). Equivalently,

$$
K=\frac{2}{\gamma+1}\left(\frac12mv'^2\right).
$$

As \(v\to0\), \(\gamma\to1\) and this conversion factor approaches 1; both expressions then approach the classical kinetic energy \(\tfrac12mv^2\).


## Corollary 4 — Taylor-series representation of the kinetic-energy factor

The preceding result can be written as

$$
K=mv^2\left(\frac{1}{\gamma^{-1}+\gamma^{-2}}\right).
$$

We now represent the dimensionless factor in parentheses through a Taylor series.


### Step 1. Introduce the normalized ratio and the field-sector notation

Let

$$
x=\frac{mv^2}{mc^2}=\frac{v^2}{c^2},\qquad 0\le x<1.
$$

In the proposed Computational Relativity notation, define

$$
F(x)=1-\sqrt{1-x}=\sum_{n=1}^{\infty}f_n(x),
$$

where

$$
f_n(x)=\frac{\binom{2n}{n}}{4^n(2n-1)}x^n.
$$

The first terms are

$$
F(x)=\frac12x+\frac18x^2+\frac1{16}x^3+\frac5{128}x^4+\cdots.
$$


### Step 2. Substitute into the denominator

Because

$$
\gamma^{-1}=\sqrt{1-x}=1-F(x),\qquad \gamma^{-2}=1-x,
$$

we obtain

$$
\begin{aligned}
\gamma^{-1}+\gamma^{-2}
&=(1-F(x))+(1-x)\\[4pt]
&=2-x-\sum_{n=1}^{\infty}f_n(x).
\end{aligned}
$$

Therefore,

$$
K=
\frac{mv^2}{\displaystyle
2-\frac{mv^2}{mc^2}
-\sum_{n=1}^{\infty}
\frac{\binom{2n}{n}}{4^n(2n-1)}
\left(\frac{mv^2}{mc^2}\right)^n}.
$$

Expanding the denominator gives

$$
K=\frac{mv^2}{\displaystyle 2-\frac32x-\frac18x^2-\frac1{16}x^3-\frac5{128}x^4-\cdots}.
$$


### Step 3. Expand the reciprocal directly

For \(x>0\), use \(K=mc^2(\gamma-1)\) and \(mv^2=mc^2x\):

$$
\frac{1}{\gamma^{-1}+\gamma^{-2}}=\frac{\gamma-1}{x}.
$$

The Taylor series of the Lorentz factor is

$$
\gamma=(1-x)^{-1/2}
=\sum_{n=0}^{\infty}\frac{\binom{2n}{n}}{4^n}x^n.
$$

Subtract the constant term and divide by \(x\):

$$
\frac{\gamma-1}{x}
=\sum_{n=0}^{\infty}
\frac{\binom{2n+2}{n+1}}{4^{n+1}}x^n.
$$

Thus, the **general Taylor-series coefficient** after factoring out \(mv^2\) is

$$
a_n=\frac{\binom{2n+2}{n+1}}{4^{n+1}},\qquad n=0,1,2,\ldots
$$

and the reciprocal factor expands as

$$
\frac{1}{\gamma^{-1}+\gamma^{-2}}
=\frac12+\frac38x+\frac5{16}x^2+\frac{35}{128}x^3+\frac{63}{256}x^4+\cdots.
$$

At \(x=0\), the expression \((\gamma-1)/x\) is interpreted by its continuous limit \(1/2\).


### Step 4. Write the final expression with \(mv^2\) outside

$$

K=mv^2\sum_{n=0}^{\infty}
\frac{\binom{2n+2}{n+1}}{4^{n+1}}
\left(\frac{mv^2}{mc^2}\right)^n

$$

or, expanded,

$$

K=mv^2\left[
\frac12
+\frac38\left(\frac{mv^2}{mc^2}\right)
+\frac5{16}\left(\frac{mv^2}{mc^2}\right)^2
+\frac{35}{128}\left(\frac{mv^2}{mc^2}\right)^3
+\cdots
\right].

$$

**Interpretation.** Mathematically, both the subtracted-denominator series and the positive reciprocal series represent the same relativistic kinetic-energy function. In the proposed Computational Relativity framework, the terms \(f_n(x)\) are called *field-sector contributions*. Their identification with physical particle sectors is a **hypothesis**, not a consequence of the Taylor expansion itself.

## Summary of the derivation

The chain of results is

$$
\underbrace{\frac12mv'^2=\frac12mc^2(\gamma^2-1)}_{\text{algebra}}
\quad\xrightarrow{\ d\ }
\quad mc^2\gamma\,d\gamma=\gamma\,dK
\quad\xrightarrow{\ \text{integrate}\ }
\quad K=mc^2(\gamma-1).
$$

This leads to the equivalent representations

$$
K=\frac{mv'^2}{\gamma+1}
=\frac{mv^2}{\gamma^{-1}+\gamma^{-2}}
=mv^2\sum_{n=0}^{\infty}\frac{\binom{2n+2}{n+1}}{4^{n+1}}
\left(\frac{mv^2}{mc^2}\right)^n.
$$

Together with the standard rest-energy relation \(E_0=mc^2\), the same kinetic-energy result gives \(E=\gamma mc^2\).
