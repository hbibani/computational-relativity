## Proposition 2: Energy can be partitioned by a computational field

In computational relativity, energy is partitioned within a framework that draws upon both thermodynamics and standard relativity, while modifying their interpretation through the underlying philosophy of computational relativity. Depending on the system used to describe probability dilation, the state of an individual particle may be determined in relation to a broader multi-particle system. As the system approaches equilibrium, its available energy is distributed among its possible states according to the structure of that system.

In this proposition, we introduce the principle that a system may partition its energy in different ways depending on the nature of the field acting upon it. The form of the energy distribution is therefore not treated as universally identical across all systems, but as dependent upon the physical and computational structure of the relevant field.

Within computational relativity, this structure is connected to standard relativity through the Taylor-series expansion of the relativistic dilation term. The individual terms of the series are interpreted as representing distinct contributions within a multi-particle system, providing a mathematical framework through which the partitioning of energy among different states can be described.


The kinetic energy of a moving body is

$$
E_k=\frac{1}{2}mv^2.
$$

This energy may be transferred or partitioned among different physical processes:

$$
E=E_1+E_2+\cdots+E_n.
$$

### Corollary 1

Energy may be transferred as work:

$$
W=\int \mathbf{F}\cdot d\mathbf{x}.
$$

When applied to another body, this work may change its kinetic energy, momentum, or trajectory:

$$
\Delta E_k=W.
$$

Thus, the same total energy may produce different physical effects.
