### Corollary 8: Velocity Dilation Derivation and Thought Experiment



In this section, we formally derive velocity dilation using a direct mathematical methodology, distinct from the method of postulation used to establish the terms introduced in Proposition 3, Corollary 1. As we can see the skeleton below:


$$
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}\left(\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}\right)mv'^2
$$

can be used to derive the following terms:

$$v'=\frac{v}{\sqrt{1-\frac{v^2}{c^2}}}$$

However, a distinct methodology is required to independently arrive at the same conclusion. Below is the methodology described through a thought experiment:

Consider a train moving horizontally at velocity

$$
V=0.9c.
$$

Inside the train, an object moves vertically upward with velocity

$$
v.
$$

The velocity (v) is measured in the train’s own reference frame. We want to determine the effective vertical velocity measured by an observer standing on the ground.

We compare two cases:

1. The train is at rest relative to the ground observer.
2. The train moves horizontally at (0.9c).

A particle moving perpendicular to the horizontal motion will have its observed perpendicular velocity contracted according to the following relation:

$$
v=\frac{s}{t},
\qquad
t'=\frac{t}{\sqrt{1-\frac{v^2}{c^2}}}
$$

$$
v=\frac{s}{t'}
=\frac{s}{t}\sqrt{1-\frac{v^2}{c^2}}
$$

$$
v
=v'\sqrt{1-\frac{v^2}{c^2}}
$$


If the **observed perpendicular velocity** is set equal to the train velocity (V), then

$$
v=v'\sqrt{1-\frac{v^2}{c^2}}=V
$$

Therefore,

$$
\boxed{
v'=\frac{v}{\sqrt{1-\frac{v^2}{c^2}}}
}
$$
