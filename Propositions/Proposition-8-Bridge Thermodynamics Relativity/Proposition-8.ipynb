## Proposition 8: Energy–temperature bridge

The computational-relativity model begins with the energy-allocation relation

$$
K'=K-e.
$$

This expands into the following form:

$$
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}\left(\frac{mv^2}{2k_e}\right)\left(\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}\right)mv'^2
$$


For relativistic motion, the dilation term

$$
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
$$

is more physically explicit than the reduced form because Galilean relativity describes motion kinematically while omitting the underlying physical causes and energetic structure of that motion.

$$
\frac{v^2}{c^2},
$$

because it preserves the underlying energy ratio. This provides a direct bridge to thermodynamics, where relations must be expressed in compatible energy units. By contrast, the reduced velocity ratio omits the mass and momentum structure required for this mapping.

Which relates to the equipartition theorem in thermodynamics

$$
\frac{3}{2}N k_B T
=
\frac{3}{2}N k_B T'
-
\left(
\frac{k_B T}{k_e}
\right)
\frac{3}{2}N k_B T',
$$


Therefore, the temperature ratio for probability dilation is

$$
T'
=
\frac{T}
{
1 - \frac{k_B T}{k_e}
}.
$$

and the temperature ratio for dilation of motion is

$$
T'
=
\frac{T}
{
1 - \frac{k_B T}{k_m}
}.
$$

### Corollary 2

The methodology used in Proposition 7 produces a temperature-dilation factor through the thermodynamic relation

$$
\Delta E=-k_B T\ln\gamma,
$$

whereas this proposition derives the same general energy-allocation structure by converting motion and probability factors into temperature ratios.

The two propositions therefore reach compatible dilation structures through different mechanisms: one through thermodynamic postulation and the other through an energy–temperature mapping.
