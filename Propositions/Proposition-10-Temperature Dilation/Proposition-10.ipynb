## Proposition 10: Temperature Dilation

The first attempt to construct a temperature-dilation estimate was developed in the preceding propositions on probability dilation. In that earlier analysis, we identified a relationship between the enumeration of particles, or multiplicity, and the dilation term itself. Within Computational Relativity, this was interpreted as suggesting that entropy can arise from the action of a computational field operating across a multi-particle system.

This interpretation also led to a remarkably close estimate of the temperature-dilation relation. The estimate was obtained by returning to the probability relation originally derived from `experiment.ipynb`, connecting it to the Boltzmann distribution, and then examining the resulting energy-dependent expression. From this relationship, the corresponding temperature-dilation structure was inferred, providing the initial form that was later reproduced more directly through the \(K-e\) energy-allocation framework.

Assume the observed probability is dilated by a factor $\gamma$:

$$
p_{\text{obs}} = \gamma\, p_{\text{exp}}.
$$

Substitute the Boltzmann form into the dilation:

$$
p_{\text{obs}} \propto \gamma\, e^{-E_{\text{exp}}/(k_B T)}.
$$


This derives to the following formula, and using the method of postulation, guided by the energy relation

$$
\Delta E=-k_B T\ln\gamma,
$$

the corresponding temperature-dilation factor is estimated as, using the derivative

$$
\boxed{
\gamma_d
\approx
\frac{1}
{1-\frac{k_B T}{k_e\gamma}}
}.
$$

This factor describes the proposed dilation of temperature produced by the combined influence of thermal energy and probability dilation.

### SCHOLIUM 1: ANALYZING THE INTERPRETATION AND MEANING OF TEMPERATURE DILATION

The important point is that the **temperature scale itself does not dilate**. Instead, the dilation appears in the amount of energy required to produce the same transition on that scale. Thus, if two systems are brought to the same temperature, both systems still register the same temperature value, but a dilated system requires a greater energy input to reach that state. In this sense, the physical scale remains fixed while the **energy cost associated with moving through the scale increases**.

The same interpretation can be applied to momentum. The dilation of energy should not be understood as changing the spatial or momentum scale itself. Rather, it means that **more energy is required to produce the same change in momentum or the same physical transition**. Therefore, two systems may exhibit the same resulting motion or occupy the same point on a temperature scale while differing in the amount of energy required to produce that state.

Symbolically, you could represent the distinction as

$$
T_1=T_2=T
$$

while

$$
E_2>E_1.
$$

If the second system has a dilation factor $\gamma$, then schematically,

$$
E_2=\gamma E_1,
$$

even though

$$
T_2=T_1.
$$

So the central principle becomes:

$$
\boxed{
\text{dilation changes the energy required per unit transition, not the scale itself}
}
$$

This is a more precise interpretation than saying that temperature itself is dilated.

### SCHOLIUM 3: Equipartition Theorem Derivation


The earlier probability-dilation analysis related the dilation factor to energy through

$$
\Delta E=-k_BT\ln\gamma,
$$

which has the same logarithmic structure as the statistical-mechanical multiplicity relation

$$
S=k_B\ln\Omega.
$$

Within Computational Relativity, this motivated the interpretation that \(\gamma\) carries a multiplicity contribution. The initial thermodynamic processing term was therefore written as

$$
\frac{k_BT}{k_e\gamma}.
$$

When the same system is expressed through equipartition,

$$
K=\frac{3}{2}Nk_BT,
$$

the corresponding normalized energy ratio becomes

$$
\boxed{
\frac{\frac{3}{2}Nk_BT}{\frac{3}{2}Nk_e}
=
\frac{k_BT}{k_e}.
}
$$

The many-particle content is now represented explicitly through \(N\), so the multiplicity contribution that had previously been carried implicitly through \(\gamma\) no longer appears as a separate factor in the reduced ratio. Thus,

$$
\boxed{
\frac{k_BT}{k_e\gamma}
\longrightarrow
\frac{k_BT}{k_e}.
}
$$

This has the same energetic structure as the mechanical relation

$$
\boxed{
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
=
\frac{v^2}{c^2},
}
$$

which represents the computational requirement for motion. The complementary remaining capacity is therefore

$$
\boxed{
1-\frac{v^2}{c^2}.
}
$$

It is important to distinguish multiplicity \(\Omega\) from particle number \(N\): they are not mathematically identical. The claim here is that, within the proposed mapping, the multiplicity contribution is re-expressed through the explicit many-particle energy formulation rather than remaining as a separate \(\gamma\) factor.
