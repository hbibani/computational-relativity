# Proposition 11: Transition Systems, Multiplicity, and the Production of Observed States

A transition system is a mathematical model describing how a system evolves from one state to another through a set of available actions.

In Computational Relativity, we propose using transition systems to examine the underlying processes responsible for producing the physical states encountered by observers.

The central idea is that **an observed state is the result of preceding transitions, while a sequence of observed states provides information about how the system evolves over time**.

This distinction becomes particularly important when two observers occupy approximately the same spatial region during the same period but report substantially different distributions of events.

Formally, a transition system is defined as

$$
TS=(S,Act,\rightarrow,I,AP,L)
$$

where:

- $S$ is the set of possible states of the system.
- $Act$ is the set of available actions capable of changing the system.
- $\rightarrow$ is the transition relation describing how actions move the system between states.
- $I$ is the set of initial states.
- $AP$ is the set of atomic propositions describing properties of the states.
- $L$ is the labelling function assigning those properties to states.

---

### Corollary 1: Observer-Dependent Atomic Propositions and the Multiplicity of Transition Systems

Within Computational Relativity, we propose that two observers in approximately the same spatial region, experiencing the same elapsed time, may report different sequences of observed states, even when they observe the same category of physical events.

A transition system represents how a physical system evolves through states and actions. In the vehicle experiment, the states are the physical configurations of the vehicles, and the actions are changes in those configurations, such as movement to the left, movement to the right, acceleration, or a change of position.

We define the transition system as

$$
TS=(S,Act,\rightarrow,I,AP,L)
$$

where the set of atomic propositions, $AP$, describes properties that may be true of the physical states the system passes through.

where:
- $S$ represents the possible physical states of the system which is the human body in this case
- $Act$ represents the actions performed by the human body after reaching a state, such as moving left, moving right, accelerating, or stopping.
- $\rightarrow$ represents the transitions between physical states resulting from those actions.
- $I$ represents the initial states.
- $AP$ represents the atomic propositions describing properties of the physical states of the human body.
- $L$ represents the labelling function that determines which atomic propositions are true in each state.

Consider two observers, $A$ and $B$, with the observational transition systems

$$
TS_A=(S_A,Act_A,\rightarrow_A,I_A,AP_A,L_A)
$$

$$
TS_B=(S_B,Act_B,\rightarrow_B,I_B,AP_B,L_B)
$$

We introduce two distinct collections of atomic propositions:

$$
AP_A=X,
\qquad
AP_B=Z.
$$

Here, $X$ and $Z$ are different collections of propositions used to characterise the observed states.

---


### Corollary 2: Different Atomic Propositions Correspond to Different Action Sequences

The collections $X$ and $Z$ describe completely different types of states. A state that satisfies a proposition in $Z$ is effectively unreachable when the system follows the dynamics described by $X$, and the reverse is also true.

$$
s\models X \quad\text{(binomial dynamics)},
\qquad
s\models Z \quad\text{(DA-prefix dynamics)}.
$$

The two types of state are reached through different sequences of actions.

Let the actions be

$$
Act=\{R,\;L\},
$$

where $R$ is a movement to the right and $L$ is a movement to the left. A path through the transition system is a sequence of these actions.

**Observer A ($X$, binomial dynamics).** The actions are mixed, with no long runs:

$$
\sigma_A = R\,L\,R\,L\,R\,L\,R\,L
$$

**Observer B ($Z$, DA-prefix dynamics).** The actions are grouped into long runs:

$$
\sigma_B = R\,R\,R\,R\,L\,L\,L\,L
$$

Both sequences contain the same number of each action (four $R$ and four $L$). There are

$$
\Omega=\binom{8}{4}=70
$$

sequences of this kind, and only 2 of them are fully grouped:

$$
R\,R\,R\,R\,L\,L\,L\,L
\qquad\text{and}\qquad
L\,L\,L\,L\,R\,R\,R\,R.
$$

Under binomial dynamics every sequence has the same probability, so a grouped sequence occurs with probability

$$
\frac{2}{70}\approx0.029.
$$

As the number of actions grows, this fraction quickly approaches zero, and mixed sequences such as $\sigma_A$ dominate. A system that repeatedly produces grouped sequences such as $\sigma_B$ is therefore not following binomial dynamics. Its transitions are shaped by a different multiplicity form, $\gamma$.

$$
X:\ \text{mixed action sequences}\ \longrightarrow\ \text{binomial distribution}
$$

$$
Z:\ \text{grouped action sequences}\ \longrightarrow\ \text{DA-prefix states}
$$

The two observers therefore differ not only in the states they report but in the sequences of actions that lead to those states.
