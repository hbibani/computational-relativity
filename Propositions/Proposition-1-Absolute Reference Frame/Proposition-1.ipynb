## Proposition 1: Absolute Reference Frame and Computational Relativity

We reinterpret Lorentz’s results using a computational-field view in which only an absolute reference frame is truly inertial. Within this framework, the absolute reference frame is defined by the following premises:

1. the absolute reference frame is one which is non-moving, or infinitely far away from a gravitational field
2. everyone agrees on the same amount of time that elapsed relative to the absolute reference frame
3. the greatest amount of time elapses in the absolute reference frame

Let $k_B=\frac{B}{S}$ be the bibs-rate constant (units: bibs/s), representing a system-specific computational overhead that scales how proper time is counted in bibs. The bibs elapsed over a proper-time interval $\Delta\tau$ is

$$
\boxed{\Delta b=\frac{k_B}{\rho}\,\Delta\tau}
$$

Where 

$$ \rho = \sqrt{1 - p} = \sqrt{1 - \frac{v^2}{u^2}} $$

### Corollary 1: Interpreting Process Requirements


Within Computational Relativity, dilation occurs because a process is taking too long to complete, the term

$$
1-\frac{v^2}{c^2}
=
\frac{c^2}{c^2}-\frac{v^2}{c^2}
$$

represents the proportion of computational capacity remaining available to the system as its velocity increases. The term may also be interpreted as a measure of computational complexity, representing the amount of physical change that can be produced through space and time for a given amount of energy. As this capacity decreases, more energy is required to produce or maintain the same degree of motion or change.

For example, the increased energy required to maintain the same velocity can be seen in the following comparison:

$$
0.9c \text{ without dilation}
\longrightarrow
3.64\times10^{16}\mathrm{J}
$$

$$
0.9c \text{ with dilation}
\longrightarrow
1.16\times10^{17}\mathrm{J}
$$

Therefore, additional energy is required to compensate for this reduction and preserve the same level of complexity in terms of motion. Furthermore, when the available energy is partitioned, the amount of physical change that can be produced through space and time is correspondingly reduced.
