## Proposition 6: Energy may be statistically partitioned by different forms of multiplicity

The statistical weight of state (i) is

$$
w_i=\Omega_i e^{-E_i/(k_B T)}.
$$

Here, $\Omega_i$ represents the multiplicity associated with the entropy of state (i). The multiplicity factor is defined as a representation of the computational effects of a field on the probabilistic dynamics and effective energy structure of a system. Without the $\Omega_i$ term, the model does not include the additional configurational effects attributed to the computational field. In that case, the expression reduces to the standard Boltzmann distribution, in which state probabilities are determined by the exchange of thermal energy between a microscopic system and a heat reservoir (when reaching equilibrium) [REFER PROPOSITION 4].

Its probability is

$$
p_i=\frac{\Omega_i e^{-E_i/(k_B T)}}{Z},
$$

where

$$
Z=\sum_j \Omega_j e^{-E_j/(k_B T)}.
$$

When discussing multiplicity, we are describing how the rules governing movement generate the dynamics of a system. Multiplicity therefore determines how energy is partitioned across those dynamics (REFER PROPOSITION 1). Two forms of multiplicity are proposed. 


### Multiplicity 1: Random entropic multiplicity

Ordinary thermodynamic multiplicity is denoted by

$$
\Omega_{i,\mathrm{r}},
$$

with entropy

$$
S_{i,\mathrm{r}}=k_B\ln\Omega_{i,\mathrm{r}}.
$$

The statistical weight becomes

$$
w_{i,\mathrm{r}}
=

e^{-\left(E_i-TS_{i,\mathrm{r}}\right)/(k_B T)}.
$$

The term

$$
-TS_{i,\mathrm{r}}
$$

represents the contribution of random microscopic multiplicity.

### Multiplicity 2: Biological multiplicity

For a biological state, a computational field introduces a dilation factor which changes the dynamics of how the energy is partitioned within the thermodynamic system

$$
\gamma_i.
$$

The statistical weight becomes

$$
w_{i,\mathrm{b}}
=

\gamma_i e^{-E_{i,\mathrm{b}}/(k_B T)}.
$$

Equivalently,

$$
w_{i,\mathrm{b}}
=

e^{-\left(E_{i,\mathrm{b}}-k_B T\ln\gamma_i\right)/(k_B T)}.
$$

Therefore, the effective energy is

$$
E_{i,\mathrm{eff}}
=

E_{i,\mathrm{b}}-k_B T\ln\gamma_i.
$$

The term

$$
-k_B T\ln\gamma_i
$$

has the same mathematical form as (-TS), but is proposed to represent ordered biological energy partitioning rather than random microscopic multiplicity.

*** IMPORTANT *** Evidence for the existence of a multiplicity in physical reality that is distinct from $\Omega$, where $\Omega$ primarily represents the combinatorial character of randomness and entropy, is developed from the experiment presented in `experiment.ipynb`. This second form of multiplicity is represented through the $gamma$-multiplicity.

### Corollary 2: Partitioning and Energy Terms

Although the total physical energy of state (i) remains (E_i), the multiplicity of that state introduces an effective statistical energy contribution,

$$
w_i=\Omega_i e^{-(E_i)/(k_B T)}.
$$

Which derives to and includes the term $-k_B T\ln\Omega_i$

$$
w_i = e^{-(E_i-k_B T\ln\Omega_i)/(k_B T)}.
$$

The term below describes an effective “energy reduction.” This should not be interpreted as a reduction in the total physical energy of the system because the total energy of the state still remains $E_i$. Rather, it represents the effect that a computational field has on the statistical weight of a particular state, expressed through an equivalent shift in its effective energy.

$$
-k_B T\ln\Omega_i.
$$

Accordingly, the statistical behaviour of the state can be represented by the effective energy

$$
E_i-k_B T\ln\Omega_i.
$$

Thus, multiplicity does not directly remove energy from (E_i). Instead, it changes how (E_i) is represented, because the total energy is partitioned to include the configurational contribution associated with the computational field. In this description, a portion of the effective energetic contribution is represented by (that is, a portion of $E_i$ includes the energy description below)

$$
k_B T\ln\Omega_i.
$$

### Corollary 3: Energy Terms

The energy terms in the Boltzmann distribution are modified when the multiplicity factors $\gamma$ or $\Omega$ are introduced into the statistical weight. When these factors multiply the Boltzmann probability term, their effect can be represented as an equivalent shift in the effective energy associated with each state, as shown below:


For random multiplicity,

$$
E_{i,\mathrm{eff}}
=
E_i-TS_{i,\mathrm{r}}.
$$

For biological multiplicity,

$$
E_{i,\mathrm{eff}}
=

E_{i,\mathrm{b}}-k_B T\ln\gamma_i.
$$

Thus, $E_{i,\mathrm{b}}$ remains the physical energy of the state, while the dilation term modifies its statistical weight. In both cases $E_i$ is modified to include either $TS_{i,\mathrm{r}}$ or $k_B T\ln\gamma_i$ within the contents of $E_{i,\mathrm{b}}$ depending upon the type of multiplicity.
