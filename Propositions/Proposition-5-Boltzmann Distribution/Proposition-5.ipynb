## Proposition 4: The Boltzmann distribution describes equilibrium probabilities

For a microscopic system in thermal equilibrium with a heat bath,

$$
p_i=\frac{e^{-E_i/(k_B T)}}{Z},
$$

where

$$
Z=\sum_j e^{-E_j/(k_B T)}.
$$

The distribution gives the probability that the system occupies the energy state (E_i).

The interaction with the heat bath is represented implicitly through the equilibrium temperature (T). The distribution does not specify how energy is exchanged or how (E_i) is physically composed. 

Without an additional multiplicity factor,

$$
p_i\propto e^{-E_i/(k_B T)},
$$



### Corollary 1

The Boltzmann distribution describes the relative likelihood of equilibrium energy states, not the origin, structure, or partitioning of the energy within them.

The relationship between the Boltzmann distribution and relativity in a multiparticle system can be expressed through probability dilation. When probability dilation is represented as a computational field, the multiplicity provides the corresponding representation of that field within the system. The presence of this multiplicity changes the rate at which the system generates its energetic configuration as it approaches equilibrium. Consequently, the equilibrium state can be reached over a reduced elapsed time relative to the corresponding system in the absence of the multiplicity field.
