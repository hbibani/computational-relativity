## Proposition 4: The \(K-e\) Energy-Allocation Principle

The starting point of the physical formulation is the \(K-e\) relation

$$
\boxed{
K' = K - e
}
$$

which states that the resulting kinetic state is determined by the available kinetic energy after a portion of that energy has been allocated to another physical or computational requirement.

Within the relativistic formulation, this principle was represented as

$$
\boxed{
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}
\left(
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
\right)
mv'^2
}
$$

The important feature of this expression is that the quantity controlling the subtraction is itself an **energy ratio**:

$$
\boxed{
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}
=
\frac{mv^2}{mc^2}
=
\frac{v^2}{c^2}
}
$$

The factors of \(\frac{1}{2}\) and \(m\) cancel, leaving the familiar normalized relativistic term

$$
\frac{v^2}{c^2}.
$$

Within Computational Relativity, however, this quantity is interpreted not merely as a ratio between the squared velocity \(v^2\) and the limiting squared velocity \(c^2\), but equivalently as the ratio between the kinetic-energy scale associated with the current motion and the corresponding limiting energy scale:

$$
\boxed{
\frac{v^2}{c^2}
\equiv
\frac{\frac{1}{2}mv^2}{\frac{1}{2}mc^2}.
}
$$

It can therefore be interpreted as the **fraction of the available energetic capacity represented by the current state of motion**. This energetic interpretation provides the basis for subsequently relating the relativistic term to the \(k_e\) energy scale and to the thermodynamic expression involving \(k_B T\).

## Corollary 1: Thought experiment to confirm velocity dilation


Using the method of postulation we managed to infer dilation of momentum within computational relativity using the structure:

$$
K-e
$$

From the skeleton we produced the structure:

$$
\frac{1}{2}mv^2
=
\frac{1}{2}mv'^2
-
\frac{1}{2}\left(\frac{v^2}{c^2}\right)mv'^2
$$

Which implies the following:

$$
v'
=
\frac{v}
{\sqrt{1-\frac{v^2}{c^2}}}.
$$


This methodology implies that time dilation must also have a quadratic structure. 


The purpose of this document is to derive the structure presented above using two independent approaches: standard relativistic kinematics based on the Lorentz transformations, and a separate methodology developed within this framework. That is, we need to demonstrate that the (K-e) structure proposed within computational relativity can also be represented geometrically through a coordinate-based formulation. Both approaches aim to recover the relation

$$
v'
=
\frac{v}
{\sqrt{1-\frac{v^2}{c^2}}}.
$$

## Thought Experiment: Perpendicular Motion Inside a Relativistic Train

Consider a train moving horizontally at velocity

$$
V=0.9c.
$$

Inside the train, an object moves vertically upward with velocity

$$
v.
$$

The velocity (v) is measured in the train’s own reference frame. We want to determine the effective vertical velocity measured by an observer standing on the ground.

We compare two cases:

1. The train is at rest relative to the ground observer.
2. The train moves horizontally at (0.9c).

## 2. Derivation Using Distance and Dilated Time

## Method 1: Relativistic velocity transformation

$$
u'_x=0,
\qquad
u'_y=v
$$

$$
u_y=\frac{u'_y}{\gamma\left(1+\frac{u'_xV}{c^2}\right)}
=\frac{v}{\gamma}
$$

$$
u_y=v\sqrt{1-\frac{V^2}{c^2}}
$$

$$
V=0.9c
\quad\Longrightarrow\quad
\boxed{u_y\approx0.4359v}
$$

## Method 2: Distance over dilated time

$$
v=\frac{s}{t},
\qquad
t'=\frac{t}{\sqrt{1-\frac{V^2}{c^2}}}
$$

$$
v_{\mathrm{eff}}=\frac{s}{t'}
=\frac{s}{t}\sqrt{1-\frac{V^2}{c^2}}
$$

$$
v_{\mathrm{eff}}
=v\sqrt{1-\frac{V^2}{c^2}}
$$

$$
V=0.9c
\quad\Longrightarrow\quad
\boxed{v_{\mathrm{eff}}\approx0.4359v}
$$


---

If the **observed perpendicular velocity** is set equal to the train velocity (V), then

$$
u_{\mathrm{eff}}=u\sqrt{1-\frac{V^2}{c^2}}=V
$$

Therefore,

$$
\boxed{
u=\frac{V}{\sqrt{1-\frac{V^2}{c^2}}}
