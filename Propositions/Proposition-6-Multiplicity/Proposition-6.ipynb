## Proposition 5: Multiplicity Definition

**Multiplicity** is the number of possible configurations or states that a system can have under a given set of rules.

It can be separated into two parts:

* **Form:** the mathematical structure or rule used to determine the multiplicity, such as

$$
\Omega(r)=\binom{N}{r}.
$$

* **Actual multiplicity:** the numerical value produced by that form. For example,

$$
\Omega(2)=\binom{5}{2}=10.
$$


*** IMPORTANT *** Evidence for the existence of a multiplicity in physical reality that is distinct from $\Omega$, where $\Omega$ primarily represents the combinatorial character of randomness and entropy, is developed from the experiment presented in `experiment.ipynb`. This second form of multiplicity is represented through the $\gamma$-multiplicity.


---

### Corollary 1: Multiplicity Form Determines Energy Partitioning

Two multiplicities can have the same **actual value** while differing in their **form**.

$$
\Omega_A=\Omega_B,
\qquad
F_A\neq F_B.
$$


The form describes the structure of the multiplicity. Therefore, when multiplicity is used to describe the distribution of energy within a system, different multiplicity forms can represent different structures of energy partitioning.

For example,

$$
k_B T\ln\Omega
$$

can represent the energy associated with the entropy of the system, while

$$
k_B T\ln\gamma
$$

can represent the biological-energy contribution associated with the $\gamma$-multiplicity in the proposed framework.

### Corollary 2: 

The multiplicity terms

$$
k_B T \ln\gamma
$$

and

$$
k_B T \ln\Omega
$$

imply that, without multiplicity, the system could not favor the states determined by its form. The existence of the $\gamma$ multiplicity is demonstrated by the experiment in `experiment.ipynb`. This implies two things: that the system is computational, and that a multiplicity is replaceable. By replaceable, I mean that the computational system shaping the dynamics of a given thermodynamic system can be swapped for a different form over time. This is also shown in Corollary 6 of Proposition 4, which covers the description in standard relativity.

For example:

- **Time 1, location 1:** Form 1 (entropic) is instantiated.
- **Time 2, location 1:** Form 2 (biological) is instantiated.

### Corollary 3: Understanding the Terms as Descriptions of Thermodynamic Equilibrium


Thermodynamic expressions involving

$$
k_B T\ln\Omega
$$

The model can describe the energy associated with the structure of a system at equilibrium. However, reaching equilibrium requires a sequence of transitions that occur before the equilibrium state is established, and the model does not explicitly describe these transitions themselves.

Instead, the transitions are determined by the internal structure of the multiplicity term, represented by its **form** (as described in the proposition related to this corollary). We interpret this underlying structure as the **computational component** of the system.

Within this framework, without the multiplicity corresponding to a particular form, the transitions required to realize that state could not occur. The time required to produce the equilibrium state would therefore approach infinity. That is, multiplicity reduces the time required to produce a biological state. In Computational Relativity, this effect is represented as a corresponding reduction in the dilation term. That is, $\sqrt{1-\frac{v^2}{c^2}} = 1$


$$
\boxed{
\text{Multiplicity}\rightarrow
\text{finite partitioning time}
}
$$

$$
\boxed{
\text{No multiplicity}\rightarrow
t_{\mathrm{eq}}\rightarrow\infty
}
$$

### Demonstration 1: Multiplicity as a representation of reducing dilation factor

Consider the Boltzmann factor

$$
p=e^{-E/(k_BT)}.
$$

For a system with

$$
\frac{E}{k_BT}=10,
$$

we obtain

| $E/(k_BT)$ | $p(E)=e^{-E/k_BT}$ |
|---:|---:|
| $0$ | $1.000000$ |
| $1$ | $0.367879$ |
| $2$ | $0.135335$ |
| $3$ | $0.049787$ |
| $4$ | $0.018316$ |
| $5$ | $0.006738$ |
| $6$ | $0.002479$ |
| $7$ | $0.000912$ |
| $8$ | $0.000335$ |
| $9$ | $0.000123$ |
| $10$ | $0.0000454$ |
| $E\to\infty$ | $\to 0$ |

The final case follows directly from

$$
p=\Omega e^{-E/(k_BT)}.
$$

Setting

$$
\Omega=e^{10}
$$

gives

$$
p=e^{10}e^{-10}
=e^0
=1.
$$

Therefore, the multiplicity progressively offsets the exponential suppression until the effective term becomes zero and the statistical weight reaches

$$
\boxed{p=1}.
$$

In Computational Relativity, this is compared with the limiting form of the inverse Lorentz dilation factor,

$$
\sqrt{1-\frac{v^2}{c^2}}.
$$

When there is no dilation,

$$
\sqrt{1-\frac{v^2}{c^2}}=1.
$$

Thus, the proposed correspondence is

$$
\boxed{
\Omega\rightarrow e^{10}
\quad\Longrightarrow\quad
p\rightarrow1
\quad\Longleftrightarrow\quad
\text{dilation factor}\rightarrow1
}
$$

In this interpretation, increasing multiplicity reduces the effective statistical or computational dilation until the system reaches the limiting value of $1$.
