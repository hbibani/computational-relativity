# Galilean Relativity and the Lorentz Transformation

## 1. Definition of the Two Reference Frames

Consider two inertial reference frames.

Let

$$
S
$$

represent the external or stationary reference frame.

Let

$$
S'
$$

represent a reference frame moving horizontally relative to $S$ at constant velocity

$$
V.
$$

The motion of $S'$ is taken to be along the positive $x$-direction.

We choose the origins of the two coordinate systems so that they coincide at the initial instant:

$$
x=x'=0
$$

when

$$
t=t'=0.
$$

This means that at the beginning of the experiment, the origins of both coordinate systems occupy exactly the same position.

After time has elapsed, the origin of the moving frame $S'$ has travelled relative to $S$.

In classical mechanics, its displacement is simply

$$
Vt.
$$

The central problem of Galilean and Lorentzian relativity is therefore:

> Given the position and motion of an object measured in $S'$, what position and motion will be measured in $S$?

---

# 2. Galilean Position Transformation

Suppose an object has position

$$
x'
$$

relative to the moving frame.

At the same time, the origin of the moving frame itself has moved a distance

$$
Vt
$$

relative to the external frame.

Therefore, the external observer assigns the object the position

$$
\boxed{
x=x'+Vt
}.
$$

This is the fundamental Galilean transformation for the horizontal spatial coordinate.

The expression can be interpreted as

$$
\text{external position}
=

\text{internal position}
+
\text{frame displacement}.
$$

That is,

$$
x
=

x'
+
Vt.
$$

If the object remains exactly at the origin of the moving frame,

$$
x'=0,
$$

then

$$
x=Vt.
$$

Therefore,

$$
\boxed{
x=Vt
}.
$$

This is simply uniform linear motion.

---

# 3. Example: Object Fixed at the Origin of the Moving Frame

Take

$$
V=10\text{ m/s}.
$$

Initially,

$$
x=x'=0.
$$

Suppose

$$
t=5\text{ s}.
$$

Because the object remains at the origin of the moving frame,

$$
x'=0.
$$

The Galilean transformation gives

$$
x=x'+Vt.
$$

Therefore,

$$
x
=

0+(10)(5).
$$

Hence,

$$
\boxed{
x=50\text{ m}
}.
$$

The interpretation is straightforward.

The object has not moved relative to the train:

$$
x'=0.
$$

However, the train itself has travelled

$$
50\text{ m}.
$$

Therefore, relative to the external observer, the object has also travelled

$$
50\text{ m}.
$$

---

# 4. Galilean Absolute Time

The second fundamental assumption of Galilean relativity is that time is universal.

Thus,

$$
\boxed{
t=t'
}.
$$

This means that if one observer measures

$$
5\text{ s},
$$

every other inertial observer also measures

$$
5\text{ s}.
$$

Time is independent of the motion of the observer.

Mathematically,

$$
dt=dt'.
$$

This assumption is crucial because it allows velocities to add directly.

In Galilean mechanics, space may be described differently because the origins of the coordinate systems are moving relative to each other, but the time coordinate itself remains unchanged.

Thus,

$$
x\neq x'
$$

in general, while

$$
t=t'.
$$

---

# 5. Deriving Galilean Velocity Addition

Begin with the Galilean position transformation:

$$
x=x'+Vt.
$$

Differentiate:

$$
dx=dx'+Vdt.
$$

The velocity measured in the external frame is

$$
u_x
=

\frac{dx}{dt}.
$$

Therefore,

$$
u_x
=

\frac{dx'+Vdt}{dt}.
$$

Separate the terms:

$$
u_x
=

\frac{dx'}{dt}
+
V.
$$

Because Galilean relativity assumes

$$
dt=dt',
$$

we have

$$
\frac{dx'}{dt}
=

\frac{dx'}{dt'}.
$$

Define the internal velocity as

$$
u'_x
=

\frac{dx'}{dt'}.
$$

Therefore,

$$
\boxed{
u_x=u'_x+V
}.
$$

This is the classical velocity-addition law.

It states that the velocity observed externally is simply the object's velocity inside the moving frame plus the velocity of the moving frame itself.

---

# 6. Example: Horizontal Motion Inside a Train

Suppose the train moves at

$$
V=20\text{ m/s}.
$$

An object moves forward inside the train at

$$
u'_x=5\text{ m/s}.
$$

The Galilean transformation gives

$$
u_x=u'_x+V.
$$

Therefore,

$$
u_x=5+20.
$$

Hence,

$$
\boxed{
u_x=25\text{ m/s}
}.
$$

We can also derive this result from position.

Suppose

$$
t=t'=4\text{ s}.
$$

The object travels inside the train

$$
x'
=

u'_x t'.
$$

Therefore,

$$
x'
=

(5)(4)

20\text{ m}.
$$

During the same period, the train travels

$$
Vt
=

(20)(4)

80\text{ m}.
$$

Consequently,

$$
x
=

x'+Vt.
$$

Thus,

$$
x
=

20+80

100\text{ m}.
$$

The external velocity is therefore

$$
u_x
=

\frac{x}{t}.
$$

Hence,

$$
u_x
=

\frac{100}{4}.
$$

Therefore,

$$
\boxed{
u_x=25\text{ m/s}
}.
$$

This agrees exactly with

$$
u_x=u'_x+V.
$$

---

# 7. Galilean Perpendicular Motion

Suppose the moving frame travels horizontally along the $x$-axis.

The perpendicular coordinate is the $y$-coordinate.

Galilean relativity assumes

$$
\boxed{
y=y'
}.
$$

Since time is also unchanged,

$$
t=t',
$$

we have

$$
dy=dy'
$$

and

$$
dt=dt'.
$$

Therefore,

$$
u_y
=

\frac{dy}{dt}

\frac{dy'}{dt'}.
$$

Hence,

$$
\boxed{
u_y=u'_y
}.
$$

This means that horizontal motion of the reference frame has no effect on the perpendicular component of velocity in Galilean relativity.

---

# 8. Example: Purely Vertical Motion Inside the Train

Suppose

$$
V=20\text{ m/s}
$$

and an object moves vertically inside the train with

$$
u'_y=4\text{ m/s}.
$$

In Galilean relativity,

$$
u_y=u'_y.
$$

Therefore,

$$
\boxed{
u_y=4\text{ m/s}
}.
$$

The external observer sees the object moving horizontally because of the train:

$$
u_x=V=20\text{ m/s}.
$$

At the same time, the vertical velocity remains

$$
u_y=4\text{ m/s}.
$$

The total velocity is therefore

$$
u
=

\sqrt{u_x^2+u_y^2}.
$$

Substituting,

$$
u
=

\sqrt{20^2+4^2}.
$$

Therefore,

$$
u
=

\sqrt{416}.
$$

Hence,

$$
\boxed{
u\approx20.40\text{ m/s}
}.
$$

The horizontal and vertical components are treated independently.

---

# 9. The Classical Geometry of Galilean Relativity

Galilean relativity therefore has the basic structure

$$
\boxed{
x=x'+Vt
}
$$

$$
\boxed{
y=y'
}
$$

$$
\boxed{
t=t'
}.
$$

This produces

$$
\boxed{
u_x=u'_x+V
}
$$

and

$$
\boxed{
u_y=u'_y
}.
$$

The central feature is that time is common to every component.

There is one universal differential

$$
dt=dt'.
$$

Consequently, horizontal motion of the reference frame changes the horizontal position and velocity but does not alter the perpendicular velocity.

---

# 10. The Problem at Relativistic Velocities

The Galilean equations work extremely well when

$$
V\ll c.
$$

However, if they were applied directly to light, they would predict that velocities greater than $c$ could occur.

For example, suppose

$$
V=0.6c
$$

and

$$
u'_x=0.8c.
$$

Galilean velocity addition would give

$$
u_x
=

u'_x+V.
$$

Therefore,

$$
u_x
=

0.8c+0.6c.
$$

Hence,

$$
u_x=1.4c.
$$

This conflicts with the relativistic requirement that light in vacuum has invariant speed

$$
c.
$$

Lorentz transformations replace the Galilean transformations so that the space and time coordinates change together.

---

# 11. Lorentz Factor

The central factor is

$$
\boxed{
\gamma_V
=

\frac{1}{
\sqrt{1-\frac{V^2}{c^2}}
}
}.
$$

When

$$
V\ll c,
$$

the quantity

$$
\frac{V^2}{c^2}
$$

is very small.

Thus,

$$
\gamma_V\approx1.
$$

Consequently, the Lorentz transformation approaches the Galilean transformation.

When $V$ becomes comparable with $c$,

$$
\gamma_V>1,
$$

and relativistic corrections become significant.

---

# 12. Lorentz Position Transformation

The Galilean expression is

$$
x=x'+Vt'.
$$

The corresponding Lorentz transformation is

$$
\boxed{
x
=

\gamma_V
\left(
x'+Vt'
\right)
}.
$$

The term inside the parentheses,

$$
x'+Vt',
$$

has the same structural form as the Galilean transformation.

The important difference is the multiplicative Lorentz factor

$$
\gamma_V.
$$

If

$$
x'=0,
$$

then

$$
x
=

\gamma_V Vt'.
$$

However, this does not mean that the frame origin physically has velocity

$$
\gamma_VV
$$

in $S$.

The reason is that the corresponding external time is also transformed.

For

$$
x'=0,
$$

the Lorentz time transformation gives

$$
t=\gamma_Vt'.
$$

Therefore,

$$
\frac{x}{t}
=

\frac{\gamma_VVt'}{\gamma_Vt'}.
$$

The factors cancel:

$$
\boxed{
\frac{x}{t}=V
}.
$$

Thus, the train still moves at velocity

$$
V.
$$

This cancellation is fundamental.

Space and time are transformed together.

---

# 13. Example: Lorentz Position at (V=0.6c)

Let

$$
V=0.6c.
$$

Then

$$
\gamma_V
=

\frac{1}{
\sqrt{1-0.6^2}
}.
$$

Therefore,

$$
\gamma_V
=

\frac{1}{
\sqrt{0.64}
}.
$$

Hence,

$$
\gamma_V=1.25.
$$

Let

$$
x'=0
$$

and

$$
t'=1\text{ s}.
$$

The Lorentz position transformation gives

$$
x
=

\gamma_V
\left(
0+Vt'
\right).
$$

Therefore,

$$
x
=

(1.25)(0.6c)(1).
$$

Hence,

$$
x=0.75c\text{ s}.
$$

Using

$$
c=3.00\times10^8\text{ m/s},
$$

we obtain

$$
x
=

0.75
\left(
3.00\times10^8
\right).
$$

Therefore,

$$
\boxed{
x=2.25\times10^8\text{ m}
}.
$$

At first sight this appears larger than the classical displacement

$$
Vt'
=

0.6c(1)

1.80\times10^8\text{ m}.
$$

However, the external elapsed time is not

$$
1\text{ s}.
$$

Instead,

$$
t
=

\gamma_Vt'

1.25\text{ s}.
$$

Therefore,

$$
\frac{x}{t}
=

\frac{2.25\times10^8}{1.25}.
$$

Hence,

$$
\boxed{
\frac{x}{t}
=
 1.80\times10^8\text{ m/s}

0.6c
}.
$$

So the velocity remains exactly

$$
V=0.6c.
$$

---

# 14. Lorentz Time Transformation

The complete Lorentz time transformation is

$$
\boxed{
t
=

\gamma_V
\left(
t'
+
\frac{Vx'}{c^2}
\right)
}.
$$

This is one of the most important differences between Galilean and relativistic mechanics.

Galilean relativity has

$$
t=t'.
$$

Lorentz relativity instead makes the external time depend on both

$$
t'
$$

and

$$
x'.
$$

Therefore, time and space can no longer be transformed independently.

The term

$$
\frac{Vx'}{c^2}
$$

is known as the relativity-of-simultaneity term.

It means that events occurring at different positions in $S'$ generally do not have the same time separation when described in $S$.

---

# 15. Special Case (x'=0)

If the event occurs at the origin of the moving frame,

$$
x'=0,
$$

then the time transformation simplifies to

$$
t
=

\gamma_V t'.
$$

Thus,

$$
\boxed{
t=\gamma_Vt'
}.
$$

Equivalently,

$$
t'
=

\frac{t}{\gamma_V}.
$$

Since

$$
\frac{1}{\gamma_V}
=

\sqrt{
1-\frac{V^2}{c^2}
},
$$

we obtain

$$
\boxed{
t'
==

t
\sqrt{
1-\frac{V^2}{c^2}
}
}.
$$

This is the ordinary time-dilation relationship for a clock at rest in the moving frame.

---

Let

$$
V=0.8c.
$$

Then

$$
\gamma_V
=

\frac{1}{
\sqrt{1-0.8^2}
}.
$$

Therefore,

$$
\gamma_V
=

\frac{1}{
\sqrt{0.36}
}.
$$

Thus,

$$
\gamma_V
=

\frac{1}{0.6}.
$$

Hence,

$$
\boxed{
\gamma_V
=

1.6667
}.
$$

If the moving clock measures

$$
t'=6\text{ s},
$$

then

$$
t
=

\gamma_Vt'.
$$

Therefore,

$$
t
=

(1.6667)(6).
$$

Hence,

$$
\boxed{
t=10\text{ s}
}.
$$

The moving clock records

$$
6\text{ s}
$$

while the external frame records

$$
10\text{ s}.
$$

# 17. Derivation of Relativistic Horizontal Velocity Addition

Start with

$$
x
=

\gamma_V(x'+Vt').
$$

Differentiate:

$$
dx
=

\gamma_V(dx'+Vdt').
$$

The time transformation is

$$
t
=

\gamma_V
\left(
t'
+
\frac{Vx'}{c^2}
\right).
$$

Differentiate:

$$
dt
=

\gamma_V
\left(
dt'
+
\frac{Vdx'}{c^2}
\right).
$$

The external horizontal velocity is

$$
u_x
=

\frac{dx}{dt}.
$$

Therefore,

$$
u_x
=

\frac{
\gamma_V(dx'+Vdt')
}{
\gamma_V
\left(
dt'+\frac{Vdx'}{c^2}
\right)
}.
$$

Cancel

$$
\gamma_V.
$$

Then

$$
u_x
=

\frac{
dx'+Vdt'
}{
dt'+\frac{Vdx'}{c^2}
}.
$$

Divide the numerator and denominator by

$$
dt'.
$$

Thus,

$$
u_x
=

\frac{
\frac{dx'}{dt'}+V
}{
1+
\frac{V}{c^2}
\frac{dx'}{dt'}
}.
$$

Since

$$
u'_x
=

\frac{dx'}{dt'},
$$

we obtain

$$
\boxed{
u_x
=

\frac{
u'_x+V
}{
1+\frac{u'_xV}{c^2}
}
}.
$$

This is the relativistic horizontal velocity-addition formula.

---

# 18. Connection to Galilean Velocity Addition

Consider

$$
u_x
=

\frac{
u'_x+V
}{
1+\frac{u'_xV}{c^2}
}.
$$

At ordinary velocities,

$$
u'_xV\ll c^2.
$$

Therefore,

$$
\frac{u'_xV}{c^2}
\approx0.
$$

The denominator becomes approximately

$$
1.
$$

Thus,

$$
u_x
\approx
u'_x+V.
$$

Therefore,

$$
\boxed{
u_x
\longrightarrow
u'_x+V
}
$$

in the classical limit.

The Galilean transformation is consequently the low-velocity approximation of the Lorentz transformation.

---

# 19. Example: Relativistic Horizontal Velocity

Let

$$
V=0.6c
$$

and

$$
u'_x=0.5c.
$$

Then

$$
u_x
=

\frac{
0.5c+0.6c
}{
1+
\frac{(0.5c)(0.6c)}{c^2}
}.
$$

Simplify the numerator:

$$
0.5c+0.6c
=

1.1c.
$$

Simplify the denominator:

$$
1+
\frac{(0.5c)(0.6c)}{c^2}
=

1+0.30.
$$

Therefore,

$$
u_x
=

\frac{1.1c}{1.30}.
$$

Hence,

$$
\boxed{
u_x
\approx0.8462c
}.
$$

Galilean relativity would instead predict

$$
u_x
=

1.1c.
$$

Thus, the relativistic denominator

$$
1+\frac{u'_xV}{c^2}
$$

prevents ordinary velocity addition from producing a velocity greater than $c$.

---

# 20. The Origin of the Velocity Denominator

The term

$$
1+\frac{u'_xV}{c^2}
$$

does not come from the spatial transformation alone.

It comes from the transformation of time.

Recall

$$
dt
=

\gamma_V
\left(
dt'
+
\frac{Vdx'}{c^2}
\right).
$$

Factor out

$$
dt'.
$$

Then

$$
dt
=

\gamma_Vdt'
\left(
1+
\frac{V}{c^2}
\frac{dx'}{dt'}
\right).
$$

Since

$$
u'_x
=

\frac{dx'}{dt'},
$$

we obtain

$$
\boxed{
dt
=

\gamma_Vdt'
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

This formula is particularly important.

For an object with

$$
u'_x=0,
$$

we obtain

$$
dt=\gamma_Vdt'.
$$

But for an object that itself moves horizontally within $S'$,

$$
u'_x\neq0,
$$

we obtain an additional factor:

$$
1+\frac{u'_xV}{c^2}.
$$

Thus, the transformation of the object's trajectory involves more than the time dilation associated with the train alone.

---

# 21. Lorentz Transformation of the Perpendicular Coordinate

For motion along the $x$-axis, Lorentz transformations leave the perpendicular spatial coordinate unchanged:

$$
\boxed{
y=y'
}.
$$

Therefore,

$$
dy=dy'.
$$

However, unlike Galilean relativity,

$$
dt\neq dt'.
$$

The transverse velocity is

$$
u_y
=

\frac{dy}{dt}.
$$

Since

$$
dy=dy',
$$

we obtain

$$
u_y
=

\frac{dy'}{
\gamma_Vdt'
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

Therefore,

$$
u_y
=

\frac{
\frac{dy'}{dt'}
}{
\gamma_V
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

Since

$$
u'_y
=

\frac{dy'}{dt'},
$$

we obtain

$$
\boxed{
u_y
=

\frac{
u'_y
}{
\gamma_V
\left(
1+\frac{u'_xV}{c^2}
\right)
}
}.
$$

Using

$$
\frac{1}{\gamma_V}
=

\sqrt{
1-\frac{V^2}{c^2}
},
$$

this becomes

$$
\boxed{
u_y
=

u'_y
\frac{
\sqrt{
1-\frac{V^2}{c^2}
}
}{
1+\frac{u'_xV}{c^2}
}
}.
$$

---

# 22. Purely Perpendicular Internal Motion

Now set

$$
u'_x=0.
$$

Then

$$
1+\frac{u'_xV}{c^2}
=

1.

$$

Therefore,

$$
u_y
=

\frac{u'_y}{\gamma_V}.
$$

Hence,

$$
\boxed{
u_y
=

u'_y
\sqrt{
1-\frac{V^2}{c^2}
}
}.
$$

This result has a simple interpretation.

The perpendicular displacement is unchanged:

$$
dy=dy'.
$$

However, the external observer associates a longer time interval with the motion:

$$
dt=\gamma_Vdt'.
$$

Since velocity is

$$
u_y
=

\frac{dy}{dt},
$$

the same perpendicular displacement divided by a larger external time gives a smaller perpendicular velocity.

---

# 24. Motion with Both Horizontal and Vertical Internal Components

Suppose now that

$$
u'_x\neq0
$$

and

$$
u'_y\neq0.
$$

Then the horizontal component is

$$
\boxed{
u_x
=

\frac{
u'_x+V
}{
1+\frac{u'_xV}{c^2}
}
}
$$

and the vertical component is

$$
\boxed{
u_y
=

\frac{
u'_y
}{
\gamma_V
\left(
1+\frac{u'_xV}{c^2}
\right)
}
}.
$$

Notice that both expressions contain

$$
1+\frac{u'_xV}{c^2}.
$$

This occurs because both velocity components are measured using the same transformed external time

$$
dt.
$$

The horizontal internal motion therefore affects not only the transformed horizontal component but also the observed vertical component.

and

$$
u'_y=0.4c.
$$

First suppose

$$
u'_x=0.
$$

Then

$$
u_y
=

0.4c(0.6).
$$

Therefore,

$$
\boxed{
u_y=0.24c
}.
$$

Now introduce

$$
u'_x=0.3c.
$$

Then

$$
u_y
=

\frac{0.24c}{1.24}.
$$

Hence,

$$
\boxed{
u_y\approx0.1935c
}.
$$

Therefore, the additional horizontal internal motion changes the transformed vertical velocity from

$$
0.24c
$$

to

$$
0.1935c.
$$

The ratio is

$$
\frac{0.1935}{0.24}
\approx0.8065.
$$

This is precisely

$$
\frac{1}{1.24}.
$$

Therefore, the additional reduction is generated by

$$
\boxed{
\frac{1}{
1+\frac{u'_xV}{c^2}
}
}.
$$

# 27. Important Interpretation of the Additional Factor

It is tempting to interpret

$$
1+\frac{u'_xV}{c^2}
$$

as though the internal horizontal motion consumes some finite quantity of the train's original time dilation.

That is not the standard relativistic interpretation.

The train itself still has the Lorentz factor

$$
\boxed{
\gamma_V
=

\frac{1}{
\sqrt{1-\frac{V^2}{c^2}}
}
}.
$$

This factor depends only on the relative velocity

$$
V
$$

between $S$ and $S'$.

The additional term

$$
1+\frac{u'_xV}{c^2}
$$

appears because the object is moving through different positions

$$
x'
$$

inside the moving frame.

Since Lorentz time depends on both

$$
t'
$$

and

$$
x',
$$

the object's horizontal trajectory modifies the relation between

$$
dt
$$

and

$$
dt'.
$$

Thus,

$$
dt
=

\gamma_Vdt'
$$

for an object at fixed

$$
x',
$$

but

$$
dt
=

\gamma_Vdt'
\left(
1+\frac{u'_xV}{c^2}
\right)
$$

for an object moving horizontally inside the frame.

# 28. Four Distinct Cases

The formulas become particularly clear when separated into four cases.

## Case 1: Object stationary inside the train

$$
u'_x=0,
\qquad
u'_y=0.
$$

Then

$$
u_x=V
$$

and

$$
u_y=0.
$$

The object's clock follows

$$
dt=\gamma_Vdt'.
$$

---

## Case 2: Pure horizontal internal motion

$$
u'_x\neq0,
\qquad
u'_y=0.
$$

Then

$$
u_x
=

\frac{
u'_x+V
}{
1+\frac{u'_xV}{c^2}
}
$$

and

$$
u_y=0.
$$

The trajectory has

$$
dt
==

\gamma_Vdt'
\left(
1+\frac{u'_xV}{c^2}
\right).
$$

---

## Case 3: Pure vertical internal motion

$$
u'_x=0,
\qquad
u'_y\neq0.
$$

Then

$$
u_x=V
$$

and

$$
u_y
===

\frac{u'_y}{\gamma_V}.
$$

Thus,

$$
u_y
=

u'_y
\sqrt{
1-\frac{V^2}{c^2}
}.
$$

---

## Case 4: Horizontal and vertical internal motion

$$
u'_x\neq0,
\qquad
u'_y\neq0.
$$

Then

$$
u_x
=

\frac{
u'_x+V
}{
1+\frac{u'_xV}{c^2}
}
$$

and

$$
u_y
=

\frac{
u'_y
}{
\gamma_V
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

This is the most general two-dimensional case considered here.

# 30. Central Difference Between Galilean and Lorentz Relativity

The fundamental Galilean assumption is

$$
\boxed{
dt=dt'
}.
$$

Because time is universal, velocities can be added independently:

$$
u_x=u'_x+V
$$

and

$$
u_y=u'_y.
$$

The fundamental Lorentzian change is

$$
\boxed{
dt
=

\gamma_V
\left(
dt'
+
\frac{Vdx'}{c^2}
\right)
}.
$$

Since

$$
dx'=u'_xdt',
$$

this becomes

$$
\boxed{
dt
=

\gamma_Vdt'
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

This one equation explains why both transformed velocity components acquire relativistic corrections.

For the horizontal component,

$$
u_x
=

\frac{
u'_x+V
}{
1+\frac{u'_xV}{c^2}
}.
$$

For the perpendicular component,

$$
u_y
=

\frac{
u'_y
}{
\gamma_V
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

Thus, the essential transition from Galilean to Lorentzian relativity can be summarized as the replacement

$$
\boxed{
dt=dt'
}
$$

with

$$
\boxed{
dt
=

\gamma_Vdt'
\left(
1+\frac{u'_xV}{c^2}
\right)
}.
$$

Once the time coordinate becomes dependent upon both motion of the reference frame and the object's position within that frame, the simple Galilean addition of velocities is replaced by the Lorentz velocity-transformation equations.
