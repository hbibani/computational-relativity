# **SECTION 1: INTRODUCTION**

---

In this section, we expand the definition of the temperature-dilation formula introduced through the **K-e bridge in Computational Relativity**. The formula was initially developed through two approaches.

The first approach used the Boltzmann distribution, which connects probability and energy through the relation

$$
\Delta E=-k_BT\ln\gamma.
$$

Taking the derivative with respect to $\gamma$ gives a term proportional to $k_BT/\gamma$. Comparing this with the computational energy scale $k_e$ led us to propose the temperature-dilation formula

$$
\boxed{
\gamma_d\approx
\frac{1}{1-\frac{k_BT}{k_e\gamma}}
}.
$$

The second approach began with the proposed energy-allocation relation

$$
K'=K-e.
$$

By applying this relation to the thermodynamic energy of a system, we obtained

$$
\frac{3}{2}Nk_BT
=
\frac{3}{2}Nk_BT'
\left(1-\frac{k_BT}{k_e}\right).
$$

Rearranging gives

$$
\boxed{
\frac{T'}{T}
=
\frac{1}{1-\frac{k_BT}{k_e}}
}.
$$

Both approaches produce a similar mathematical structure, although the first includes the additional factor $1/\gamma$. They become identical if that factor is included in the definition of the effective $k_e$.

**The next step is to expand the definition of $k_e$ using the relativistic analogue developed in the K-e bridge.**

In the relativistic formulation, we proposed that the total motion contribution could be divided into separate motion and computational-field contributions:

$$
D_{\mathrm{motion}}+D_{\mathrm{field}}
=
\frac{mv^2}{mc^2}.
$$

By comparing this structure with the Taylor-series expansion of the relativistic dilation factor, we proposed that the computational-field contribution could contain multiple terms, rather than a single contribution.

We now apply the same idea to the thermodynamic formulation. Instead of representing the computational-field contribution by one term, we propose expanding it into a sum of temperature-dependent terms:

$$
\boxed{
\frac{k_BT}{k_e\gamma}
\longrightarrow
\sum_{N=1}^{M}
\left(\frac{T}{T_0}\right)^N
\frac{1}{2^N}
}
$$

where $T_0=6.49\times10^{13}\ \mathrm K$ is the reference temperature, $N$ identifies the order of each contribution, and $M$ is the total number of terms included. The weighting factor $1/2^N$ is the proposed form used in this expansion.

Substituting this summation into the original temperature-dilation formula gives

$$
\boxed{
\gamma_d(T,M)=
\frac{1}{
1-\displaystyle\sum_{N=1}^{M}
\left(\frac{T}{2T_0}\right)^N
}
}
$$

**The purpose of this section is to explain how we arrived at this expanded formula and how its individual terms relate to the computational-field contributions proposed in the relativistic analogue.**

We will also examine how the resulting dilation changes with temperature and with the number of terms included in the summation. These relationships are mathematical predictions of the proposed model and still require physical validation.

# **SECTION 2: Boltzmann Distribution**

---


The Boltzmann distribution describes a small system, which may be as small as a single particle, in thermal contact with a large heat reservoir with which it exchanges energy. It gives the probability of finding the small system in a state of energy $E$ as $P(E)\propto e^{-E/k_BT}$, where $T$ is the temperature of the reservoir.

The phenomenon observed in the experiment was traced to this distribution: the effect arose from the occurrence of low-probability states. Substituting the postulated relation into the Boltzmann form gave a strong estimate of the dilation factor in terms of both time and temperature. It also showed that a computational field which alters the probabilistic dynamics of a thermodynamic system must alter its internal energy, changing the forms of energy present. In the present framework, these changes appear in biological form. Translating the result into relativistic terms then yielded two distinct dilation factors.

Assume the observed probability is dilated by a factor $\gamma$:

$$
p_{\text{obs}} = \gamma\, p_{\text{exp}}.
$$

Substitute the Boltzmann form into the dilation:

$$
p_{\text{obs}} \propto \gamma\, e^{-E_{\text{exp}}/(k_B T)}.
$$


This leads to the following formula, which specifies how the computational field shapes the formation of internal energy. In particular, it identifies the energy as taking a biological form, a result obtained by postulation and guided by the energy relation:

$$
\Delta E=-k_B T\ln\gamma,
$$

# **SECTION 3: BOLTZMANN DISTRIBUTION MULTIPLE STATES**

---

The temperature-dilation factor associated with a multiplicity-dependent contribution was initially expressed as

$$
\boxed{
\gamma_d
\approx
\frac{1}
{1-\frac{k_B T}{k_e\gamma}}
}.
$$

Within Computational Relativity, this expression represents the proposed relationship between temperature, computational energy, and dilation. The term $\frac{k_BT}{k_e\gamma}$ describes the computational contribution associated with a particular multiplicity of particles, where $k_e$ represents the effective computational energy parameter and $\gamma$ provides the corresponding weighting factor.

**This contribution is associated with a particular $N$-tuple configuration of a multi-particle system**, rather than necessarily representing the total contribution of the entire system.


The small system described by the Boltzmann distribution can occupy any of several states, each with its own energy $E_i$ and an associated probability. As the diagram shows, the system may be found in any one of these states.

</p



In the relativistic analogue, the total dilation is proposed to include multiple contributions, each associated with a different multiplicity order. Therefore, instead of considering a single computational contribution, we extend the formulation to include **the summation of all multiplicity-dependent contributions**, with each order possessing its own values of $k_{e,N}$ and $\gamma_N$.

The proposed temperature-dilation expression consequently becomes

$$
\boxed{
\gamma_d
\approx
\frac{1}{
1-\displaystyle\sum_{N=1}^{M}
\frac{k_BT}{k_{e,N}\gamma_N}
}
}.
$$

Taking the limit as $M\to\infty$ gives

$$
\boxed{
\gamma_d
\approx
\frac{1}{
1-\displaystyle\sum_{N=1}^{\infty}
\frac{k_BT}{k_{e,N}\gamma_N}
}
}.
$$

This extends the original single-term formulation into a **spectrum of multiplicity-dependent computational contributions**, whose combined effect determines the proposed total temperature-dilation factor, provided the series converges and the denominator remains nonzero.

# **SECTION 4: TEMPERATURE DILATION FORMULAS**

---

Our initial attempt to formulate the temperature-dilation relationship produced the following expression:

$$
\boxed{
\gamma_d
\approx
\frac{1}
{1-\frac{k_B T}{k_e\gamma}}
}.
$$

to transfer the terms above to relativistic terms we have the total:


$$\boxed{\frac{k_B T}{k_e\gamma} \approx \frac{\frac{1}{2}mc^2}{k_e} \approx  \frac{mv^2}{2k_e}  \approx \frac{mv^2}{2k_e k_m} = \frac{mv^2}{mc^2} \times \frac{1}{k_e}}$$

The proposed correspondence between thermodynamic energy and relativistic motion was initially expressed through the relationship

$$
\boxed{
\frac{k_B T}{k_e\gamma}
\approx
\frac{mv^2}{2k_e}
}.
$$

To reconcile certain aspects of the thermodynamic and relativistic formulations, we introduced the proposition that **the total dilation contribution could be striated into distinct components**, representing different contributions to the system's motion and internal physical organization.

The initial reasoning was that the total squared velocity could be decomposed into separate contributions, expressed as

$$
\frac{mv^2}{mc^2}
=
\frac{mv_1^2}{mc^2}
+
\frac{mv_2^2}{mc^2}
+
\frac{mv_3^2}{mc^2}
+\cdots,
$$

where the equality assumes a decomposition satisfying $v^2=\sum_i v_i^2$, as occurs for mutually orthogonal velocity components.

Extending this reasoning, we proposed that **the total relativistic motion contribution could be partitioned into two complementary components: one associated with ordinary motion and the other with the influence of computational fields upon the system's internal organization.**

This led to the formulation

$$
\boxed{
D_{\mathrm{motion}}
+
D_{\mathrm{field}}
=
\frac{mv^2}{mc^2}
\left(
1-
\sum_n
\frac{mv^2}{2k_{e,n}}
\right)
+
\frac{mv^2}{mc^2}
\sum_n
\frac{mv^2}{2k_{e,n}}
}.
$$

In this representation, $D_{\mathrm{motion}}$ describes the remaining contribution attributed to ordinary motion, while $D_{\mathrm{field}}$ represents the portion assigned to computational-field interactions. Their sum preserves the original dimensionless quantity $mv^2/mc^2$, while allowing the proposed field contribution to be distributed across multiple terms.

**An important development was that this formulation arose through a line of reasoning initially independent of the Taylor-series expansion of the relativistic dilation factor.** Rather than beginning with the established expansion and constructing an interpretation around its coefficients, we first approached the problem through the statistical formulation, the proposed partitioning of energy, and the distinction between motion and computational-field contributions.

At the time these relationships were being developed, the Taylor-series representation had not been used as the foundation of the derivation. Indeed, a substantial portion of the proposed mathematical structure had already been formulated before we became aware of the relevant Taylor-series expansion.

Subsequent comparison revealed a possible mathematical correspondence between the multiplicity-dependent field contributions and the individual terms appearing in the Taylor-series expansion of the Lorentz factor.

Within Computational Relativity, this correspondence is proposed to take the form

$$
\boxed{
\frac{mv^2}{mc^2}
\frac{mv^2}{2k_{e,n}}
=
a_n
\left(
\frac{mv^2}{mc^2}
\right)^n
}
$$

where $a_n$ denotes the coefficient associated with the $n$th term of the relevant expansion, and $k_{e,n}$ represents the corresponding effective computational-field parameter.

Under this proposed identification, **each multiplicity-dependent contribution can be associated with a distinct order of the relativistic expansion**, providing a mathematical framework for investigating how different computational-field contributions might be distributed within the total dilation term.

The correspondence is an algebraic identification within the proposed framework; establishing that these terms represent physically distinct energy contributions requires an independent physical derivation.



In extending the thermodynamic formulation to its relativistic analogue, we arrived at a more complete interpretation of the effective computational energy parameter $k_e$. Rather than representing a single contribution, **$k_e$ was proposed to incorporate a spectrum of multiplicity-dependent terms**, expressed through a summation of the form $\sum_N (T/T_0)^N(1/\gamma_N)$, where the weighting factor $\gamma_N$ is determined from the underlying mathematical principles of the framework.

This formulation allows the proposed computational-field contribution to be represented as a series of temperature-dependent terms, each corresponding to a distinct order of the system's internal organization. Taking $\gamma_N=2^N$, and assuming the summation terminates at a finite order $M$, the resulting dilation expression becomes

$$
\boxed{
\gamma_d(T,M)=
\frac{1}{
1-\displaystyle\sum_{N=1}^{M}
\left(\frac{T}{2T_0}\right)^N
}
}
$$

# **SECTION 5: DISCUSSION MATHEMATICAL STRUCTURE**

---


In thermodynamics, the multiplicity of a system is conventionally represented by $\Omega$, while within Computational Relativity, we introduce $\gamma$ as a proposed multiplicity-dependent term. We initially proposed that the dilation contribution would be inversely proportional to the multiplicity term, taking the form $1/\gamma$. This places the multiplicity factor in the denominator, producing a mathematical structure similar to the relativistic analogue, whose Taylor-series coefficients contain the factor $1/4^n$.

Although these structures are closely related in form, **they are not necessarily mathematically equivalent**. Their similarity suggests a possible correspondence between thermodynamic multiplicity and relativistic dilation, but the precise relationship between them requires further derivation.

Furthermore, the thermodynamic expressions $k_BT\ln\gamma$ and $k_BT\ln\Omega$ are interpreted within the proposed framework as describing the energy associated with the system's organization and multiplicity. **These expressions do not explicitly describe the underlying transitions through which particles reach equilibrium.** Instead, the mechanisms responsible for those transitions are abstracted from the mathematical description, leaving the resulting equilibrium state and its associated energy or multiplicity as the quantities being represented.

Within Computational Relativity, this distinction is important because the proposed computational field is intended to describe the underlying mechanism responsible for changes in the system's internal organization, rather than merely describing the equilibrium state that results from those changes.

# **SECTION 6: FREE ENERGY DEFINITION AND MULTIPLICITY**

---

In the previous section, we proposed that temperature dilation could be expressed as a summation of multiplicity-dependent contributions. We now introduce the Boltzmann distribution to account for the energy associated with these contributions.

For simplicity, we assume that **all configurations have the same fixed energy $E$**. The partition function of an ensemble containing $n$ binary elements is then

$$
\boxed{
Z_n=\sum_{r=0}^{n}\binom{n}{r}e^{-E/(k_BT)}
}
$$

Using the identity

$$
\sum_{r=0}^{n}\binom{n}{r}=2^n,
$$

we obtain

$$
\boxed{
Z_n=2^n e^{-E/(k_BT)}
}
$$

The partition function is related to the Helmholtz free energy through

$$
F=-k_BT\ln Z_n.
$$

Therefore,

$$
\boxed{
F=E-nk_BT\ln2
}
$$

This shows how the free energy depends on both the fixed energy $E$ and the multiplicity of the ensemble.

## Introducing Boltzmann Weighting into Temperature Dilation

We can now extend our proposed temperature-dilation formula by replacing the original multiplicity weighting $1/2^N$ with the reciprocal partition function $1/Z_N$:

$$
\boxed{
\gamma_d(T)=
\frac{1}{
1-\displaystyle\sum_{N=1}^{\infty}
\left(\frac{T}{T_0}\right)^N
\frac{1}{2^N e^{-E/(k_BT)}}
}
}
$$

Since $E$ is constant across all contributions, we can factor the exponential term out of the summation:

$$
\boxed{
\gamma_d(T)=
\frac{1}{
1-e^{E/(k_BT)}
\displaystyle\sum_{N=1}^{\infty}
\left(\frac{T}{2T_0}\right)^N
}
}
$$

Using the infinite geometric series,

$$
\sum_{N=1}^{\infty}
\left(\frac{T}{2T_0}\right)^N
=
\frac{T}{2T_0-T},
$$

we obtain

$$
\boxed{
\gamma_d(T)=
\frac{2T_0-T}{
2T_0-T-Te^{E/(k_BT)}
}
}
$$

Here, $N$ represents the order of each proposed contribution. The summation extends to infinity because no maximum order is assumed. The geometric series converges for $0\leq T<2T_0$, and the resulting dilation expression additionally requires a nonzero denominator.

The use of $1/Z_N$ as the dilation weighting is a proposed modelling assumption.
