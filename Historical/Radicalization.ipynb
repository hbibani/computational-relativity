# Radicalization and the Motivational Effects of Fanaticism

### Proposition 1: There are two classes of attributes with respect to necessity.

**a. Essential attributes** — attributes regarded as necessary or beneficial to the development and functioning of the individual, such as intelligence, physical strength, discipline, and reason.

**b. Non-essential attributes** — attributes that are not necessary to the individual's proper development and may, in some circumstances, be detrimental, such as excessive anger, hatred, and envy.

---

### Proposition 2: Dissociation

**Dissociation**, in this context, refers to a process of separation in which an individual becomes psychologically distanced from a particular non-essential attribute.

Where the attribute being removed is harmful, this separation constitutes a **positive modification** of the individual.

### Corollary 1

Certain internal attributes or motivational states can increase the rate at which this separation occurs.

---

### Proposition 3: Radicalization

**Radicalization** is an internal state generated through education, experience, or a combination of both, which produces an intense desire either:

1. to eradicate particular non-essential attributes; or
2. to strengthen particular essential attributes.

Radicalization therefore describes the intensification of a person's commitment toward some perceived end.

---

### Proposition 4: Religious Feeling

A **religious feeling or experience** is one generated through an individual's investigation of, or confrontation with, questions concerning his purpose, existence, and proper mode of life.

### Corollary 1

Radicalization may arise through an intense theological investigation because theological inquiry frequently concerns questions of purpose, obligation, and the meaning of one's existence.

### Scholium

When radicalization originates from a person's conception of his purpose, it may be classified as a **religious attribute** in the broader sense: an internal motive force derived from the individual's understanding of what he believes he ought to become or accomplish.

---

### Corollary 2: Radicalization may be divided into two forms according to motive.

**a. Destructive or hateful radicalization** — radicalization that rationalizes the destruction, degradation, or deconstruction of people, institutions, or things surrounding the individual.

**b. Constructive or reasonable radicalization** — radicalization directed toward construction, education, scientific investigation, self-development, or the improvement of people and things surrounding the individual.

**Scientific radicalization** belongs to the second category when its motivating force is directed toward knowledge, discovery, construction, and disciplined inquiry.

---

### Corollary 3: Radicalization can increase endurance.

An intensely committed individual may tolerate levels of difficulty, discomfort, sacrifice, or sustained effort that he would ordinarily be unwilling to endure.

### Scholium 1

Extreme ideological commitment can sometimes lead individuals to acts of severe self-sacrifice, including destructive or fatal acts undertaken for the sake of some perceived objective. Such behaviour demonstrates the strength that an internal motive can exert over ordinary self-preservation, although its destructive expression is neither useful nor desirable.

If a similarly intense degree of motivation is instead directed toward scientific, intellectual, or constructive activity, the increased persistence and endurance may contribute to extraordinary levels of work and achievement.

### Scholium 2

A person who has developed an intense commitment to a purpose before encountering hardship may be capable of enduring considerably greater suffering in pursuit of that purpose than he otherwise would.

---

### Corollary 4: Scientific radicalization can produce a strong internal desire for education.

When the object of radicalization is scientific understanding, the resulting motive may intensify curiosity, study, discipline, experimentation, and the pursuit of knowledge.

### Scholium 1

Moral instructions are often difficult to follow because they require restraint, discipline, sacrifice, or obedience to principles that may conflict with immediate desires.

An intense commitment to those principles may increase an individual's willingness to endure the discomfort associated with acting according to them.

---

### Corollary 5: Radicalization can increase adherence to moral instructions.

Where the object of radicalization is a moral system, the increased motivational force may strengthen an individual's willingness to obey and persist in accordance with its instructions.

This consequence depends upon the moral system toward which the commitment is directed: increased obedience amplifies the practical consequences of the principles being obeyed, whether constructive or destructive.
